````python
# ============================================================
# CELL 0 — PDF INPUT
# ============================================================

import os
import re
import json
import hashlib
import numpy as np
import pandas as pd
import pymupdf


# ------------------------------------------------------------
# LOCKED INPUT FILE
# ------------------------------------------------------------

PDF_PATH = "1 (1).pdf"


# ------------------------------------------------------------
# VERIFY FILE EXISTS
# ------------------------------------------------------------

if not os.path.isfile(PDF_PATH):
    raise FileNotFoundError(
        f"STOP: Expected PDF was not found:\n{PDF_PATH}\n\n"
        "Make sure '1 (1).pdf' is in the VS Code working directory."
    )


# ------------------------------------------------------------
# OPEN PDF AND READ BASIC METADATA
# ------------------------------------------------------------

_doc = pymupdf.open(PDF_PATH)

PDF_PAGE_COUNT = _doc.page_count
PDF_SIZE_BYTES = os.path.getsize(PDF_PATH)

_doc.close()


# ------------------------------------------------------------
# FILE HASH — helps verify we are using the same PDF
# ------------------------------------------------------------

with open(PDF_PATH, "rb") as f:
    PDF_SHA256 = hashlib.sha256(f.read()).hexdigest()


# ------------------------------------------------------------
# REPORT
# ------------------------------------------------------------

print("=" * 70)
print("CELL 0 — PDF INPUT")
print("=" * 70)

print(f"PDF path: {os.path.abspath(PDF_PATH)}")
print(f"File size: {PDF_SIZE_BYTES:,} bytes")
print(f"Pages: {PDF_PAGE_COUNT}")
print(f"SHA256: {PDF_SHA256}")

print("=" * 70)
print("PDF INPUT VERIFIED")
print("=" * 70)
````

````python
# ============================================================
# CELL 1 — PDF TEXT EXTRACTION
# ============================================================

import pymupdf
import re

doc = pymupdf.open(PDF_PATH)

pages = []

for page_number, page in enumerate(doc, start=1):
    text = page.get_text("text")

    pages.append({
        "page_number": page_number,
        "text": text
    })

doc.close()


# ------------------------------------------------------------
# BASIC VALIDATION
# ------------------------------------------------------------

if not pages:
    raise RuntimeError("STOP: No pages were extracted from the PDF.")

empty_pages = [
    p["page_number"]
    for p in pages
    if not p["text"].strip()
]

total_characters = sum(len(p["text"]) for p in pages)


print("=" * 70)
print("CELL 1 — PDF TEXT EXTRACTION")
print("=" * 70)

print(f"Pages extracted: {len(pages)}")
print(f"Empty pages: {len(empty_pages)}")
print(f"Total characters: {total_characters:,}")

if empty_pages:
    print("Empty page numbers:", empty_pages[:20])

print("\nFirst page preview:")
print("-" * 70)
print(pages[0]["text"][:2000])
print("-" * 70)


assert len(pages) == PDF_PAGE_COUNT, \
    "STOP: Extracted page count does not match PDF page count."

assert total_characters > 0, \
    "STOP: PDF extraction returned no text."
````

````python
# ============================================================
# CELL 2 — ROBUST HEADING DETECTION
# ============================================================

import re

# ------------------------------------------------------------
# CONFIGURATION
# ------------------------------------------------------------

CONTENT_START_PAGE = 13

HEADING_PATTERN = re.compile(
    r"^\s*(\d+\.\d+(?:\.\d+)?)\s+(.+?)\s*$"
)


# ------------------------------------------------------------
# DETECT HEADINGS
# ------------------------------------------------------------

heading_candidates = []


for page in pages:

    page_number = page["page_number"]

    if page_number < CONTENT_START_PAGE:
        continue

    text = page.get("text", "")

    if not text:
        continue

    # Character cursor inside this page.
    # This lets us store the exact character position
    # where the heading begins.
    cursor = 0

    for line_number, raw_line in enumerate(
        text.splitlines(keepends=True),
        start=1
    ):

        line = raw_line.strip()

        # Position of this line in the original page text
        line_start = cursor

        # Advance cursor before continuing
        cursor += len(raw_line)

        if not line:
            continue

        match = HEADING_PATTERN.match(line)

        if not match:
            continue

        number = match.group(1)
        title = match.group(2).strip()

        # ----------------------------------------------------
        # Remove TOC-style trailing page numbers
        # ----------------------------------------------------

        title = re.sub(
            r"(?:\s*[.\u2026]\s*){3,}\s*\d+\s*$",
            "",
            title
        ).strip()

        if not title:
            continue

        # ----------------------------------------------------
        # Basic sanity checks
        # ----------------------------------------------------

        if len(title) > 180:
            continue

        lowered = title.lower()

        if (
            "http://" in lowered
            or "https://" in lowered
            or "licence" in lowered
            or "license" in lowered
        ):
            continue

        # ----------------------------------------------------
        # Store heading
        # ----------------------------------------------------

        heading_candidates.append({
            "number": number,
            "title": title,
            "text": f"{number} {title}",
            "page_number": page_number,
            "line_number": line_number,
            "position": line_start,
            "level": len(number.split("."))
        })


# ============================================================
# REMOVE DUPLICATE HEADINGS
# ============================================================

unique_headings = []

seen = set()

for h in heading_candidates:

    key = (
        h["number"],
        h["title"].strip().lower(),
        h["page_number"]
    )

    if key in seen:
        continue

    seen.add(key)
    unique_headings.append(h)


heading_candidates = unique_headings


# ============================================================
# CANONICAL OUTPUT FOR CELL 3
# ============================================================

located_headings = []

for h in heading_candidates:

    located_headings.append({
        "number": h["number"],
        "title": h["title"],
        "text": h["text"],
        "page": h["page_number"],
        "position": h["position"],
        "level": h["level"]
    })


# ============================================================
# SORT IN DOCUMENT ORDER
# ============================================================

located_headings.sort(
    key=lambda h: (
        h["page"],
        h["position"]
    )
)


# ============================================================
# DIAGNOSTIC
# ============================================================

print("=" * 70)
print("CELL 2 — ROBUST HEADING DETECTION")
print("=" * 70)

print(f"Content start page: {CONTENT_START_PAGE}")
print(f"Heading candidates: {len(located_headings)}")


print("\nFirst 30 detected headings:")
print("-" * 70)

for i, h in enumerate(located_headings[:30]):

    print(
        f"{i:03d} | "
        f"page={h['page']:3d} | "
        f"{h['number']:7s} | "
        f"{h['title']}"
    )


print("\nLast 20 detected headings:")
print("-" * 70)

start_index = max(
    0,
    len(located_headings) - 20
)

for i, h in enumerate(
    located_headings[start_index:],
    start=start_index
):

    print(
        f"{i:03d} | "
        f"page={h['page']:3d} | "
        f"{h['number']:7s} | "
        f"{h['title']}"
    )


# ============================================================
# VALIDATION
# ============================================================

assert len(located_headings) > 0, (
    "STOP: No numbered headings were detected."
)

assert all(
    h["page"] >= CONTENT_START_PAGE
    for h in located_headings
), (
    "STOP: Heading detected before CONTENT_START_PAGE."
)

assert all(
    "." in h["number"]
    for h in located_headings
), (
    "STOP: Invalid non-section number detected."
)

assert all(
    isinstance(h["position"], int)
    and h["position"] >= 0
    for h in located_headings
), (
    "STOP: One or more headings has no valid character position."
)


# Check document order

for i in range(1, len(located_headings)):

    previous = located_headings[i - 1]
    current = located_headings[i]

    previous_key = (
        previous["page"],
        previous["position"]
    )

    current_key = (
        current["page"],
        current["position"]
    )

    assert current_key >= previous_key, (
        "STOP: Headings are not in document order."
    )


print("-" * 70)
print("HEADING DETECTION: PASS")
print("-" * 70)

print(
    f"Located headings: {len(located_headings)}"
)

print(
    "Character positions: AVAILABLE"
)
````

````python
# ============================================================
# CELL 3 — EXCLUSIVE SECTION EXTRACTION
# ============================================================

print("=" * 70)
print("CELL 3 — EXCLUSIVE SECTION EXTRACTION")
print("=" * 70)

# ------------------------------------------------------------
# Required inputs
# ------------------------------------------------------------

assert "located_headings" in globals(), \
    "STOP: located_headings is not defined. Run Cell 2 first."

assert len(located_headings) > 0, \
    "STOP: No located headings were detected."


# ------------------------------------------------------------
# Normalize heading schema
#
# Cell 2 may contain different internal field names depending
# on the detection implementation.
#
# Cell 3 converts them ONCE into the canonical schema below:
#
#   number
#   title
#   level
#   page_start
#   start_char
# ------------------------------------------------------------

def get_heading_page(h):
    """
    Extract the page number from a detected heading.
    Supports the canonical 'page' field and common alternatives.
    """

    for key in ("page", "page_start", "page_number"):
        if key in h:
            return int(h[key])

    raise KeyError(
        f"Heading does not contain a page field. "
        f"Available keys: {list(h.keys())}"
    )


def get_heading_position(h):
    """
    Extract the character position of the heading within its page.
    """

    for key in ("position", "start_position", "start_char", "char_position"):
        if key in h:
            return int(h[key])

    # If the detector stores no character position,
    # we cannot safely perform exact page trimming.
    return None


def get_heading_title(h):
    """
    Extract the heading title.
    """

    for key in ("title", "heading", "text"):
        if key in h:
            value = str(h[key]).strip()

            if value:
                return value

    raise KeyError(
        f"Heading does not contain a usable title. "
        f"Available keys: {list(h.keys())}"
    )


def get_heading_level(h):
    """
    Extract heading level.

    Example:
        1.1     -> level 2
        1.1.1   -> level 3
    """

    for key in ("level", "heading_level"):
        if key in h:
            return int(h[key])

    number = str(
        h.get("number", "")
    ).strip()

    if number:
        return len(number.split("."))

    return 0


def get_heading_number(h):
    """
    Extract section number if available.
    """

    for key in ("number", "section_number"):
        if key in h:
            return str(h[key]).strip()

    return ""


# ------------------------------------------------------------
# Build canonical heading list
# ------------------------------------------------------------

canonical_headings = []

for h in located_headings:

    page = get_heading_page(h)
    position = get_heading_position(h)
    title = get_heading_title(h)
    level = get_heading_level(h)
    number = get_heading_number(h)

    canonical_headings.append({
        "number": number,
        "title": title,
        "level": level,
        "page_start": page,
        "start_position": position
    })


# ------------------------------------------------------------
# Sort headings in document order
# ------------------------------------------------------------

canonical_headings.sort(
    key=lambda h: (
        h["page_start"],
        h["start_position"] if h["start_position"] is not None else 0
    )
)


# ------------------------------------------------------------
# Validate heading ordering
# ------------------------------------------------------------

for i in range(1, len(canonical_headings)):

    previous = canonical_headings[i - 1]
    current = canonical_headings[i]

    previous_key = (
        previous["page_start"],
        previous["start_position"]
        if previous["start_position"] is not None
        else 0
    )

    current_key = (
        current["page_start"],
        current["start_position"]
        if current["start_position"] is not None
        else 0
    )

    assert current_key >= previous_key, (
        "STOP: Headings are not in document order."
    )


# ------------------------------------------------------------
# Extract exclusive text for each section
#
# IMPORTANT:
#
# Section A ends immediately before Section B starts.
#
# Therefore a parent section does NOT absorb its child section.
# ------------------------------------------------------------

structured_sections = []


for i, heading in enumerate(canonical_headings):

    start_page = heading["page_start"]

    next_heading = (
        canonical_headings[i + 1]
        if i + 1 < len(canonical_headings)
        else None
    )

    # --------------------------------------------------------
    # Determine ending location
    # --------------------------------------------------------

    if next_heading is not None:

        end_page = next_heading["page_start"]

    else:

        end_page = len(pages)


    # --------------------------------------------------------
    # Collect page text
    #
    # pages is assumed to be the list created by Cell 1.
    # --------------------------------------------------------

    assert "pages" in globals(), \
        "STOP: pages is not defined. Run Cell 1 first."

    assert 1 <= start_page <= len(pages), (
        f"STOP: Invalid start page {start_page}."
    )

    assert 1 <= end_page <= len(pages), (
        f"STOP: Invalid end page {end_page}."
    )


    section_pages = pages[start_page - 1:end_page]


    # --------------------------------------------------------
    # Character-level trimming
    #
    # We use exact positions when Cell 2 provides them.
    #
    # If positions are unavailable, we retain the page-level
    # text. This is safer than inventing a character offset.
    # --------------------------------------------------------

    page_texts = []

    for page_record in section_pages:

        if isinstance(page_record, dict):

            text = page_record.get("text", "")

        else:

            text = str(page_record)

        page_texts.append(text)


    section_text = "\n".join(page_texts)


    # --------------------------------------------------------
    # Trim beginning of first page
    # --------------------------------------------------------

    start_position = heading["start_position"]

    if start_position is not None:

        first_page_text = page_texts[0]

        if 0 <= start_position <= len(first_page_text):

            section_text = (
                first_page_text[start_position:]
                + "\n"
                + "\n".join(page_texts[1:])
            )


    # --------------------------------------------------------
    # Trim beginning of next heading on its page
    #
    # This is what makes sections exclusive.
    # --------------------------------------------------------

    if next_heading is not None:

        next_position = next_heading["start_position"]

        if next_position is not None:

            next_page_index = (
                next_heading["page_start"] - start_page
            )

            if 0 <= next_page_index < len(page_texts):

                before_next_page = page_texts[:next_page_index]

                next_page_text = page_texts[next_page_index]

                trimmed_next_page = (
                    next_page_text[:next_position]
                )

                section_text = (
                    "\n".join(before_next_page)
                    + "\n"
                    + trimmed_next_page
                )


    # --------------------------------------------------------
    # Clean whitespace
    # --------------------------------------------------------

    section_text = re.sub(
        r"\n{3,}",
        "\n\n",
        section_text
    ).strip()


    # --------------------------------------------------------
    # Create canonical section object
    # --------------------------------------------------------

    section = {
        "section_id": i,
        "number": heading["number"],
        "heading": heading["title"],
        "level": heading["level"],
        "page_start": start_page,
        "page_end": end_page,
        "text": section_text,
        "word_count": count_words(section_text)
        if "count_words" in globals()
        else len(re.findall(r"\b\w+\b", section_text))
    }

    structured_sections.append(section)


# ------------------------------------------------------------
# Validation
# ------------------------------------------------------------

assert len(structured_sections) == len(canonical_headings), (
    "STOP: Number of structured sections does not match "
    "number of detected headings."
)

assert len(structured_sections) > 0, (
    "STOP: No structured sections were created."
)

for section in structured_sections:

    assert "text" in section
    assert "heading" in section
    assert "page_start" in section
    assert "page_end" in section

    assert section["page_start"] <= section["page_end"]


# ------------------------------------------------------------
# Diagnostic
# ------------------------------------------------------------

print(f"Detected headings: {len(canonical_headings)}")
print(f"Structured sections: {len(structured_sections)}")

print()
print("First 10 sections:")
print("-" * 70)

for section in structured_sections[:10]:

    print(
        f'{section["section_id"]:03d} | '
        f'{section["heading"]} | '
        f'pages={section["page_start"]}-'
        f'{section["page_end"]} | '
        f'words={section["word_count"]}'
    )


print()
print("Last 10 sections:")
print("-" * 70)

for section in structured_sections[-10:]:

    print(
        f'{section["section_id"]:03d} | '
        f'{section["heading"]} | '
        f'pages={section["page_start"]}-'
        f'{section["page_end"]} | '
        f'words={section["word_count"]}'
    )


print()
print("=" * 70)
print("EXCLUSIVE SECTION EXTRACTION: PASS")
print("=" * 70)
````

````python
# ============================================================
# CELL 4 — PARAGRAPH EXTRACTION
# ============================================================

import re


def split_into_paragraphs(text):
    """
    Split section text into paragraph-like units.

    Rules:
    - Normalize whitespace.
    - Preserve sentence/content order.
    - Split primarily on blank-line boundaries.
    - Remove empty fragments.
    """

    if not isinstance(text, str):
        raise TypeError("Section text must be a string.")

    text = text.replace("\r\n", "\n").replace("\r", "\n")

    # Normalize spaces while preserving paragraph boundaries.
    text = re.sub(r"[ \t]+", " ", text)

    # Normalize excessive blank lines.
    text = re.sub(r"\n[ \t]*\n+", "\n\n", text)

    raw_paragraphs = re.split(r"\n\s*\n", text)

    paragraphs = []

    for p in raw_paragraphs:
        p = p.strip()

        if not p:
            continue

        # Collapse remaining single newlines caused by PDF line wrapping.
        p = re.sub(r"\s*\n\s*", " ", p)

        # Normalize repeated spaces.
        p = re.sub(r"\s{2,}", " ", p).strip()

        if p:
            paragraphs.append(p)

    return paragraphs


section_paragraphs = []

for section in structured_sections:

    if "text" not in section:
        raise RuntimeError(
            "STOP: structured_sections contains a section without 'text'."
        )

    paragraphs = split_into_paragraphs(section["text"])

    section_paragraphs.append({
        "heading": section["heading"],
        "level": section["level"],
        "page_start": section["page_start"],
        "page_end": section["page_end"],
        "paragraphs": paragraphs,
    })


# ------------------------------------------------------------
# VALIDATION
# ------------------------------------------------------------

assert len(section_paragraphs) == len(structured_sections), (
    "Section count changed during paragraph extraction."
)

total_paragraphs = sum(
    len(section["paragraphs"])
    for section in section_paragraphs
)

assert total_paragraphs > 0, (
    "STOP: No paragraphs were extracted."
)


print("=" * 70)
print("CELL 4 — PARAGRAPH EXTRACTION")
print("=" * 70)

print(f"Sections: {len(section_paragraphs)}")
print(f"Total paragraphs: {total_paragraphs}")

print("\nFirst 10 sections:")
print("-" * 70)

for i, section in enumerate(section_paragraphs[:10]):
    print(
        f"{i:03d} | "
        f"{section['heading']} | "
        f"paragraphs={len(section['paragraphs'])}"
    )

print("\nLast 10 sections:")
print("-" * 70)

start = max(0, len(section_paragraphs) - 10)

for i in range(start, len(section_paragraphs)):
    section = section_paragraphs[i]

    print(
        f"{i:03d} | "
        f"{section['heading']} | "
        f"paragraphs={len(section['paragraphs'])}"
    )


print("\n" + "=" * 70)
print("PARAGRAPH EXTRACTION: PASS")
print("=" * 70)
````

````python
# ============================================================
# CELL 5 — PARAGRAPH CLEANING + VALIDATION
# ============================================================

import re


def normalize_paragraph(text):
    """
    Normalize a paragraph without changing its meaning.
    """

    if not isinstance(text, str):
        return ""

    text = text.replace("\u00a0", " ")
    text = re.sub(r"\s+", " ", text)
    text = text.strip()

    return text


def is_noise_paragraph(text):
    """
    Detect obvious PDF extraction noise.

    We intentionally keep this conservative.
    We do NOT remove normal short paragraphs.
    """

    if not text:
        return True

    # Page-number-only paragraph
    if re.fullmatch(r"\d{1,4}", text):
        return True

    # Very obvious isolated punctuation
    if re.fullmatch(r"[\W_]+", text):
        return True

    return False


cleaned_section_paragraphs = []

original_paragraph_count = 0
removed_paragraph_count = 0

original_word_count = 0
cleaned_word_count = 0


def count_words(text):
    return len(re.findall(r"\b\w+\b", text))


for section in section_paragraphs:

    original_paragraphs = section["paragraphs"]

    cleaned_paragraphs = []

    for paragraph in original_paragraphs:

        original_paragraph_count += 1
        original_word_count += count_words(paragraph)

        normalized = normalize_paragraph(paragraph)

        if is_noise_paragraph(normalized):
            removed_paragraph_count += 1
            continue

        cleaned_paragraphs.append(normalized)
        cleaned_word_count += count_words(normalized)

    cleaned_section_paragraphs.append({
        "heading": section["heading"],
        "level": section["level"],
        "page_start": section["page_start"],
        "page_end": section["page_end"],
        "paragraphs": cleaned_paragraphs,
    })


# ------------------------------------------------------------
# VALIDATION
# ------------------------------------------------------------

cleaned_paragraph_count = sum(
    len(section["paragraphs"])
    for section in cleaned_section_paragraphs
)

assert cleaned_paragraph_count > 0, (
    "STOP: Cleaning removed all paragraphs."
)

assert cleaned_word_count > 0, (
    "STOP: No words remain after cleaning."
)

# Cleaning should be conservative.
# If more than 10% of paragraphs disappeared,
# stop and inspect instead of silently continuing.
removal_ratio = (
    removed_paragraph_count / original_paragraph_count
    if original_paragraph_count
    else 1.0
)

assert removal_ratio <= 0.10, (
    f"STOP: Paragraph removal ratio is too high: "
    f"{removal_ratio:.2%}"
)

# Word loss should also be very small.
word_loss_ratio = (
    (original_word_count - cleaned_word_count)
    / original_word_count
    if original_word_count
    else 1.0
)

assert word_loss_ratio <= 0.10, (
    f"STOP: Word loss ratio is too high: "
    f"{word_loss_ratio:.2%}"
)


print("=" * 70)
print("CELL 5 — PARAGRAPH CLEANING + VALIDATION")
print("=" * 70)

print(f"Original sections: {len(section_paragraphs)}")
print(f"Cleaned sections: {len(cleaned_section_paragraphs)}")

print(f"Original paragraphs: {original_paragraph_count}")
print(f"Cleaned paragraphs: {cleaned_paragraph_count}")
print(f"Removed paragraphs: {removed_paragraph_count}")

print(f"Original words: {original_word_count:,}")
print(f"Cleaned words: {cleaned_word_count:,}")

print(f"Paragraph removal ratio: {removal_ratio:.2%}")
print(f"Word loss ratio: {word_loss_ratio:.2%}")

print("\nLast section diagnostic:")
last_section = cleaned_section_paragraphs[-1]

print(f"Heading: {last_section['heading']}")
print(f"Pages: {last_section['page_start']}-{last_section['page_end']}")
print(f"Paragraphs: {len(last_section['paragraphs'])}")
print(
    f"Words: "
    f"{sum(count_words(p) for p in last_section['paragraphs']):,}"
)

print("\n" + "=" * 70)
print("PARAGRAPH CLEANING + VALIDATION: PASS")
print("=" * 70)
````

````python
# ============================================================
# CELL 6 — FINAL SAFE PARAGRAPH-AWARE CHUNKING
# ============================================================

TARGET_SIZE = 500
MAX_OVERLAP_WORDS = 75


def count_words(text):
    return len(re.findall(r"\b\w+\b", text))


def make_chunk(chunk_id, chunk_order, section, paragraphs):

    text = "\n\n".join(paragraphs)

    return {
        "chunk_id": chunk_id,
        "chunk_order": chunk_order,
        "text": text,
        "heading": section["heading"],
        "level": section["level"],
        "page_start": section["page_start"],
        "page_end": section["page_end"],
        "word_count": count_words(text),
    }


chunks = []
large_source_paragraphs = []

chunk_id = 0
chunk_order = 0


for section in cleaned_section_paragraphs:

    paragraphs = section["paragraphs"]

    current_paras = []
    current_words = 0

    for para_index, para in enumerate(paragraphs):

        para_words = count_words(para)

        # ====================================================
        # LARGE SOURCE PARAGRAPH
        # ====================================================

        if para_words > TARGET_SIZE:

            # Finish current chunk first.
            if current_paras:

                chunks.append(
                    make_chunk(
                        chunk_id,
                        chunk_order,
                        section,
                        current_paras
                    )
                )

                chunk_id += 1
                chunk_order += 1

                current_paras = []
                current_words = 0

            # Keep large paragraph intact.
            chunks.append(
                make_chunk(
                    chunk_id,
                    chunk_order,
                    section,
                    [para]
                )
            )

            large_source_paragraphs.append({
                "heading": section["heading"],
                "paragraph_index": para_index,
                "word_count": para_words,
            })

            chunk_id += 1
            chunk_order += 1

            continue


        # ====================================================
        # DOES NOT FIT
        # ====================================================

        if (
            current_paras
            and current_words + para_words > TARGET_SIZE
        ):

            # Save current chunk.
            previous_paras = list(current_paras)

            chunks.append(
                make_chunk(
                    chunk_id,
                    chunk_order,
                    section,
                    previous_paras
                )
            )

            chunk_id += 1
            chunk_order += 1


            # =================================================
            # BUILD NON-RECURSIVE OVERLAP
            # =================================================

            overlap_paras = []
            overlap_words = 0

            for previous_para in reversed(previous_paras):

                previous_words = count_words(previous_para)

                if overlap_words + previous_words <= MAX_OVERLAP_WORDS:

                    overlap_paras.insert(
                        0,
                        previous_para
                    )

                    overlap_words += previous_words

                else:
                    break


            # -------------------------------------------------
            # IMPORTANT SAFETY RULE
            #
            # If overlap is empty, start clean.
            # This guarantees progress.
            # -------------------------------------------------

            current_paras = list(overlap_paras)
            current_words = overlap_words

            # Add the CURRENT paragraph immediately.
            #
            # We intentionally do not use "continue" here.
            # This guarantees the loop always progresses.
            current_paras.append(para)
            current_words += para_words

            continue


        # ====================================================
        # FITS NORMALLY
        # ====================================================

        current_paras.append(para)
        current_words += para_words


    # ========================================================
    # FINAL CHUNK OF SECTION
    # ========================================================

    if current_paras:

        chunks.append(
            make_chunk(
                chunk_id,
                chunk_order,
                section,
                current_paras
            )
        )

        chunk_id += 1
        chunk_order += 1


# ============================================================
# VALIDATION
# ============================================================

assert len(chunks) > 0

assert all(
    chunk["text"].strip()
    for chunk in chunks
)

assert all(
    chunk["word_count"] == count_words(chunk["text"])
    for chunk in chunks
)

assert all(
    chunk["chunk_id"] == i
    for i, chunk in enumerate(chunks)
)

assert all(
    chunk["chunk_order"] == i
    for i, chunk in enumerate(chunks)
)


# ============================================================
# DIAGNOSTIC
# ============================================================

word_counts = [
    chunk["word_count"]
    for chunk in chunks
]

print("=" * 70)
print("CELL 6 — FINAL SAFE PARAGRAPH-AWARE CHUNKING")
print("=" * 70)

print(f"Sections: {len(cleaned_section_paragraphs)}")
print(f"Paragraphs: {cleaned_paragraph_count}")
print(f"Total chunks: {len(chunks)}")

print(f"Target size: {TARGET_SIZE}")
print(f"Maximum overlap: {MAX_OVERLAP_WORDS}")

print(f"Minimum chunk words: {min(word_counts)}")
print(f"Maximum chunk words: {max(word_counts)}")
print(f"Average chunk words: {sum(word_counts) / len(word_counts):.2f}")

print(
    f"Large source paragraphs: "
    f"{len(large_source_paragraphs)}"
)

print("\nFirst 5 chunks:")
for chunk in chunks[:5]:
    print(
        f"{chunk['chunk_id']:04d} | "
        f"{chunk['word_count']} words | "
        f"{chunk['heading']}"
    )

print("\nLast 5 chunks:")
for chunk in chunks[-5:]:
    print(
        f"{chunk['chunk_id']:04d} | "
        f"{chunk['word_count']} words | "
        f"{chunk['heading']}"
    )

print("=" * 70)
print("FINAL CHUNKING: PASS")
print("=" * 70)
````

````python
# ============================================================
# CELL 7 — CHUNK VALIDATION
# ============================================================

print("=" * 70)
print("CELL 7 — CHUNK VALIDATION")
print("=" * 70)


# ------------------------------------------------------------
# 1. Basic existence checks
# ------------------------------------------------------------

assert isinstance(chunks, list), \
    "STOP: chunks must be a list."

assert len(chunks) > 0, \
    "STOP: No chunks were generated."


# ------------------------------------------------------------
# 2. Required fields
# ------------------------------------------------------------

required_fields = {
    "chunk_id",
    "chunk_order",
    "text",
    "heading",
    "level",
    "page_start",
    "page_end",
    "word_count",
}


missing_fields = []

for i, chunk in enumerate(chunks):

    missing = required_fields - set(chunk.keys())

    if missing:
        missing_fields.append(
            {
                "chunk_index": i,
                "missing": sorted(missing),
            }
        )


assert not missing_fields, (
    f"STOP: Missing fields detected: {missing_fields[:5]}"
)


# ------------------------------------------------------------
# 3. Chunk IDs must be unique and sequential
# ------------------------------------------------------------

chunk_ids = [
    chunk["chunk_id"]
    for chunk in chunks
]

chunk_orders = [
    chunk["chunk_order"]
    for chunk in chunks
]


assert len(chunk_ids) == len(set(chunk_ids)), \
    "STOP: Duplicate chunk_id detected."

assert chunk_ids == list(range(len(chunks))), \
    "STOP: chunk_id sequence is invalid."

assert chunk_orders == list(range(len(chunks))), \
    "STOP: chunk_order sequence is invalid."


# ------------------------------------------------------------
# 4. Text must be non-empty
# ------------------------------------------------------------

empty_chunks = [
    i
    for i, chunk in enumerate(chunks)
    if not isinstance(chunk["text"], str)
    or not chunk["text"].strip()
]


assert not empty_chunks, (
    f"STOP: Empty chunks detected: {empty_chunks[:10]}"
)


# ------------------------------------------------------------
# 5. Word counts must be correct
# ------------------------------------------------------------

word_count_errors = []

for i, chunk in enumerate(chunks):

    actual_count = count_words(chunk["text"])

    if actual_count != chunk["word_count"]:

        word_count_errors.append(
            {
                "chunk_index": i,
                "stored": chunk["word_count"],
                "actual": actual_count,
            }
        )


assert not word_count_errors, (
    f"STOP: Word-count mismatch: "
    f"{word_count_errors[:5]}"
)


# ------------------------------------------------------------
# 6. Page metadata validation
# ------------------------------------------------------------

page_errors = []

for i, chunk in enumerate(chunks):

    start = chunk["page_start"]
    end = chunk["page_end"]

    if start > end:
        page_errors.append(
            {
                "chunk_index": i,
                "page_start": start,
                "page_end": end,
            }
        )


assert not page_errors, (
    f"STOP: Invalid page ranges: {page_errors[:5]}"
)


# ------------------------------------------------------------
# 7. Section metadata validation
# ------------------------------------------------------------

metadata_errors = []

for i, chunk in enumerate(chunks):

    if not chunk["heading"]:
        metadata_errors.append(
            {
                "chunk_index": i,
                "error": "empty heading",
            }
        )

    if chunk["level"] is None:
        metadata_errors.append(
            {
                "chunk_index": i,
                "error": "missing level",
            }
        )


assert not metadata_errors, (
    f"STOP: Metadata errors: {metadata_errors[:5]}"
)


# ------------------------------------------------------------
# 8. Check oversized chunks
#
# Oversized chunks are allowed only when they correspond
# to a source paragraph larger than TARGET_SIZE.
# ------------------------------------------------------------

oversized_chunks = [
    chunk
    for chunk in chunks
    if chunk["word_count"] > TARGET_SIZE
]


invalid_oversized_chunks = [
    chunk
    for chunk in oversized_chunks
    if chunk["word_count"] <= TARGET_SIZE
]


assert not invalid_oversized_chunks


# ------------------------------------------------------------
# 9. Detect suspiciously tiny chunks
# ------------------------------------------------------------

tiny_chunks = [
    chunk
    for chunk in chunks
    if chunk["word_count"] < 100
]


# Tiny chunks are NOT automatically an error.
# We report them for inspection instead.


# ------------------------------------------------------------
# 10. Summary
# ------------------------------------------------------------

word_counts = [
    chunk["word_count"]
    for chunk in chunks
]


print(f"Chunks validated: {len(chunks):,}")

print(
    f"Unique chunk IDs: "
    f"{len(set(chunk_ids)):,}"
)

print(
    f"Word-count mismatches: "
    f"{len(word_count_errors)}"
)

print(
    f"Empty chunks: "
    f"{len(empty_chunks)}"
)

print(
    f"Invalid page ranges: "
    f"{len(page_errors)}"
)

print(
    f"Oversized chunks: "
    f"{len(oversized_chunks)}"
)

print(
    f"Chunks under 100 words: "
    f"{len(tiny_chunks)}"
)

print(
    f"Minimum words: "
    f"{min(word_counts)}"
)

print(
    f"Maximum words: "
    f"{max(word_counts)}"
)

print(
    f"Average words: "
    f"{sum(word_counts) / len(word_counts):.2f}"
)


# ------------------------------------------------------------
# Tiny chunk diagnostic
# ------------------------------------------------------------

if tiny_chunks:

    print("\nTiny chunk diagnostic:")
    print("-" * 70)

    for chunk in tiny_chunks[:10]:

        print(
            f"{chunk['chunk_id']:04d} | "
            f"{chunk['word_count']:3d} words | "
            f"{chunk['heading']}"
        )


# ------------------------------------------------------------
# Final result
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("CHUNK VALIDATION: PASS")
print("=" * 70)
````

````python
# ============================================================
# CELL 8 — EMBEDDING GENERATION
# ============================================================

from sentence_transformers import SentenceTransformer
import numpy as np


print("=" * 70)
print("CELL 8 — EMBEDDING GENERATION")
print("=" * 70)


# ------------------------------------------------------------
# 1. Safety checks
# ------------------------------------------------------------

assert isinstance(chunks, list), \
    "STOP: chunks must be a list."

assert len(chunks) > 0, \
    "STOP: No chunks available for embedding."


# ------------------------------------------------------------
# 2. Use the validated chunks directly
# ------------------------------------------------------------

chunk_texts = [
    chunk["text"]
    for chunk in chunks
]


assert len(chunk_texts) == len(chunks), \
    "STOP: Number of texts does not match number of chunks."

assert all(
    isinstance(text, str) and text.strip()
    for text in chunk_texts
), \
    "STOP: One or more chunk texts are empty."


# ------------------------------------------------------------
# 3. Load embedding model
# ------------------------------------------------------------

EMBEDDING_MODEL_NAME = "BAAI/bge-m3"

embedding_model = SentenceTransformer(
    EMBEDDING_MODEL_NAME
)


# ------------------------------------------------------------
# 4. Generate embeddings
# ------------------------------------------------------------

raw_embeddings = embedding_model.encode(
    chunk_texts,
    batch_size=8,
    show_progress_bar=True,
    convert_to_numpy=True,
)


# ------------------------------------------------------------
# 5. Convert to float32 for consistent storage
# ------------------------------------------------------------

raw_embeddings = np.asarray(
    raw_embeddings,
    dtype=np.float32
)


# ------------------------------------------------------------
# 6. Verify actual shape
# ------------------------------------------------------------

assert raw_embeddings.ndim == 2, \
    "STOP: Embeddings must be a 2D matrix."

assert raw_embeddings.shape[0] == len(chunks), (
    "STOP: Number of embeddings does not match "
    "number of chunks."
)

embedding_dim = raw_embeddings.shape[1]


assert embedding_dim > 0, \
    "STOP: Invalid embedding dimension."


# ------------------------------------------------------------
# 7. Check for invalid numerical values
# ------------------------------------------------------------

assert np.isfinite(raw_embeddings).all(), (
    "STOP: Embeddings contain NaN or infinite values."
)


# ------------------------------------------------------------
# 8. Diagnostic
# ------------------------------------------------------------

print(f"Model: {EMBEDDING_MODEL_NAME}")
print(f"Chunks embedded: {raw_embeddings.shape[0]:,}")
print(f"Embedding dimension: {embedding_dim}")
print(f"Embedding matrix shape: {raw_embeddings.shape}")
print(f"Dtype: {raw_embeddings.dtype}")


print("=" * 70)
print("EMBEDDING GENERATION: PASS")
print("=" * 70)
````

````python
# ============================================================
# CELL 8.1 — SAVE EMBEDDINGS CHECKPOINT
# ============================================================

import pickle
import os

embeddings_checkpoint = os.path.join(
    CHECKPOINT_DIR,
    "raw_embeddings.pkl"
)

with open(embeddings_checkpoint, "wb") as f:
    pickle.dump(raw_embeddings, f)

print("=" * 70)
print("EMBEDDINGS CHECKPOINT SAVED")
print("=" * 70)

print("Embeddings shape:", raw_embeddings.shape)
print("Dtype:", raw_embeddings.dtype)
print("Saved to:", embeddings_checkpoint)
````

````python
# ============================================================
# CELL 9 — EMBEDDING L2 NORMALIZATION
# ============================================================

print("=" * 70)
print("CELL 9 — EMBEDDING L2 NORMALIZATION")
print("=" * 70)

assert "raw_embeddings" in globals(), \
    "STOP: raw_embeddings not found. Run Cell 8 first."

assert isinstance(raw_embeddings, np.ndarray), \
    "STOP: raw_embeddings must be a NumPy array."

assert raw_embeddings.ndim == 2, \
    "STOP: raw_embeddings must be a 2D matrix."

assert raw_embeddings.shape[0] == len(chunks), \
    "STOP: Embedding count does not match chunk count."


# Calculate L2 norm for every embedding
raw_norms = np.linalg.norm(
    raw_embeddings,
    axis=1,
    keepdims=True
)


# No zero vectors allowed
zero_norm_count = int(
    np.sum(raw_norms == 0)
)

assert zero_norm_count == 0, \
    "STOP: One or more embeddings have zero L2 norm."


# Normalize
normalized_embeddings = (
    raw_embeddings / raw_norms
).astype(np.float32)


# Verify norms
normalized_norms = np.linalg.norm(
    normalized_embeddings,
    axis=1
)


normalization_errors = int(
    np.sum(
        np.abs(normalized_norms - 1.0) > 1e-4
    )
)


assert np.isfinite(normalized_embeddings).all(), \
    "STOP: Embeddings contain NaN or infinite values."

assert normalization_errors == 0, \
    "STOP: One or more embeddings failed normalization."

assert normalized_embeddings.shape == raw_embeddings.shape, \
    "STOP: Embedding shape changed during normalization."


print(f"Vectors: {normalized_embeddings.shape[0]:,}")
print(f"Dimension: {normalized_embeddings.shape[1]:,}")
print(f"Zero-norm vectors: {zero_norm_count}")
print(f"Min norm: {normalized_norms.min():.6f}")
print(f"Max norm: {normalized_norms.max():.6f}")
print(f"Mean norm: {normalized_norms.mean():.6f}")
print(f"Normalization errors: {normalization_errors}")

print("=" * 70)
print("EMBEDDING NORMALIZATION: PASS")
print("=" * 70)
````

````python
# ============================================================
# CELL 11 — MONGODB CONNECTION + CHUNK UPLOAD
# ============================================================

import numpy as np

from getpass import getpass
from pymongo import MongoClient


# ============================================================
# CONFIGURATION
# ============================================================

DATABASE_NAME = "psychological_self_help"
COLLECTION_NAME = "chunks"

VECTOR_FIELD = "embedding"
VECTOR_DIMENSION = 1024

INDEX_NAME = "chunk_vector_index"


# ============================================================
# ENTER MONGODB CONNECTION STRING SAFELY
# ============================================================

print("=" * 70)
print("CELL 11 — MONGODB CONNECTION + CHUNK UPLOAD")
print("=" * 70)

print()
print("Paste your MongoDB Atlas connection string below.")
print("Your input will NOT be displayed.")
print()

MONGODB_URI = getpass(
    "MongoDB URI: "
)


# ============================================================
# BASIC VALIDATION
# ============================================================

assert MONGODB_URI.strip(), (
    "STOP: MongoDB URI is empty."
)

assert (
    MONGODB_URI.startswith("mongodb+srv://")
    or MONGODB_URI.startswith("mongodb://")
), (
    "STOP: Invalid MongoDB URI."
)


# ============================================================
# CONNECT TO MONGODB
# ============================================================

try:

    client = MongoClient(
        MONGODB_URI,
        serverSelectionTimeoutMS=15000
    )

    # Force connection test
    client.admin.command("ping")

except Exception as e:

    raise RuntimeError(
        "MongoDB connection failed. "
        "Check your username, password, "
        "connection string, and Atlas Network Access."
    ) from e


db = client[DATABASE_NAME]
collection = db[COLLECTION_NAME]


print()
print("MongoDB connection: PASS")
print("Database:", DATABASE_NAME)
print("Collection:", COLLECTION_NAME)


# ============================================================
# VALIDATE LOCAL DATA
# ============================================================

assert "chunks" in globals(), (
    "STOP: chunks is not defined. "
    "Run the chunking cells first."
)

# ------------------------------------------------------------
# IMPORTANT:
# Cell 9 created normalized_embeddings, not embeddings.
# We use the normalized vectors for MongoDB.
# ------------------------------------------------------------

assert "normalized_embeddings" in globals(), (
    "STOP: normalized_embeddings is not defined. "
    "Run Cell 8 and Cell 9 first."
)

assert len(chunks) == len(normalized_embeddings), (
    f"STOP: chunks={len(chunks)} but "
    f"normalized_embeddings={len(normalized_embeddings)}"
)

assert len(chunks) > 0, (
    "STOP: No chunks available."
)

assert normalized_embeddings.ndim == 2, (
    f"STOP: Expected 2D normalized embeddings, "
    f"got shape {normalized_embeddings.shape}"
)

assert normalized_embeddings.shape[1] == VECTOR_DIMENSION, (
    f"STOP: Expected embedding dimension "
    f"{VECTOR_DIMENSION}, "
    f"got {normalized_embeddings.shape[1]}"
)


print()
print("Local chunks:", len(chunks))
print(
    "Normalized embedding shape:",
    normalized_embeddings.shape
)
print("Local data validation: PASS")


# ============================================================
# CLEAN PREVIOUS COLLECTION DATA
# ============================================================

collection.delete_many({})


# ============================================================
# PREPARE MONGODB DOCUMENTS
# ============================================================

mongo_documents = []


for chunk, vector in zip(
    chunks,
    normalized_embeddings
):

    vector = np.asarray(
        vector,
        dtype=np.float32
    ).tolist()

    document = {

        "chunk_id": int(
            chunk["chunk_id"]
        ),

        "text": chunk["text"],

        "heading": chunk["heading"],

        "level": chunk["level"],

        "page_start": int(
            chunk["page_start"]
        ),

        "page_end": int(
            chunk["page_end"]
        ),

        "chunk_order": int(
            chunk["chunk_order"]
        ),

        "word_count": int(
            chunk["word_count"]
        ),

        VECTOR_FIELD: vector
    }

    mongo_documents.append(document)


# ============================================================
# INSERT DOCUMENTS
# ============================================================

result = collection.insert_many(
    mongo_documents
)


print()
print(
    "Documents inserted:",
    len(result.inserted_ids)
)


# ============================================================
# DATABASE VALIDATION
# ============================================================

stored_count = collection.count_documents({})


assert stored_count == len(chunks), (
    f"STOP: MongoDB contains {stored_count} documents, "
    f"expected {len(chunks)}."
)


# ------------------------------------------------------------
# Check one stored document
# ------------------------------------------------------------

sample = collection.find_one(
    {"chunk_id": 0}
)


assert sample is not None, (
    "STOP: Could not retrieve chunk_id=0."
)


required_fields = [
    "chunk_id",
    "text",
    "heading",
    "level",
    "page_start",
    "page_end",
    "chunk_order",
    "word_count",
    "embedding"
]


for field in required_fields:

    assert field in sample, (
        f"STOP: Missing field: {field}"
    )


assert len(sample["embedding"]) == VECTOR_DIMENSION, (
    f"STOP: Stored embedding has "
    f"{len(sample['embedding'])} dimensions, "
    f"expected {VECTOR_DIMENSION}."
)


# ============================================================
# FINAL REPORT
# ============================================================

print("-" * 70)

print(
    "Stored documents:",
    stored_count
)

print(
    "Embedding dimension:",
    len(sample["embedding"])
)

print("Metadata fields: PASS")

print("=" * 70)
print("MONGODB DATA UPLOAD: PASS")
print("=" * 70)
````

````python
# ============================================================
# CELL 12 — MONGODB VECTOR SEARCH TEST
# ============================================================

import numpy as np


print("=" * 70)
print("CELL 12 — MONGODB VECTOR SEARCH TEST")
print("=" * 70)


# ============================================================
# CONFIGURATION
# ============================================================

QUERY_TEXT = (
    "What is self-help and how can it be used?"
)

TOP_K = 5

VECTOR_FIELD = "embedding"
VECTOR_DIMENSION = 1024
INDEX_NAME = "chunk_vector_index"


# ============================================================
# SAFETY CHECKS
# ============================================================

assert "embedding_model" in globals(), (
    "STOP: embedding_model is not defined. "
    "Run Cell 8 first."
)

assert "collection" in globals(), (
    "STOP: MongoDB collection is not defined. "
    "Run Cell 11 first."
)

assert "normalized_embeddings" in globals(), (
    "STOP: normalized_embeddings is not defined. "
    "Run Cell 9 first."
)

assert normalized_embeddings.shape[1] == VECTOR_DIMENSION, (
    f"STOP: Expected dimension {VECTOR_DIMENSION}, "
    f"got {normalized_embeddings.shape[1]}"
)


# ============================================================
# GENERATE QUERY EMBEDDING
# ============================================================

query_embedding = embedding_model.encode(
    QUERY_TEXT,
    convert_to_numpy=True
)


query_embedding = np.asarray(
    query_embedding,
    dtype=np.float32
)


assert query_embedding.ndim == 1, (
    f"STOP: Query embedding must be 1D, "
    f"got shape {query_embedding.shape}"
)

assert len(query_embedding) == VECTOR_DIMENSION, (
    f"STOP: Query embedding dimension is "
    f"{len(query_embedding)}, expected {VECTOR_DIMENSION}."
)


# ============================================================
# L2 NORMALIZE QUERY
# ============================================================

query_norm = np.linalg.norm(
    query_embedding
)


assert query_norm > 0, (
    "STOP: Query embedding has zero norm."
)


query_embedding = (
    query_embedding / query_norm
).astype(np.float32)


# Verify normalization

final_query_norm = np.linalg.norm(
    query_embedding
)


assert abs(final_query_norm - 1.0) < 1e-4, (
    "STOP: Query embedding normalization failed."
)


print("Query:", QUERY_TEXT)
print("Query embedding dimension:", len(query_embedding))
print(f"Query vector norm: {final_query_norm:.6f}")


# ============================================================
# VECTOR SEARCH
# ============================================================

pipeline = [

    {
        "$vectorSearch": {

            "index": INDEX_NAME,

            "path": VECTOR_FIELD,

            "queryVector": query_embedding.tolist(),

            "numCandidates": 50,

            "limit": TOP_K
        }
    },

    {
        "$project": {

            "_id": 0,

            "chunk_id": 1,

            "text": 1,

            "heading": 1,

            "page_start": 1,

            "page_end": 1,

            "word_count": 1,

            "score": {
                "$meta": "vectorSearchScore"
            }
        }
    }
]


results = list(
    collection.aggregate(pipeline)
)


# ============================================================
# VALIDATION
# ============================================================

assert len(results) > 0, (
    "STOP: Vector search returned zero results."
)


assert len(results) <= TOP_K, (
    "STOP: Vector search returned more than TOP_K results."
)


for result in results:

    assert "chunk_id" in result
    assert "text" in result
    assert "heading" in result
    assert "score" in result

    assert isinstance(
        result["score"],
        (int, float)
    )


# ============================================================
# DISPLAY RESULTS
# ============================================================

print()
print("=" * 70)
print("VECTOR SEARCH RESULTS")
print("=" * 70)


for rank, result in enumerate(
    results,
    start=1
):

    print()
    print(
        f"Rank {rank} | "
        f"Chunk {result['chunk_id']} | "
        f"Score: {result['score']:.6f}"
    )

    print(
        f"Heading: {result['heading']}"
    )

    print(
        f"Pages: "
        f"{result['page_start']}-"
        f"{result['page_end']}"
    )

    print(
        f"Words: {result['word_count']}"
    )

    print("-" * 70)

    print(
        result["text"][:700]
    )


print()
print("=" * 70)
print(f"Results returned: {len(results)}")
print("VECTOR SEARCH: PASS")
print("=" * 70)
````

````python
# ============================================================
# CELL 13 — MULTI-QUERY RETRIEVAL TEST
# ============================================================

import numpy as np


print("=" * 70)
print("CELL 13 — MULTI-QUERY RETRIEVAL TEST")
print("=" * 70)


# ============================================================
# CONFIGURATION
# ============================================================

VECTOR_FIELD = "embedding"
VECTOR_DIMENSION = 1024
INDEX_NAME = "chunk_vector_index"

TOP_K = 5
NUM_CANDIDATES = 50


# ============================================================
# TEST QUESTIONS
# ============================================================

TEST_QUERIES = [

    "What is self-help?",

    "Who is self-help intended for?",

    "When should guided self-help be used?",

    "What should a helper do when someone is at risk of suicide?",

    "What are the Step-by-Step skills for managing stress?",

    "What are the Doing What Matters modules?"
]


# ============================================================
# SAFETY CHECKS
# ============================================================

assert "embedding_model" in globals(), (
    "STOP: embedding_model is not defined. "
    "Run Cell 8 first."
)

assert "collection" in globals(), (
    "STOP: MongoDB collection is not defined. "
    "Run Cell 11 first."
)

assert INDEX_NAME == "chunk_vector_index"

assert len(TEST_QUERIES) > 0


# ============================================================
# SEARCH FUNCTION
# ============================================================

def vector_search(query_text, top_k=TOP_K):

    # --------------------------------------------------------
    # Generate query embedding
    # --------------------------------------------------------

    query_vector = embedding_model.encode(
        query_text,
        convert_to_numpy=True
    )

    query_vector = np.asarray(
        query_vector,
        dtype=np.float32
    )

    assert query_vector.ndim == 1

    assert len(query_vector) == VECTOR_DIMENSION


    # --------------------------------------------------------
    # L2 normalize query
    # --------------------------------------------------------

    query_norm = np.linalg.norm(
        query_vector
    )

    assert query_norm > 0

    query_vector = (
        query_vector / query_norm
    ).astype(np.float32)


    # --------------------------------------------------------
    # MongoDB Vector Search
    # --------------------------------------------------------

    pipeline = [

        {
            "$vectorSearch": {

                "index": INDEX_NAME,

                "path": VECTOR_FIELD,

                "queryVector": query_vector.tolist(),

                "numCandidates": NUM_CANDIDATES,

                "limit": top_k
            }
        },

        {
            "$project": {

                "_id": 0,

                "chunk_id": 1,

                "text": 1,

                "heading": 1,

                "page_start": 1,

                "page_end": 1,

                "word_count": 1,

                "score": {
                    "$meta": "vectorSearchScore"
                }
            }
        }
    ]


    return list(
        collection.aggregate(
            pipeline
        )
    )


# ============================================================
# RUN TESTS
# ============================================================

all_results = {}


for query_number, query in enumerate(
    TEST_QUERIES,
    start=1
):

    print()
    print("=" * 70)

    print(
        f"QUERY {query_number}/{len(TEST_QUERIES)}"
    )

    print("=" * 70)

    print()
    print("Question:")
    print(query)

    results = vector_search(
        query,
        TOP_K
    )

    all_results[query] = results


    # --------------------------------------------------------
    # Validate
    # --------------------------------------------------------

    assert len(results) > 0, (
        f"STOP: No results returned for query: {query}"
    )

    assert len(results) <= TOP_K


    # --------------------------------------------------------
    # Display
    # --------------------------------------------------------

    for rank, result in enumerate(
        results,
        start=1
    ):

        print()
        print(
            f"{rank}. "
            f"Chunk {result['chunk_id']} | "
            f"Score {result['score']:.6f}"
        )

        print(
            f"   Heading: {result['heading']}"
        )

        print(
            f"   Pages: "
            f"{result['page_start']}-"
            f"{result['page_end']}"
        )


# ============================================================
# FINAL REPORT
# ============================================================

print()
print("=" * 70)
print("MULTI-QUERY RETRIEVAL TEST: PASS")
print("=" * 70)

print(
    f"Queries tested: {len(TEST_QUERIES)}"
)

print(
    f"Top-K per query: {TOP_K}"
)

print(
    f"Vector index: {INDEX_NAME}"
)

print("=" * 70)
````

````python
# ============================================================
# CELL 14 — RAG RETRIEVER
# ============================================================

import numpy as np


print("=" * 70)
print("CELL 14 — RAG RETRIEVER")
print("=" * 70)


# ============================================================
# CONFIGURATION
# ============================================================

VECTOR_FIELD = "embedding"
VECTOR_DIMENSION = 1024
INDEX_NAME = "chunk_vector_index"

DEFAULT_TOP_K = 5
DEFAULT_NUM_CANDIDATES = 50


# ============================================================
# SAFETY CHECKS
# ============================================================

assert "embedding_model" in globals(), (
    "STOP: embedding_model is not defined. "
    "Run Cell 8 first."
)

assert "collection" in globals(), (
    "STOP: MongoDB collection is not defined. "
    "Run Cell 11 first."
)

assert callable(embedding_model.encode), (
    "STOP: embedding_model is not usable."
)


# ============================================================
# RETRIEVER FUNCTION
# ============================================================

def retrieve_chunks(
    query,
    top_k=DEFAULT_TOP_K,
    num_candidates=DEFAULT_NUM_CANDIDATES
):

    # --------------------------------------------------------
    # Validate input
    # --------------------------------------------------------

    assert isinstance(query, str), (
        "STOP: Query must be a string."
    )

    assert query.strip(), (
        "STOP: Query cannot be empty."
    )

    assert top_k > 0, (
        "STOP: top_k must be greater than zero."
    )

    assert num_candidates >= top_k, (
        "STOP: num_candidates must be >= top_k."
    )


    # --------------------------------------------------------
    # Generate query embedding
    # --------------------------------------------------------

    query_vector = embedding_model.encode(
        query,
        convert_to_numpy=True
    )


    query_vector = np.asarray(
        query_vector,
        dtype=np.float32
    )


    # --------------------------------------------------------
    # Validate embedding
    # --------------------------------------------------------

    assert query_vector.ndim == 1, (
        f"STOP: Query embedding must be 1D, "
        f"got {query_vector.shape}"
    )

    assert len(query_vector) == VECTOR_DIMENSION, (
        f"STOP: Expected {VECTOR_DIMENSION} dimensions, "
        f"got {len(query_vector)}"
    )

    assert np.isfinite(query_vector).all(), (
        "STOP: Query embedding contains invalid values."
    )


    # --------------------------------------------------------
    # L2 normalization
    # --------------------------------------------------------

    norm = np.linalg.norm(
        query_vector
    )


    assert norm > 0, (
        "STOP: Query embedding has zero norm."
    )


    query_vector = (
        query_vector / norm
    ).astype(np.float32)


    # --------------------------------------------------------
    # MongoDB Vector Search
    # --------------------------------------------------------

    pipeline = [

        {
            "$vectorSearch": {

                "index": INDEX_NAME,

                "path": VECTOR_FIELD,

                "queryVector": query_vector.tolist(),

                "numCandidates": num_candidates,

                "limit": top_k
            }
        },

        {
            "$project": {

                "_id": 0,

                "chunk_id": 1,

                "text": 1,

                "heading": 1,

                "level": 1,

                "page_start": 1,

                "page_end": 1,

                "chunk_order": 1,

                "word_count": 1,

                "score": {
                    "$meta": "vectorSearchScore"
                }
            }
        }
    ]


    results = list(
        collection.aggregate(
            pipeline
        )
    )


    # --------------------------------------------------------
    # Validate results
    # --------------------------------------------------------

    assert len(results) > 0, (
        "STOP: MongoDB returned zero results."
    )


    for result in results:

        assert "chunk_id" in result
        assert "text" in result
        assert "heading" in result
        assert "score" in result


    return results


# ============================================================
# TEST RETRIEVER
# ============================================================

TEST_QUERY = (
    "What is self-help and who can benefit from it?"
)


retrieved = retrieve_chunks(
    TEST_QUERY,
    top_k=5
)


# ============================================================
# DISPLAY
# ============================================================

print()
print("Test query:")
print(TEST_QUERY)

print()
print("-" * 70)

for rank, result in enumerate(
    retrieved,
    start=1
):

    print(
        f"Rank {rank} | "
        f"Chunk {result['chunk_id']} | "
        f"Score: {result['score']:.6f}"
    )

    print(
        f"Heading: {result['heading']}"
    )

    print(
        f"Pages: "
        f"{result['page_start']}-"
        f"{result['page_end']}"
    )

    print(
        f"Words: {result['word_count']}"
    )

    print("-" * 70)


# ============================================================
# FINAL REPORT
# ============================================================

print()
print(
    f"Retrieved chunks: {len(retrieved)}"
)

print(
    f"Vector index: {INDEX_NAME}"
)

print(
    f"Embedding dimension: {VECTOR_DIMENSION}"
)

print("=" * 70)
print("RAG RETRIEVER: PASS")
print("=" * 70)
````

````python
# ============================================================
# CELL 16 — PRECISION@5 — BGE-M3
# ============================================================

import numpy as np
from pymongo import MongoClient


print("=" * 70)
print("CELL 16 — PRECISION@5 — BGE-M3")
print("=" * 70)


# ============================================================
# CONFIGURATION
# ============================================================

TOP_K = 5

MODEL_NAME = "BAAI/bge-m3"

EVALUATION_SET = [
    {
        "question": "What is self-help?",
        "expected_chunks": [7],
    },
    {
        "question": "Who is self-help intended for?",
        "expected_chunks": [12, 13],
    },
    {
        "question": "When should guided self-help be used?",
        "expected_chunks": [14],
    },
    {
        "question": "What should a helper do when someone is at risk of suicide?",
        "expected_chunks": [47, 82],
    },
    {
        "question": "What are the Step-by-Step skills for managing stress?",
        "expected_chunks": [86, 89, 93],
    },
    {
        "question": "What are the Doing What Matters modules?",
        "expected_chunks": [104, 105, 106, 108, 114],
    },
]


# ============================================================
# SAFETY CHECKS
# ============================================================

assert "embedding_model" in globals(), (
    "STOP: embedding_model is not defined. "
    "Run Cell 8 first."
)

assert "collection" in globals(), (
    "STOP: MongoDB collection is not defined. "
    "Run Cell 11 first."
)

assert "normalized_embeddings" in globals(), (
    "STOP: normalized_embeddings not found. "
    "Run Cell 9 first."
)

assert normalized_embeddings.shape[1] == 1024, (
    "STOP: Expected 1024-dimensional BGE-M3 embeddings."
)


# ============================================================
# RETRIEVAL FUNCTION
# ============================================================

def retrieve_top_k_bge_m3(question, k=TOP_K):

    query_vector = embedding_model.encode(
        question,
        convert_to_numpy=True
    ).astype(np.float32)

    query_norm = np.linalg.norm(query_vector)

    assert query_norm > 0, (
        "STOP: Query embedding has zero norm."
    )

    query_vector = query_vector / query_norm

    pipeline = [
        {
            "$vectorSearch": {
                "index": "chunk_vector_index",
                "path": "embedding",
                "queryVector": query_vector.tolist(),
                "numCandidates": max(k * 10, 50),
                "limit": k,
            }
        },
        {
            "$project": {
                "_id": 0,
                "chunk_id": 1,
                "score": {
                    "$meta": "vectorSearchScore"
                }
            }
        }
    ]

    return list(
        collection.aggregate(pipeline)
    )


# ============================================================
# RUN EVALUATION
# ============================================================

results = []

total_precision = 0.0

print()
print("=" * 70)
print("PRECISION@5 RESULTS")
print("=" * 70)


for i, item in enumerate(EVALUATION_SET, start=1):

    question = item["question"]
    expected = set(item["expected_chunks"])

    retrieved = retrieve_top_k_bge_m3(
        question,
        TOP_K
    )

    retrieved_ids = [
        int(r["chunk_id"])
        for r in retrieved
    ]

    hits = [
        chunk_id
        for chunk_id in retrieved_ids
        if chunk_id in expected
    ]

    precision_at_5 = len(hits) / TOP_K

    total_precision += precision_at_5

    results.append({
        "question": question,
        "expected_chunks": sorted(expected),
        "retrieved_chunks": retrieved_ids,
        "hits": hits,
        "precision_at_5": precision_at_5,
    })

    print()
    print(f"Q{i}: {question}")
    print(f"Expected:  {sorted(expected)}")
    print(f"Retrieved: {retrieved_ids}")
    print(f"Hits:      {hits}")
    print(f"Precision@5: {precision_at_5:.3f}")


# ============================================================
# OVERALL SCORE
# ============================================================

mean_precision_at_5 = (
    total_precision / len(EVALUATION_SET)
)


print()
print("=" * 70)
print("BGE-M3 PRECISION@5 SUMMARY")
print("=" * 70)

print(
    f"Model: {MODEL_NAME}"
)

print(
    f"Queries evaluated: {len(EVALUATION_SET)}"
)

print(
    f"Top-K: {TOP_K}"
)

print(
    f"Mean Precision@5: {mean_precision_at_5:.4f}"
)

print("=" * 70)
print("PRECISION@5 — BGE-M3: PASS")
print("=" * 70)
````

````python
# ============================================================
# CELL 16 CHECKPOINT — SAVE PRECISION@5 RESULTS
# ============================================================

import pickle
import os

precision_checkpoint = os.path.join(
    CHECKPOINT_DIR,
    "precision_at_5_bge_m3.pkl"
)

precision_checkpoint_data = {
    "model_name": MODEL_NAME,
    "top_k": TOP_K,
    "evaluation_set": EVALUATION_SET,
    "results": results,
    "mean_precision_at_5": mean_precision_at_5,
}

with open(precision_checkpoint, "wb") as f:
    pickle.dump(
        precision_checkpoint_data,
        f
    )

print("=" * 70)
print("PRECISION@5 CHECKPOINT SAVED")
print("=" * 70)

print(
    "Mean Precision@5:",
    f"{mean_precision_at_5:.4f}"
)

print(
    "Saved to:",
    precision_checkpoint
)
````

````python
# ============================================================
# CELL 17 — PRECISION@5 — SECOND EMBEDDING MODEL
# ============================================================

from sentence_transformers import SentenceTransformer
import numpy as np


print("=" * 70)
print("CELL 17 — PRECISION@5 — SECOND EMBEDDING MODEL")
print("=" * 70)


# ============================================================
# CONFIGURATION
# ============================================================

SECOND_MODEL_NAME = "sentence-transformers/all-MiniLM-L6-v2"

TOP_K = 5


# ============================================================
# REUSE THE SAME EVALUATION SET
# ============================================================

assert "retrieval_test_set" in globals(), (
    "STOP: retrieval_test_set is not defined. "
    "Run Cell 15 first."
)

assert len(retrieval_test_set) == 6, (
    "STOP: Expected the 6-question evaluation set."
)


# ============================================================
# LOAD SECOND MODEL
# ============================================================

print()
print("Loading second embedding model:")
print(SECOND_MODEL_NAME)

second_embedding_model = SentenceTransformer(
    SECOND_MODEL_NAME
)


# ============================================================
# VERIFY MODEL DIMENSION
# ============================================================

second_embedding_dimension = (
    second_embedding_model.get_sentence_embedding_dimension()
)

assert second_embedding_dimension > 0, (
    "STOP: Invalid embedding dimension."
)


print(
    "Second model dimension:",
    second_embedding_dimension
)


# ============================================================
# PREPARE LOCAL CHUNK EMBEDDINGS
# ============================================================

assert "chunks" in globals(), (
    "STOP: chunks is not defined."
)

assert len(chunks) > 0, (
    "STOP: No chunks available."
)


chunk_texts_second_model = [
    chunk["text"]
    for chunk in chunks
]


# ============================================================
# GENERATE EMBEDDINGS FOR ALL CHUNKS
# ============================================================

print()
print("Generating embeddings for 166 chunks...")

second_chunk_embeddings = (
    second_embedding_model.encode(
        chunk_texts_second_model,
        batch_size=16,
        show_progress_bar=True,
        convert_to_numpy=True,
        normalize_embeddings=True,
    )
)


second_chunk_embeddings = np.asarray(
    second_chunk_embeddings,
    dtype=np.float32
)


# ============================================================
# VALIDATE CHUNK EMBEDDINGS
# ============================================================

assert second_chunk_embeddings.ndim == 2, (
    "STOP: Second-model embeddings must be 2D."
)

assert second_chunk_embeddings.shape[0] == len(chunks), (
    "STOP: Second-model embedding count does not "
    "match number of chunks."
)

assert second_chunk_embeddings.shape[1] == (
    second_embedding_dimension
), (
    "STOP: Second-model embedding dimension mismatch."
)

assert np.isfinite(
    second_chunk_embeddings
).all(), (
    "STOP: Second-model embeddings contain "
    "NaN or infinite values."
)


# ============================================================
# PREPARE CHUNK IDs
# ============================================================

chunk_ids = np.array(
    [
        int(chunk["chunk_id"])
        for chunk in chunks
    ]
)


# ============================================================
# RETRIEVAL USING SECOND MODEL
# ============================================================

def retrieve_top_k_second_model(
    question,
    k=TOP_K
):

    query_embedding = (
        second_embedding_model.encode(
            question,
            convert_to_numpy=True,
            normalize_embeddings=True,
        )
    )

    query_embedding = np.asarray(
        query_embedding,
        dtype=np.float32
    )

    assert query_embedding.ndim == 1, (
        "STOP: Query embedding must be 1D."
    )

    assert len(query_embedding) == (
        second_embedding_dimension
    ), (
        "STOP: Query embedding dimension mismatch."
    )

    # Cosine similarity because embeddings are normalized.
    scores = (
        second_chunk_embeddings
        @ query_embedding
    )

    top_indices = np.argsort(
        scores
    )[::-1][:k]

    return [
        {
            "chunk_id": int(chunk_ids[index]),
            "score": float(scores[index]),
        }
        for index in top_indices
    ]


# ============================================================
# RUN PRECISION@5 EVALUATION
# ============================================================

second_model_results = []

total_precision = 0.0


print()
print("=" * 70)
print("PRECISION@5 RESULTS")
print("=" * 70)


for i, item in enumerate(
    retrieval_test_set,
    start=1
):

    question = item["question"]

    expected = set(
        item["expected_chunks"]
    )

    retrieved = retrieve_top_k_second_model(
        question,
        TOP_K
    )

    retrieved_ids = [
        result["chunk_id"]
        for result in retrieved
    ]

    hits = [
        chunk_id
        for chunk_id in retrieved_ids
        if chunk_id in expected
    ]

    precision_at_5 = (
        len(hits) / TOP_K
    )

    total_precision += precision_at_5

    second_model_results.append({
        "question": question,
        "expected_chunks": sorted(expected),
        "retrieved_chunks": retrieved_ids,
        "hits": hits,
        "precision_at_5": precision_at_5,
    })

    print()
    print(
        f"Q{i}: {question}"
    )

    print(
        f"Expected:  {sorted(expected)}"
    )

    print(
        f"Retrieved: {retrieved_ids}"
    )

    print(
        f"Hits:      {hits}"
    )

    print(
        f"Precision@5: {precision_at_5:.3f}"
    )


# ============================================================
# OVERALL SCORE
# ============================================================

second_model_mean_precision_at_5 = (
    total_precision / len(retrieval_test_set)
)


# ============================================================
# FINAL REPORT
# ============================================================

print()
print("=" * 70)
print("SECOND MODEL PRECISION@5 SUMMARY")
print("=" * 70)

print(
    "Model:",
    SECOND_MODEL_NAME
)

print(
    "Queries evaluated:",
    len(retrieval_test_set)
)

print(
    "Top-K:",
    TOP_K
)

print(
    "Embedding dimension:",
    second_embedding_dimension
)

print(
    f"Mean Precision@5: "
    f"{second_model_mean_precision_at_5:.4f}"
)

print("=" * 70)
print("PRECISION@5 — SECOND MODEL: PASS")
print("=" * 70)
````

````python
 # ============================================================
# CELL 19 — CHUNK SIZE / OVERLAP EXPERIMENT
# ============================================================

import re
import numpy as np


print("=" * 70)
print("CELL 19 — CHUNK SIZE / OVERLAP EXPERIMENT")
print("=" * 70)


# ============================================================
# CONFIGURATION
# ============================================================

EXPERIMENT_CONFIGS = [

    {
        "name": "Small",
        "target_size": 300,
        "max_overlap": 50,
    },

    {
        "name": "Baseline",
        "target_size": 500,
        "max_overlap": 75,
    },

    {
        "name": "Large",
        "target_size": 700,
        "max_overlap": 100,
    },
]


TOP_K = 5
EMBEDDING_DIMENSION = 1024


# ============================================================
# SAFETY CHECKS
# ============================================================

assert "embedding_model" in globals(), (
    "STOP: embedding_model is not defined. Run Cell 8 first."
)

assert callable(embedding_model.encode), (
    "STOP: embedding_model is not usable."
)

assert "cleaned_section_paragraphs" in globals(), (
    "STOP: cleaned_section_paragraphs not found."
)

assert len(cleaned_section_paragraphs) > 0, (
    "STOP: No cleaned sections available."
)

assert "chunks" in globals(), (
    "STOP: baseline chunks are not available."
)

assert len(chunks) > 0, (
    "STOP: baseline chunks are empty."
)


# ============================================================
# DAY 2 EVALUATION SET
#
# Defined explicitly here so Cell 19 does not depend on the
# runtime type/format of evaluation_questions.
# ============================================================

DAY2_EVALUATION = [

    {
        "question": "What is self-help?",
        "expected_chunks": [7],
    },

    {
        "question": "Who is self-help intended for?",
        "expected_chunks": [12, 13],
    },

    {
        "question": "When should guided self-help be used?",
        "expected_chunks": [14],
    },

    {
        "question": (
            "What should a helper do when someone is "
            "at risk of suicide?"
        ),
        "expected_chunks": [47, 82],
    },

    {
        "question": (
            "What are the Step-by-Step skills "
            "for managing stress?"
        ),
        "expected_chunks": [86, 89, 93],
    },

    {
        "question": (
            "What are the Doing What Matters modules?"
        ),
        "expected_chunks": [104, 105, 106, 108, 114],
    },
]


assert len(DAY2_EVALUATION) == 6, (
    "STOP: Expected exactly 6 evaluation questions."
)


# ============================================================
# HELPER FUNCTIONS
# ============================================================

def experiment_count_words(text):

    return len(
        re.findall(
            r"\b\w+\b",
            text
        )
    )


def experiment_make_chunk(
    chunk_id,
    chunk_order,
    section,
    paragraphs
):

    text = "\n\n".join(paragraphs)

    return {

        "chunk_id": chunk_id,

        "chunk_order": chunk_order,

        "text": text,

        "heading": section["heading"],

        "level": section["level"],

        "page_start": section["page_start"],

        "page_end": section["page_end"],

        "word_count": experiment_count_words(text),
    }


# ============================================================
# EXPERIMENTAL CHUNK BUILDER
# ============================================================

def build_experimental_chunks(
    target_size,
    max_overlap
):

    experimental_chunks = []

    chunk_id = 0
    chunk_order = 0


    for section in cleaned_section_paragraphs:

        paragraphs = section["paragraphs"]

        current_paras = []
        current_words = 0


        for para in paragraphs:

            para_words = experiment_count_words(para)


            # ------------------------------------------------
            # LARGE SOURCE PARAGRAPH
            # ------------------------------------------------

            if para_words > target_size:

                if current_paras:

                    experimental_chunks.append(
                        experiment_make_chunk(
                            chunk_id,
                            chunk_order,
                            section,
                            current_paras
                        )
                    )

                    chunk_id += 1
                    chunk_order += 1

                    current_paras = []
                    current_words = 0


                experimental_chunks.append(
                    experiment_make_chunk(
                        chunk_id,
                        chunk_order,
                        section,
                        [para]
                    )
                )

                chunk_id += 1
                chunk_order += 1

                continue


            # ------------------------------------------------
            # DOES NOT FIT
            # ------------------------------------------------

            if (
                current_paras
                and current_words + para_words > target_size
            ):

                previous_paras = list(
                    current_paras
                )


                # Save current chunk

                experimental_chunks.append(
                    experiment_make_chunk(
                        chunk_id,
                        chunk_order,
                        section,
                        previous_paras
                    )
                )

                chunk_id += 1
                chunk_order += 1


                # Build overlap

                overlap_paras = []
                overlap_words = 0


                for previous_para in reversed(
                    previous_paras
                ):

                    previous_words = (
                        experiment_count_words(
                            previous_para
                        )
                    )


                    if (
                        overlap_words
                        + previous_words
                        <= max_overlap
                    ):

                        overlap_paras.insert(
                            0,
                            previous_para
                        )

                        overlap_words += previous_words

                    else:

                        break


                current_paras = list(
                    overlap_paras
                )

                current_words = overlap_words


                # Always add current paragraph

                current_paras.append(
                    para
                )

                current_words += para_words

                continue


            # ------------------------------------------------
            # FITS NORMALLY
            # ------------------------------------------------

            current_paras.append(
                para
            )

            current_words += para_words


        # ----------------------------------------------------
        # FINAL CHUNK
        # ----------------------------------------------------

        if current_paras:

            experimental_chunks.append(
                experiment_make_chunk(
                    chunk_id,
                    chunk_order,
                    section,
                    current_paras
                )
            )

            chunk_id += 1
            chunk_order += 1


    # --------------------------------------------------------
    # Validation
    # --------------------------------------------------------

    assert len(experimental_chunks) > 0

    assert all(
        chunk["text"].strip()
        for chunk in experimental_chunks
    )

    assert all(
        chunk["chunk_id"] == i
        for i, chunk in enumerate(
            experimental_chunks
        )
    )

    assert all(
        chunk["chunk_order"] == i
        for i, chunk in enumerate(
            experimental_chunks
        )
    )


    return experimental_chunks


# ============================================================
# BUILD EXPECTED HEADINGS
#
# Cell 15 chunk IDs refer to the baseline 500/75 chunks.
# We convert those IDs into semantic headings.
# ============================================================

expected_headings_by_question = []


for item in DAY2_EVALUATION:

    expected_ids = item["expected_chunks"]

    headings = set()


    for chunk_id in expected_ids:

        assert (
            0 <= chunk_id < len(chunks)
        ), (
            f"STOP: Expected baseline chunk ID "
            f"{chunk_id} is outside range."
        )

        headings.add(
            chunks[chunk_id]["heading"]
        )


    expected_headings_by_question.append(
        headings
    )


# ============================================================
# PRECOMPUTE QUERY EMBEDDINGS ONCE
# ============================================================

print()
print("=" * 70)
print("PRECOMPUTING QUERY EMBEDDINGS")
print("=" * 70)


query_texts = [
    item["question"]
    for item in DAY2_EVALUATION
]


query_embeddings = embedding_model.encode(
    query_texts,
    batch_size=8,
    show_progress_bar=True,
    convert_to_numpy=True
)


query_embeddings = np.asarray(
    query_embeddings,
    dtype=np.float32
)


assert query_embeddings.ndim == 2

assert (
    query_embeddings.shape[0]
    == len(DAY2_EVALUATION)
)

assert (
    query_embeddings.shape[1]
    == EMBEDDING_DIMENSION
)


# Normalize

query_norms = np.linalg.norm(
    query_embeddings,
    axis=1,
    keepdims=True
)


assert np.all(
    query_norms > 0
)


query_embeddings = (
    query_embeddings / query_norms
).astype(
    np.float32
)


assert np.isfinite(
    query_embeddings
).all()


print(
    "Query embeddings shape:",
    query_embeddings.shape
)

print(
    "Query embedding normalization: PASS"
)


# ============================================================
# RUN EXPERIMENTS
# ============================================================

experiment_results = []


# Keep final objects available for inspection
experimental_chunks = None
experimental_embeddings = None


for config_index, config in enumerate(
    EXPERIMENT_CONFIGS,
    start=1
):

    config_name = config["name"]
    target_size = config["target_size"]
    max_overlap = config["max_overlap"]


    print()
    print("=" * 70)
    print(
        f"CONFIGURATION "
        f"{config_index}/{len(EXPERIMENT_CONFIGS)}: "
        f"{config_name}"
    )

    print(
        f"Target size: {target_size}"
    )

    print(
        f"Maximum overlap: {max_overlap}"
    )

    print("=" * 70)


    # --------------------------------------------------------
    # Build chunks
    # --------------------------------------------------------

    experimental_chunks = build_experimental_chunks(
        target_size=target_size,
        max_overlap=max_overlap
    )


    word_counts = [
        chunk["word_count"]
        for chunk in experimental_chunks
    ]


    print(
        f"\nChunks generated: "
        f"{len(experimental_chunks)}"
    )

    print(
        f"Minimum words: "
        f"{min(word_counts)}"
    )

    print(
        f"Maximum words: "
        f"{max(word_counts)}"
    )

    print(
        f"Average words: "
        f"{sum(word_counts) / len(word_counts):.2f}"
    )


    # --------------------------------------------------------
    # Generate embeddings
    # --------------------------------------------------------

    print(
        "\nGenerating chunk embeddings..."
    )


    experimental_texts = [
        chunk["text"]
        for chunk in experimental_chunks
    ]


    experimental_embeddings = (
        embedding_model.encode(
            experimental_texts,
            batch_size=32,
            show_progress_bar=True,
            convert_to_numpy=True
        )
    )


    experimental_embeddings = np.asarray(
        experimental_embeddings,
        dtype=np.float32
    )


    assert experimental_embeddings.ndim == 2

    assert (
        experimental_embeddings.shape[0]
        == len(experimental_chunks)
    )

    assert (
        experimental_embeddings.shape[1]
        == EMBEDDING_DIMENSION
    )

    assert np.isfinite(
        experimental_embeddings
    ).all()


    # --------------------------------------------------------
    # Normalize chunk embeddings
    # --------------------------------------------------------

    experimental_norms = np.linalg.norm(
        experimental_embeddings,
        axis=1,
        keepdims=True
    )


    assert np.all(
        experimental_norms > 0
    )


    experimental_embeddings = (
        experimental_embeddings
        / experimental_norms
    ).astype(
        np.float32
    )


    normalized_norms = np.linalg.norm(
        experimental_embeddings,
        axis=1
    )


    assert np.all(
        np.abs(
            normalized_norms - 1.0
        ) < 1e-4
    )


    print(
        "Chunk embedding normalization: PASS"
    )


    # --------------------------------------------------------
    # Retrieval evaluation
    # --------------------------------------------------------

    all_top5_similarities = []

    successful_heading_queries = 0


    for query_index, query_vector in enumerate(
        query_embeddings
    ):

        similarities = (
            experimental_embeddings
            @ query_vector
        )


        top_indices = np.argsort(
            similarities
        )[::-1][:TOP_K]


        top_scores = similarities[
            top_indices
        ]


        all_top5_similarities.extend(
            top_scores.tolist()
        )


        # Expected semantic headings

        expected_headings = (
            expected_headings_by_question[
                query_index
            ]
        )


        retrieved_headings = {

            experimental_chunks[
                int(index)
            ]["heading"]

            for index in top_indices
        }


        if (
            retrieved_headings
            & expected_headings
        ):

            successful_heading_queries += 1


    # --------------------------------------------------------
    # Metrics
    # --------------------------------------------------------

    mean_top5_similarity = float(
        np.mean(
            all_top5_similarities
        )
    )


    heading_hit_rate = (
        successful_heading_queries
        / len(DAY2_EVALUATION)
    )


    # --------------------------------------------------------
    # Validation
    # --------------------------------------------------------

    assert (
        0.0
        <= mean_top5_similarity
        <= 1.0
    )


    assert (
        0.0
        <= heading_hit_rate
        <= 1.0
    )


    experiment_results.append({

        "name": config_name,

        "target_size": target_size,

        "overlap": max_overlap,

        "chunks": len(
            experimental_chunks
        ),

        "avg_words": (
            sum(word_counts)
            / len(word_counts)
        ),

        "mean_top5_similarity":
            mean_top5_similarity,

        "heading_hit_rate":
            heading_hit_rate,
    })


    print()
    print(
        f"Mean Top-5 similarity: "
        f"{mean_top5_similarity:.6f}"
    )

    print(
        f"Heading hit rate: "
        f"{heading_hit_rate:.4f}"
    )

    print()
    print(
        f"Configuration {config_name}: PASS"
    )


# ============================================================
# SUMMARY
# ============================================================

print()
print("=" * 70)
print("CHUNK SIZE / OVERLAP EXPERIMENT SUMMARY")
print("=" * 70)


print()
print(
    f"{'Config':<15}"
    f"{'Target':>10}"
    f"{'Overlap':>10}"
    f"{'Chunks':>10}"
    f"{'Avg words':>13}"
    f"{'Mean Top-5':>14}"
    f"{'Hit rate':>12}"
)

print("-" * 84)


for result in experiment_results:

    print(
        f"{result['name']:<15}"
        f"{result['target_size']:>10}"
        f"{result['overlap']:>10}"
        f"{result['chunks']:>10}"
        f"{result['avg_words']:>13.2f}"
        f"{result['mean_top5_similarity']:>14.6f}"
        f"{result['heading_hit_rate']:>12.4f}"
    )


# ============================================================
# SELECT BEST CONFIGURATION
# ============================================================

baseline_target = 500
baseline_overlap = 75


best_result = max(
    experiment_results,
    key=lambda result: (
        result["mean_top5_similarity"],
        result["heading_hit_rate"],
        -abs(
            result["target_size"]
            - baseline_target
        ),
        -abs(
            result["overlap"]
            - baseline_overlap
        ),
    )
)


print()
print("-" * 70)

print(
    "Best experimental configuration:"
)

print(
    f"Name: "
    f"{best_result['name']}"
)

print(
    f"Target size: "
    f"{best_result['target_size']}"
)

print(
    f"Overlap: "
    f"{best_result['overlap']}"
)

print(
    f"Chunks: "
    f"{best_result['chunks']}"
)

print(
    f"Mean Top-5 similarity: "
    f"{best_result['mean_top5_similarity']:.6f}"
)

print(
    f"Heading hit rate: "
    f"{best_result['heading_hit_rate']:.4f}"
)


# ============================================================
# FINAL VALIDATION
# ============================================================

assert len(
    experiment_results
) == 3


assert all(
    0.0
    <= result["heading_hit_rate"]
    <= 1.0
    for result in expذeriment_results
)


assert all(
    0.0
    <= result["mean_top5_similarity"]
    <= 1.0
    for result in experiment_results
)


assert best_result in experiment_results


# ============================================================
# FINAL REPORT
# ============================================================

print()
print("=" * 70)
print("CHUNK SIZE / OVERLAP EXPERIMENT: PASS")
print("=" * 70)
````

````python
# ============================================================
# CELL 20 — FINAL CHUNK CONFIGURATION
# ============================================================

import re


print("=" * 70)
print("CELL 20 — FINAL CHUNK CONFIGURATION")
print("=" * 70)


# ============================================================
# 1. FINAL CONFIGURATION
#
# Selected by Cell 19 experiment:
#
# Small      -> 300 words / 50 overlap
# Baseline   -> 500 words / 75 overlap
# Large      -> 700 words / 100 overlap
#
# Small achieved the highest Mean Top-5 similarity:
# 0.658741
#
# All three configurations achieved:
# Heading hit rate = 1.0000
# ============================================================

FINAL_CHUNK_TARGET_SIZE = 300
FINAL_CHUNK_OVERLAP = 50

FINAL_CHUNK_CONFIG_NAME = "Small"


# ============================================================
# 2. SAFETY CHECKS
# ============================================================

assert "cleaned_section_paragraphs" in globals(), (
    "STOP: cleaned_section_paragraphs is not defined. "
    "Run the cleaning/chunk preparation cells first."
)

assert len(cleaned_section_paragraphs) > 0, (
    "STOP: No cleaned sections are available."
)


assert "best_result" in globals(), (
    "STOP: best_result is not available. "
    "Run Cell 19 first."
)


# ============================================================
# 3. VERIFY CELL 19 SELECTION
#
# Do not silently use a different configuration.
# ============================================================

assert best_result["name"] == FINAL_CHUNK_CONFIG_NAME, (
    "STOP: Cell 19 selected a different best configuration."
)

assert best_result["target_size"] == FINAL_CHUNK_TARGET_SIZE, (
    "STOP: Final target size does not match Cell 19."
)

assert best_result["overlap"] == FINAL_CHUNK_OVERLAP, (
    "STOP: Final overlap does not match Cell 19."
)


print()
print("Selected configuration:")
print(f"Name: {FINAL_CHUNK_CONFIG_NAME}")
print(f"Target size: {FINAL_CHUNK_TARGET_SIZE}")
print(f"Overlap: {FINAL_CHUNK_OVERLAP}")
print(f"Cell 19 Mean Top-5: {best_result['mean_top5_similarity']:.6f}")
print(f"Cell 19 Heading hit rate: {best_result['heading_hit_rate']:.4f}")


# ============================================================
# 4. WORD COUNT HELPER
# ============================================================

def final_count_words(text):

    return len(
        re.findall(
            r"\b\w+\b",
            text
        )
    )


# ============================================================
# 5. FINAL CHUNK CREATOR
# ============================================================

def make_final_chunk(
    chunk_id,
    chunk_order,
    section,
    paragraphs
):

    text = "\n\n".join(paragraphs)

    return {

        "chunk_id": int(chunk_id),

        "chunk_order": int(chunk_order),

        "text": text,

        "heading": section["heading"],

        "level": section["level"],

        "page_start": int(
            section["page_start"]
        ),

        "page_end": int(
            section["page_end"]
        ),

        "word_count": int(
            final_count_words(text)
        ),
    }


# ============================================================
# 6. BUILD FINAL CHUNKS
# ============================================================

final_chunks = []

chunk_id = 0
chunk_order = 0


for section in cleaned_section_paragraphs:

    paragraphs = section["paragraphs"]

    current_paras = []
    current_words = 0


    for para in paragraphs:

        para_words = final_count_words(
            para
        )


        # ----------------------------------------------------
        # SOURCE PARAGRAPH LARGER THAN TARGET
        # ----------------------------------------------------

        if para_words > FINAL_CHUNK_TARGET_SIZE:

            if current_paras:

                final_chunks.append(
                    make_final_chunk(
                        chunk_id,
                        chunk_order,
                        section,
                        current_paras
                    )
                )

                chunk_id += 1
                chunk_order += 1

                current_paras = []
                current_words = 0


            final_chunks.append(
                make_final_chunk(
                    chunk_id,
                    chunk_order,
                    section,
                    [para]
                )
            )

            chunk_id += 1
            chunk_order += 1

            continue


        # ----------------------------------------------------
        # CURRENT PARAGRAPH DOES NOT FIT
        # ----------------------------------------------------

        if (
            current_paras
            and current_words + para_words
            > FINAL_CHUNK_TARGET_SIZE
        ):

            previous_paras = list(
                current_paras
            )


            # ------------------------------------------------
            # SAVE CURRENT CHUNK
            # ------------------------------------------------

            final_chunks.append(
                make_final_chunk(
                    chunk_id,
                    chunk_order,
                    section,
                    previous_paras
                )
            )

            chunk_id += 1
            chunk_order += 1


            # ------------------------------------------------
            # BUILD OVERLAP
            # ------------------------------------------------

            overlap_paras = []
            overlap_words = 0


            for previous_para in reversed(
                previous_paras
            ):

                previous_words = final_count_words(
                    previous_para
                )


                if (
                    overlap_words
                    + previous_words
                    <= FINAL_CHUNK_OVERLAP
                ):

                    overlap_paras.insert(
                        0,
                        previous_para
                    )

                    overlap_words += previous_words

                else:

                    break


            current_paras = list(
                overlap_paras
            )

            current_words = overlap_words


            # ------------------------------------------------
            # ADD CURRENT PARAGRAPH
            # ------------------------------------------------

            current_paras.append(
                para
            )

            current_words += para_words

            continue


        # ----------------------------------------------------
        # PARAGRAPH FITS
        # ----------------------------------------------------

        current_paras.append(
            para
        )

        current_words += para_words


    # --------------------------------------------------------
    # FINAL SECTION CHUNK
    # --------------------------------------------------------

    if current_paras:

        final_chunks.append(
            make_final_chunk(
                chunk_id,
                chunk_order,
                section,
                current_paras
            )
        )

        chunk_id += 1
        chunk_order += 1


# ============================================================
# 7. FINAL CHUNK VALIDATION
# ============================================================

assert len(final_chunks) > 0, (
    "STOP: Final chunk generation produced zero chunks."
)


assert all(
    isinstance(chunk, dict)
    for chunk in final_chunks
), (
    "STOP: Invalid final chunk object."
)


required_fields = [
    "chunk_id",
    "chunk_order",
    "text",
    "heading",
    "level",
    "page_start",
    "page_end",
    "word_count",
]


for chunk in final_chunks:

    for field in required_fields:

        assert field in chunk, (
            f"STOP: Missing field '{field}' "
            f"in final chunk."
        )

    assert isinstance(
        chunk["text"],
        str
    )

    assert chunk["text"].strip(), (
        "STOP: Empty final chunk text."
    )

    assert chunk["word_count"] > 0, (
        "STOP: Final chunk has zero words."
    )


# ============================================================
# 8. CHUNK ID / ORDER VALIDATION
# ============================================================

assert all(
    chunk["chunk_id"] == index
    for index, chunk in enumerate(
        final_chunks
    )
), (
    "STOP: Final chunk IDs are not sequential."
)


assert all(
    chunk["chunk_order"] == index
    for index, chunk in enumerate(
        final_chunks
    )
), (
    "STOP: Final chunk order is not sequential."
)


# ============================================================
# 9. SOURCE COVERAGE VALIDATION
#
# Every cleaned section must contribute text to the final
# dataset.
# ============================================================

source_section_count = len(
    cleaned_section_paragraphs
)


final_headings = {
    chunk["heading"]
    for chunk in final_chunks
}


source_headings = {
    section["heading"]
    for section in cleaned_section_paragraphs
}


assert source_headings.issubset(
    final_headings
), (
    "STOP: One or more source section headings "
    "are missing from final chunks."
)


# ============================================================
# 10. CHUNK STATISTICS
# ============================================================

final_word_counts = [
    chunk["word_count"]
    for chunk in final_chunks
]


final_min_words = min(
    final_word_counts
)

final_max_words = max(
    final_word_counts
)

final_avg_words = (
    sum(final_word_counts)
    / len(final_word_counts)
)


# ============================================================
# 11. VERIFY AGAINST CELL 19 RESULT
#
# Cell 19 reported 248 chunks for 300/50.
# ============================================================

assert len(final_chunks) == best_result["chunks"], (
    "STOP: Final chunk count does not match "
    "the winning Cell 19 configuration."
)


# ============================================================
# 12. FINAL REPORT
# ============================================================

print()
print("=" * 70)
print("FINAL CHUNK DATASET")
print("=" * 70)

print()
print(f"Configuration: {FINAL_CHUNK_CONFIG_NAME}")
print(f"Target size: {FINAL_CHUNK_TARGET_SIZE}")
print(f"Maximum overlap: {FINAL_CHUNK_OVERLAP}")

print()
print(f"Source sections: {source_section_count}")
print(f"Final chunks: {len(final_chunks):,}")
print(f"Minimum words: {final_min_words}")
print(f"Maximum words: {final_max_words}")
print(f"Average words: {final_avg_words:.2f}")

print()
print(
    f"Expected chunks from Cell 19: "
    f"{best_result['chunks']:,}"
)

print(
    f"Actual final chunks: "
    f"{len(final_chunks):,}"
)

print()
print("Chunk IDs: PASS")
print("Chunk order: PASS")
print("Required metadata: PASS")
print("Source heading coverage: PASS")
print("Cell 19 configuration match: PASS")


# ============================================================
# 13. FINAL SAMPLE
# ============================================================

print()
print("=" * 70)
print("FINAL CHUNK SAMPLE")
print("=" * 70)

for sample_chunk in final_chunks[:3]:

    print()
    print(
        f"Chunk ID: {sample_chunk['chunk_id']}"
    )

    print(
        f"Heading: {sample_chunk['heading']}"
    )

    print(
        f"Pages: "
        f"{sample_chunk['page_start']}-"
        f"{sample_chunk['page_end']}"
    )

    print(
        f"Words: "
        f"{sample_chunk['word_count']}"
    )

    print("-" * 70)

    print(
        sample_chunk["text"][:500]
    )


# ============================================================
# FINAL PASS
# ============================================================

print()
print("=" * 70)
print("CELL 20 — FINAL CHUNK CONFIGURATION: PASS")
print("=" * 70)
````

````python
# ============================================================
# CELL 22 — FINAL MONGODB DATASET UPLOAD
# RECOVERY-SAFE VERSION
# ============================================================

import os
import pickle
import numpy as np


print("=" * 70)
print("CELL 22 — FINAL MONGODB DATASET UPLOAD")
print("RECOVERY-SAFE VERSION")
print("=" * 70)


# ============================================================
# 1. CONFIGURATION
# ============================================================

DATABASE_NAME = "psychological_self_help"
COLLECTION_NAME = "chunks"

VECTOR_FIELD = "embedding"
VECTOR_DIMENSION = 1024

INDEX_NAME = "chunk_vector_index"

EXPECTED_FINAL_CHUNKS = 248

EXPECTED_TARGET_SIZE = 300
EXPECTED_OVERLAP = 50

EXPECTED_MODEL_NAME = "BAAI/bge-m3"


# ============================================================
# 2. CHECK MONGODB CONNECTION
# ============================================================

assert "collection" in globals(), (
    "STOP: MongoDB collection is not available in RAM.\n"
    "Run the MongoDB connection part of Cell 11 first.\n"
    "Do NOT run the old baseline upload logic."
)


print()
print("MongoDB connection object: FOUND")
print("Database:", DATABASE_NAME)
print("Collection:", COLLECTION_NAME)


# ============================================================
# 3. RECOVER FINAL CHUNKS FROM CHECKPOINT
# ============================================================

FINAL_CHUNKS_CHECKPOINT = os.path.join(
    CHECKPOINT_DIR,
    "final_chunks_checkpoint.pkl"
)


assert os.path.exists(
    FINAL_CHUNKS_CHECKPOINT
), (
    "STOP: Final chunks checkpoint not found:\n"
    f"{FINAL_CHUNKS_CHECKPOINT}"
)


print()
print("Recovering final chunks checkpoint...")


with open(
    FINAL_CHUNKS_CHECKPOINT,
    "rb"
) as f:

    final_chunks_checkpoint = pickle.load(f)


assert isinstance(
    final_chunks_checkpoint,
    dict
), (
    "STOP: Final chunks checkpoint must be a dictionary."
)


assert "chunks" in final_chunks_checkpoint, (
    "STOP: 'chunks' key missing from final chunks checkpoint."
)


# ============================================================
# 4. RESTORE FINAL CHUNKS
# ============================================================

final_chunks = final_chunks_checkpoint["chunks"]


assert isinstance(
    final_chunks,
    list
), (
    "STOP: Recovered final_chunks must be a list."
)


assert len(final_chunks) == EXPECTED_FINAL_CHUNKS, (
    f"STOP: Expected {EXPECTED_FINAL_CHUNKS} final chunks, "
    f"got {len(final_chunks)}."
)


# ============================================================
# 5. VALIDATE CHUNK CONFIGURATION
# ============================================================

assert (
    final_chunks_checkpoint["config_name"]
    == "Small"
), (
    "STOP: Final chunk configuration is not 'Small'."
)


assert (
    final_chunks_checkpoint["target_size"]
    == EXPECTED_TARGET_SIZE
), (
    "STOP: Final chunk target size does not match."
)


assert (
    final_chunks_checkpoint["overlap"]
    == EXPECTED_OVERLAP
), (
    "STOP: Final chunk overlap does not match."
)


assert (
    final_chunks_checkpoint["num_chunks"]
    == EXPECTED_FINAL_CHUNKS
), (
    "STOP: Checkpoint chunk count does not match."
)


print()
print("Final chunks recovered:", len(final_chunks))
print(
    "Configuration:",
    final_chunks_checkpoint["config_name"]
)
print(
    "Target size:",
    final_chunks_checkpoint["target_size"]
)
print(
    "Overlap:",
    final_chunks_checkpoint["overlap"]
)
print(
    "Average words:",
    final_chunks_checkpoint["avg_words"]
)


# ============================================================
# 6. RECOVER FINAL BGE-M3 EMBEDDINGS
# ============================================================

FINAL_EMBEDDINGS_CHECKPOINT = os.path.join(
    CHECKPOINT_DIR,
    "final_bge_m3_embeddings_checkpoint.pkl"
)


assert os.path.exists(
    FINAL_EMBEDDINGS_CHECKPOINT
), (
    "STOP: Final BGE-M3 embeddings checkpoint not found:\n"
    f"{FINAL_EMBEDDINGS_CHECKPOINT}"
)


print()
print("Recovering final BGE-M3 embeddings checkpoint...")


with open(
    FINAL_EMBEDDINGS_CHECKPOINT,
    "rb"
) as f:

    final_embeddings_checkpoint = pickle.load(f)


assert isinstance(
    final_embeddings_checkpoint,
    dict
), (
    "STOP: Final embeddings checkpoint must be a dictionary."
)


# ============================================================
# 7. INSPECT EMBEDDING CHECKPOINT
# ============================================================

print()
print("Embedding checkpoint keys:")

for key in final_embeddings_checkpoint.keys():

    print(
        "  -",
        key
    )


# ============================================================
# 8. FIND EMBEDDING MATRIX INSIDE CHECKPOINT
# ============================================================

# The checkpoint created after Cell 21 should contain
# the final embedding matrix.
#
# We support the expected key names without relying
# on the RAM state of the previous runtime.


embedding_key_candidates = [

    "final_raw_embeddings",
    "final_embeddings",
    "embeddings",
    "embedding_matrix",

]


FINAL_EMBEDDING_KEY = None


for key in embedding_key_candidates:

    if key in final_embeddings_checkpoint:

        value = final_embeddings_checkpoint[key]

        if isinstance(value, np.ndarray):

            if value.shape == (
                EXPECTED_FINAL_CHUNKS,
                VECTOR_DIMENSION
            ):

                FINAL_EMBEDDING_KEY = key
                break


assert FINAL_EMBEDDING_KEY is not None, (
    "STOP: Could not find a valid 248 x 1024 embedding "
    "matrix inside final_bge_m3_embeddings_checkpoint.pkl.\n"
    "Available keys: "
    f"{list(final_embeddings_checkpoint.keys())}"
)


final_embeddings = np.asarray(
    final_embeddings_checkpoint[
        FINAL_EMBEDDING_KEY
    ],
    dtype=np.float32
)


print()
print(
    "Embedding variable recovered from checkpoint:",
    FINAL_EMBEDDING_KEY
)


# ============================================================
# 9. VALIDATE EMBEDDING CHECKPOINT METADATA
# ============================================================

if "model_name" in final_embeddings_checkpoint:

    assert (
        final_embeddings_checkpoint["model_name"]
        == EXPECTED_MODEL_NAME
    ), (
        "STOP: Embedding model does not match BAAI/bge-m3."
    )


if "embedding_dimension" in final_embeddings_checkpoint:

    assert (
        final_embeddings_checkpoint[
            "embedding_dimension"
        ]
        == VECTOR_DIMENSION
    ), (
        "STOP: Embedding dimension metadata mismatch."
    )


if "dimension" in final_embeddings_checkpoint:

    assert (
        final_embeddings_checkpoint[
            "dimension"
        ]
        == VECTOR_DIMENSION
    ), (
        "STOP: Embedding dimension metadata mismatch."
    )


# ============================================================
# 10. VALIDATE EMBEDDING MATRIX
# ============================================================

assert final_embeddings.ndim == 2, (
    "STOP: Final embedding matrix must be 2D."
)


assert final_embeddings.shape == (
    EXPECTED_FINAL_CHUNKS,
    VECTOR_DIMENSION
), (
    "STOP: Final embedding matrix shape mismatch.\n"
    f"Expected: "
    f"({EXPECTED_FINAL_CHUNKS}, {VECTOR_DIMENSION})\n"
    f"Got: {final_embeddings.shape}"
)


assert np.isfinite(
    final_embeddings
).all(), (
    "STOP: Final embeddings contain NaN or infinite values."
)


print()
print(
    "Final embedding matrix shape:",
    final_embeddings.shape
)

print(
    "Final embedding dtype:",
    final_embeddings.dtype
)

print(
    "Embedding model:",
    EXPECTED_MODEL_NAME
)

print(
    "Embedding dimension:",
    VECTOR_DIMENSION
)


# ============================================================
# 11. VALIDATE CHUNK STRUCTURE
# ============================================================

required_chunk_fields = [

    "chunk_id",
    "chunk_order",
    "text",
    "heading",
    "level",
    "page_start",
    "page_end",
    "word_count",

]


for chunk in final_chunks:

    for field in required_chunk_fields:

        assert field in chunk, (
            f"STOP: Missing chunk field: {field}"
        )


assert all(
    chunk["chunk_id"] == index
    for index, chunk in enumerate(final_chunks)
), (
    "STOP: Final chunk IDs are not sequential."
)


assert all(
    chunk["chunk_order"] == index
    for index, chunk in enumerate(final_chunks)
), (
    "STOP: Final chunk order is not sequential."
)


assert all(
    isinstance(chunk["text"], str)
    and chunk["text"].strip()
    for chunk in final_chunks
), (
    "STOP: One or more final chunks have empty text."
)


print()
print("Final chunk structure: PASS")
print("Final chunk IDs: PASS")
print("Final chunk order: PASS")


# ============================================================
# 12. VERIFY CHUNKS / EMBEDDINGS ALIGNMENT
# ============================================================

assert len(final_chunks) == len(
    final_embeddings
), (
    "STOP: Number of chunks does not match "
    "number of embeddings."
)


print(
    "Chunks / embeddings alignment: PASS"
)


# ============================================================
# 13. L2 NORMALIZATION
# ============================================================

embedding_norms = np.linalg.norm(
    final_embeddings,
    axis=1
)


assert np.all(
    embedding_norms > 0
), (
    "STOP: One or more embeddings have zero norm."
)


normalization_errors = int(
    np.sum(
        np.abs(
            embedding_norms - 1.0
        ) > 1e-4
    )
)


if normalization_errors == 0:

    print(
        "Final embeddings already L2-normalized: PASS"
    )

else:

    print(
        "Final embeddings are not L2-normalized."
    )

    print(
        "Applying L2 normalization..."
    )

    final_embeddings = (
        final_embeddings
        / embedding_norms[:, None]
    ).astype(
        np.float32
    )


    final_norms = np.linalg.norm(
        final_embeddings,
        axis=1
    )


    assert np.all(
        np.abs(
            final_norms - 1.0
        ) < 1e-4
    ), (
        "STOP: Final embedding normalization failed."
    )


    print(
        "Final embedding normalization: PASS"
    )


# ============================================================
# 14. PREPARE FINAL MONGODB DOCUMENTS
# ============================================================

mongo_documents = []


for chunk, vector in zip(
    final_chunks,
    final_embeddings
):

    document = {

        "chunk_id": int(
            chunk["chunk_id"]
        ),

        "text": chunk["text"],

        "heading": chunk["heading"],

        "level": chunk["level"],

        "page_start": int(
            chunk["page_start"]
        ),

        "page_end": int(
            chunk["page_end"]
        ),

        "chunk_order": int(
            chunk["chunk_order"]
        ),

        "word_count": int(
            chunk["word_count"]
        ),

        VECTOR_FIELD: np.asarray(
            vector,
            dtype=np.float32
        ).tolist(),

    }

    mongo_documents.append(
        document
    )


assert len(
    mongo_documents
) == EXPECTED_FINAL_CHUNKS


print()
print(
    "MongoDB documents prepared:",
    len(mongo_documents)
)


# ============================================================
# 15. VALIDATE PREPARED DOCUMENTS BEFORE DATABASE CHANGE
# ============================================================

for document in mongo_documents:

    assert VECTOR_FIELD in document

    assert len(
        document[VECTOR_FIELD]
    ) == VECTOR_DIMENSION

    assert np.isfinite(
        np.asarray(
            document[VECTOR_FIELD],
            dtype=np.float32
        )
    ).all()


print(
    "Prepared MongoDB documents: PASS"
)


# ============================================================
# 16. SHOW CURRENT DATABASE STATE
# ============================================================

current_count = collection.count_documents({})


print()
print("=" * 70)
print("CURRENT MONGODB STATE")
print("=" * 70)

print(
    "Current stored documents:",
    current_count
)

print(
    "Expected final documents:",
    EXPECTED_FINAL_CHUNKS
)


# ============================================================
# 17. REMOVE OLD BASELINE DATA
# ============================================================

print()
print(
    "Removing previous MongoDB dataset..."
)


delete_result = collection.delete_many({})


print(
    "Previous documents removed:",
    delete_result.deleted_count
)


# ============================================================
# 18. INSERT FINAL 248 DOCUMENTS
# ============================================================

print()
print(
    "Uploading final 248 chunks..."
)


insert_result = collection.insert_many(
    mongo_documents
)


assert len(
    insert_result.inserted_ids
) == EXPECTED_FINAL_CHUNKS


print(
    "Documents inserted:",
    len(insert_result.inserted_ids)
)


# ============================================================
# 19. VALIDATE DATABASE COUNT
# ============================================================

stored_count = collection.count_documents({})


assert stored_count == EXPECTED_FINAL_CHUNKS, (
    f"STOP: MongoDB contains {stored_count} documents, "
    f"expected {EXPECTED_FINAL_CHUNKS}."
)


print(
    "MongoDB document count: PASS"
)


# ============================================================
# 20. VALIDATE CHUNK IDS
# ============================================================

stored_ids = sorted(

    document["chunk_id"]

    for document in collection.find(
        {},
        {
            "_id": 0,
            "chunk_id": 1
        }
    )

)


expected_ids = list(
    range(
        EXPECTED_FINAL_CHUNKS
    )
)


assert stored_ids == expected_ids, (
    "STOP: MongoDB chunk IDs do not match "
    "the final 248-chunk dataset."
)


print(
    "Stored chunk IDs: PASS"
)


# ============================================================
# 21. VALIDATE STORED DOCUMENT
# ============================================================

sample = collection.find_one(
    {
        "chunk_id": 0
    }
)


assert sample is not None, (
    "STOP: Could not retrieve chunk_id=0."
)


for field in required_chunk_fields:

    assert field in sample, (
        f"STOP: Missing stored field: {field}"
    )


assert VECTOR_FIELD in sample, (
    "STOP: Stored document missing embedding."
)


assert len(
    sample[VECTOR_FIELD]
) == VECTOR_DIMENSION, (
    "STOP: Stored embedding dimension mismatch."
)


stored_vector = np.asarray(
    sample[VECTOR_FIELD],
    dtype=np.float32
)


assert np.isfinite(
    stored_vector
).all(), (
    "STOP: Stored embedding contains "
    "NaN or infinite values."
)


print(
    "Stored document structure: PASS"
)

print(
    "Stored embedding dimension: PASS"
)


# ============================================================
# 22. VALIDATE STORED VECTOR NORMALIZATION
# ============================================================

stored_norm = np.linalg.norm(
    stored_vector
)


assert abs(
    stored_norm - 1.0
) < 1e-4, (
    "STOP: Stored embedding is not L2-normalized."
)


print(
    "Stored embedding normalization: PASS"
)


# ============================================================
# 23. SAVE FINAL MONGODB CHECKPOINT
# ============================================================

mongodb_final_checkpoint = {

    "database": DATABASE_NAME,

    "collection": COLLECTION_NAME,

    "vector_index_name": INDEX_NAME,

    "vector_dimension": VECTOR_DIMENSION,

    "stored_documents": stored_count,

    "status": "PASS",

    "dataset": "final",

    "chunk_configuration": "Small",

    "target_size": EXPECTED_TARGET_SIZE,

    "overlap": EXPECTED_OVERLAP,

    "embedding_model": EXPECTED_MODEL_NAME,

}


MONGODB_FINAL_CHECKPOINT = os.path.join(
    CHECKPOINT_DIR,
    "mongodb_final_upload_checkpoint.pkl"
)


with open(
    MONGODB_FINAL_CHECKPOINT,
    "wb"
) as f:

    pickle.dump(
        mongodb_final_checkpoint,
        f
    )


assert os.path.exists(
    MONGODB_FINAL_CHECKPOINT
)


# ============================================================
# 24. FINAL REPORT
# ============================================================

print()
print("=" * 70)
print("FINAL MONGODB DATASET")
print("=" * 70)

print()
print(
    "Database:",
    DATABASE_NAME
)

print(
    "Collection:",
    COLLECTION_NAME
)

print(
    "Configuration: Small"
)

print(
    "Target size:",
    EXPECTED_TARGET_SIZE
)

print(
    "Maximum overlap:",
    EXPECTED_OVERLAP
)

print(
    "Final chunks:",
    EXPECTED_FINAL_CHUNKS
)

print(
    "Stored documents:",
    stored_count
)

print(
    "Embedding model:",
    EXPECTED_MODEL_NAME
)

print(
    "Embedding dimension:",
    VECTOR_DIMENSION
)

print(
    "Vector field:",
    VECTOR_FIELD
)

print(
    "Vector index:",
    INDEX_NAME
)

print()
print(
    "Chunk IDs: PASS"
)

print(
    "Chunk metadata: PASS"
)

print(
    "Embedding matrix: PASS"
)

print(
    "MongoDB document count: PASS"
)

print(
    "Stored embedding: PASS"
)

print(
    "Stored embedding normalization: PASS"
)

print()
print(
    "Final MongoDB checkpoint:"
)

print(
    MONGODB_FINAL_CHECKPOINT
)

print()
print("=" * 70)
print("CELL 22 — FINAL MONGODB DATASET UPLOAD: PASS")
print("=" * 70)
````

````python
# ============================================================
# CELL 23 — FINAL VECTOR SEARCH VALIDATION
# ============================================================

import numpy as np


print("=" * 70)
print("CELL 23 — FINAL VECTOR SEARCH VALIDATION")
print("=" * 70)


# ============================================================
# CONFIGURATION
# ============================================================

VECTOR_FIELD = "embedding"
VECTOR_DIMENSION = 1024
INDEX_NAME = "chunk_vector_index"

TOP_K = 5
NUM_CANDIDATES = 50

EXPECTED_FINAL_CHUNKS = 248


# ============================================================
# TEST QUERIES
# ============================================================

FINAL_TEST_QUERIES = [

    "What is self-help?",

    "Who is self-help intended for?",

    "When should guided self-help be used?",

    (
        "What should a helper do when someone is "
        "at risk of suicide?"
    ),

    (
        "What are the Step-by-Step skills "
        "for managing stress?"
    ),

    "What are the Doing What Matters modules?",
]


# ============================================================
# SAFETY CHECKS
# ============================================================

assert "embedding_model" in globals(), (
    "STOP: embedding_model is not defined. "
    "Run Cell 8 first."
)

assert callable(
    embedding_model.encode
), (
    "STOP: embedding_model is not usable."
)


assert "collection" in globals(), (
    "STOP: MongoDB collection not found. "
    "Run Cell 11 first."
)


# ============================================================
# VERIFY FINAL DATABASE SIZE
# ============================================================

stored_count = collection.count_documents({})


assert stored_count == EXPECTED_FINAL_CHUNKS, (
    f"STOP: Expected {EXPECTED_FINAL_CHUNKS} "
    f"documents in MongoDB, got {stored_count}."
)


print()
print(
    "MongoDB final document count:",
    stored_count
)

print(
    "Final dataset count: PASS"
)


# ============================================================
# VERIFY VECTOR DIMENSION IN DATABASE
# ============================================================

sample = collection.find_one(
    {},
    {
        "_id": 0,
        "chunk_id": 1,
        "embedding": 1
    }
)


assert sample is not None, (
    "STOP: MongoDB contains no documents."
)


assert VECTOR_FIELD in sample, (
    "STOP: Embedding field not found."
)


assert len(
    sample[VECTOR_FIELD]
) == VECTOR_DIMENSION, (
    f"STOP: Stored vector dimension is "
    f"{len(sample[VECTOR_FIELD])}, "
    f"expected {VECTOR_DIMENSION}."
)


print(
    "Stored vector dimension:",
    len(sample[VECTOR_FIELD])
)

print(
    "Vector dimension: PASS"
)


# ============================================================
# FINAL VECTOR SEARCH FUNCTION
# ============================================================

def final_vector_search(
    query,
    top_k=TOP_K,
    num_candidates=NUM_CANDIDATES
):

    assert isinstance(query, str), (
        "STOP: Query must be a string."
    )

    assert query.strip(), (
        "STOP: Query cannot be empty."
    )

    assert top_k > 0, (
        "STOP: top_k must be greater than zero."
    )

    assert num_candidates >= top_k, (
        "STOP: num_candidates must be >= top_k."
    )


    # --------------------------------------------------------
    # Generate BGE-M3 query embedding
    # --------------------------------------------------------

    query_vector = embedding_model.encode(
        query,
        convert_to_numpy=True
    )


    query_vector = np.asarray(
        query_vector,
        dtype=np.float32
    )


    # --------------------------------------------------------
    # Validate query vector
    # --------------------------------------------------------

    assert query_vector.ndim == 1, (
        f"STOP: Query vector must be 1D, "
        f"got {query_vector.shape}."
    )


    assert len(query_vector) == VECTOR_DIMENSION, (
        f"STOP: Query vector has "
        f"{len(query_vector)} dimensions, "
        f"expected {VECTOR_DIMENSION}."
    )


    assert np.isfinite(
        query_vector
    ).all(), (
        "STOP: Query vector contains "
        "NaN or infinite values."
    )


    # --------------------------------------------------------
    # L2 normalize query
    # --------------------------------------------------------

    query_norm = np.linalg.norm(
        query_vector
    )


    assert query_norm > 0, (
        "STOP: Query vector has zero norm."
    )


    query_vector = (
        query_vector / query_norm
    ).astype(
        np.float32
    )


    # --------------------------------------------------------
    # Verify normalization
    # --------------------------------------------------------

    final_query_norm = np.linalg.norm(
        query_vector
    )


    assert abs(
        final_query_norm - 1.0
    ) < 1e-4, (
        "STOP: Query normalization failed."
    )


    # --------------------------------------------------------
    # MongoDB Atlas Vector Search
    # --------------------------------------------------------

    pipeline = [

        {
            "$vectorSearch": {

                "index": INDEX_NAME,

                "path": VECTOR_FIELD,

                "queryVector":
                    query_vector.tolist(),

                "numCandidates":
                    num_candidates,

                "limit":
                    top_k,
            }
        },

        {
            "$project": {

                "_id": 0,

                "chunk_id": 1,

                "text": 1,

                "heading": 1,

                "level": 1,

                "page_start": 1,

                "page_end": 1,

                "chunk_order": 1,

                "word_count": 1,

                "score": {
                    "$meta":
                        "vectorSearchScore"
                },
            }
        },
    ]


    results = list(
        collection.aggregate(
            pipeline
        )
    )


    # --------------------------------------------------------
    # Validate results
    # --------------------------------------------------------

    assert len(results) > 0, (
        "STOP: Vector search returned zero results."
    )


    assert len(results) <= top_k, (
        "STOP: Vector search returned more "
        "than TOP_K results."
    )


    for result in results:

        assert "chunk_id" in result
        assert "text" in result
        assert "heading" in result
        assert "page_start" in result
        assert "page_end" in result
        assert "score" in result


        assert isinstance(
            result["chunk_id"],
            int
        ), (
            "STOP: chunk_id must be an integer."
        )


        assert isinstance(
            result["score"],
            (int, float)
        ), (
            "STOP: Vector search score "
            "must be numeric."
        )


        assert np.isfinite(
            result["score"]
        ), (
            "STOP: Vector search returned "
            "an invalid score."
        )


    return results


# ============================================================
# RUN FINAL RETRIEVAL TESTS
# ============================================================

final_retrieval_results = {}


print()
print("=" * 70)
print("FINAL VECTOR SEARCH RESULTS")
print("=" * 70)


for query_number, query in enumerate(
    FINAL_TEST_QUERIES,
    start=1
):

    print()
    print(
        f"QUERY {query_number}/"
        f"{len(FINAL_TEST_QUERIES)}"
    )

    print("-" * 70)

    print(
        "Question:"
    )

    print(
        query
    )


    results = final_vector_search(
        query=query,
        top_k=TOP_K,
        num_candidates=NUM_CANDIDATES
    )


    final_retrieval_results[
        query
    ] = results


    print()

    for rank, result in enumerate(
        results,
        start=1
    ):

        print(
            f"Rank {rank} | "
            f"Chunk {result['chunk_id']} | "
            f"Score: {result['score']:.6f}"
        )

        print(
            f"Heading: "
            f"{result['heading']}"
        )

        print(
            f"Pages: "
            f"{result['page_start']}-"
            f"{result['page_end']}"
        )

        print(
            f"Words: "
            f"{result['word_count']}"
        )


# ============================================================
# FINAL VALIDATION
# ============================================================

assert (
    len(final_retrieval_results)
    == len(FINAL_TEST_QUERIES)
)


assert all(
    len(results) > 0
    for results
    in final_retrieval_results.values()
)


# ============================================================
# FINAL REPORT
# ============================================================

print()
print("=" * 70)
print("FINAL RETRIEVAL VALIDATION SUMMARY")
print("=" * 70)

print(
    "Final chunks:",
    EXPECTED_FINAL_CHUNKS
)

print(
    "Embedding model:",
    "BAAI/bge-m3"
)

print(
    "Embedding dimension:",
    VECTOR_DIMENSION
)

print(
    "Vector index:",
    INDEX_NAME
)

print(
    "Top-K:",
    TOP_K
)

print(
    "Num candidates:",
    NUM_CANDIDATES
)

print(
    "Queries tested:",
    len(FINAL_TEST_QUERIES)
)

print("-" * 70)

print(
    "MongoDB dataset: PASS"
)

print(
    "Vector dimension: PASS"
)

print(
    "Query embedding generation: PASS"
)

print(
    "Query normalization: PASS"
)

print(
    "MongoDB vector search: PASS"
)

print(
    "All final queries returned results: PASS"
)

print("=" * 70)
print("CELL 23 — FINAL VECTOR SEARCH VALIDATION: PASS")
print("=" * 70)
````

````python
# ============================================================
# CELL 24 — FINAL RETRIEVAL PRECISION@5
# RECOVERY-SAFE VERSION
# ============================================================

import os
import pickle
import numpy as np


print("=" * 70)
print("CELL 24 — FINAL RETRIEVAL PRECISION@5")
print("RECOVERY-SAFE VERSION")
print("=" * 70)


# ============================================================
# CONFIGURATION
# ============================================================

CHECKPOINT_DIR = (
    "/content/drive/MyDrive/RAG_Project/checkpoints"
)

TOP_K = 5

EXPECTED_FINAL_CHUNKS = 248

VECTOR_FIELD = "embedding"
VECTOR_DIMENSION = 1024
INDEX_NAME = "chunk_vector_index"

NUM_CANDIDATES = 50

EMBEDDING_MODEL_NAME = "BAAI/bge-m3"


# ============================================================
# FINAL EVALUATION SET
# ============================================================

FINAL_EVALUATION = [

    {
        "id": 1,
        "question": "What is self-help?",
        "expected_chunks": [7],
    },

    {
        "id": 2,
        "question": "Who is self-help intended for?",
        "expected_chunks": [12, 13],
    },

    {
        "id": 3,
        "question": "When should guided self-help be used?",
        "expected_chunks": [14],
    },

    {
        "id": 4,
        "question": (
            "What should a helper do when someone is "
            "at risk of suicide?"
        ),
        "expected_chunks": [47, 82],
    },

    {
        "id": 5,
        "question": (
            "What are the Step-by-Step skills "
            "for managing stress?"
        ),
        "expected_chunks": [86, 89, 93],
    },

    {
        "id": 6,
        "question": (
            "What are the Doing What Matters modules?"
        ),
        "expected_chunks": [104, 105, 106, 108, 114],
    },
]


# ============================================================
# RECOVERY — MONGODB CONNECTION
# ============================================================

if "collection" in globals():

    print()
    print("MongoDB collection object: FOUND")

else:

    print()
    print(
        "MongoDB collection object: NOT FOUND"
    )

    print(
        "Attempting recovery from existing MongoDB "
        "client/database objects..."
    )


    if "db" in globals():

        collection = db["chunks"]

        print(
            "Collection recovered from existing db object."
        )

    elif "client" in globals():

        db = client["psychological_self_help"]

        collection = db["chunks"]

        print(
            "Database and collection recovered "
            "from existing client object."
        )

    else:

        raise RuntimeError(
            "STOP: MongoDB connection objects are not "
            "available in memory.\n\n"
            "Run the MongoDB connection portion of "
            "Cell 11 or Cell 22 first."
        )


print(
    "Database: psychological_self_help"
)

print(
    "Collection: chunks"
)


# ============================================================
# RECOVERY — EMBEDDING MODEL
# ============================================================

if "embedding_model" in globals():

    print()
    print(
        "Embedding model object: FOUND"
    )

else:

    print()
    print(
        "Embedding model object: NOT FOUND"
    )

    print(
        "Attempting recovery..."
    )


    try:

        from sentence_transformers import SentenceTransformer


        embedding_model = SentenceTransformer(
            EMBEDDING_MODEL_NAME
        )


        print(
            "Embedding model recovered successfully."
        )

    except Exception as e:

        raise RuntimeError(
            "STOP: Could not recover BGE-M3 embedding model.\n"
            "Run the embedding-model loading cell first."
        ) from e


assert callable(
    embedding_model.encode
), (
    "STOP: Embedding model is not usable."
)


# ============================================================
# VERIFY FINAL MONGODB DATASET
# ============================================================

stored_count = collection.count_documents({})


assert stored_count == EXPECTED_FINAL_CHUNKS, (

    f"STOP: Expected {EXPECTED_FINAL_CHUNKS} "
    f"documents in MongoDB, "

    f"got {stored_count}."
)


print()
print(
    "Final MongoDB documents:",
    stored_count
)

print(
    "Final dataset size: PASS"
)


# ============================================================
# VERIFY STORED VECTOR
# ============================================================

sample = collection.find_one(
    {},
    {
        "_id": 0,
        "chunk_id": 1,
        "embedding": 1
    }
)


assert sample is not None, (
    "STOP: MongoDB contains no documents."
)


assert VECTOR_FIELD in sample, (
    "STOP: Embedding field not found."
)


assert len(
    sample[VECTOR_FIELD]
) == VECTOR_DIMENSION, (

    f"STOP: Stored vector dimension is "
    f"{len(sample[VECTOR_FIELD])}, "

    f"expected {VECTOR_DIMENSION}."
)


print(
    "Stored vector dimension:",
    len(sample[VECTOR_FIELD])
)

print(
    "Vector dimension: PASS"
)


# ============================================================
# RECOVER BASELINE GOLD DATA
#
# IMPORTANT:
#
# We intentionally DO NOT use:
#
#     chunks
#
# from the current notebook RAM.
#
# Instead, we search checkpoint files for a saved baseline
# chunk list.
# ============================================================

print()
print("=" * 70)
print("RECOVERING BASELINE GOLD DATA")
print("=" * 70)


assert os.path.exists(
    CHECKPOINT_DIR
), (

    "STOP: Checkpoint directory does not exist:\n"
    f"{CHECKPOINT_DIR}"
)


checkpoint_files = sorted(
    os.path.join(
        CHECKPOINT_DIR,
        filename
    )
    for filename in os.listdir(
        CHECKPOINT_DIR
    )
    if filename.lower().endswith(
        ".pkl"
    )
)


print()
print(
    "Checkpoint files found:",
    len(checkpoint_files)
)


for filepath in checkpoint_files:

    print(
        "  -",
        os.path.basename(filepath)
    )


# ============================================================
# HELPER — FIND CHUNK LIST INSIDE CHECKPOINT
# ============================================================

def find_chunk_list(
    obj,
    path="root"
):

    # --------------------------------------------------------
    # Direct list
    # --------------------------------------------------------

    if isinstance(obj, list):

        if len(obj) == 0:

            return None


        if all(
            isinstance(item, dict)
            for item in obj
        ):

            required_fields = {
                "chunk_id",
                "heading",
                "text",
            }


            if required_fields.issubset(
                obj[0].keys()
            ):

                return obj


    # --------------------------------------------------------
    # Dictionary
    # --------------------------------------------------------

    if isinstance(obj, dict):

        for key, value in obj.items():

            result = find_chunk_list(
                value,
                f"{path}.{key}"
            )

            if result is not None:

                return result


    # --------------------------------------------------------
    # Tuple
    # --------------------------------------------------------

    if isinstance(obj, tuple):

        for index, value in enumerate(obj):

            result = find_chunk_list(
                value,
                f"{path}[{index}]"
            )

            if result is not None:

                return result


    return None


# ============================================================
# SEARCH CHECKPOINTS
# ============================================================

baseline_chunk_candidates = []


for filepath in checkpoint_files:

    try:

        with open(
            filepath,
            "rb"
        ) as f:

            checkpoint_data = pickle.load(f)


        recovered_chunks = find_chunk_list(
            checkpoint_data
        )


        if recovered_chunks is None:

            continue


        # ----------------------------------------------------
        # Validate candidate
        # ----------------------------------------------------

        valid_candidate = True


        for chunk in recovered_chunks:

            required_fields = {
                "chunk_id",
                "heading",
                "text",
            }


            if not required_fields.issubset(
                chunk.keys()
            ):

                valid_candidate = False

                break


        if not valid_candidate:

            continue


        baseline_chunk_candidates.append(
            (
                filepath,
                recovered_chunks
            )
        )


    except Exception:

        continue


# ============================================================
# DISPLAY CANDIDATES
# ============================================================

print()
print(
    "Checkpoint chunk-list candidates:"
)


for filepath, recovered_chunks in (
    baseline_chunk_candidates
):

    print(
        "  -",
        os.path.basename(filepath),
        "| chunks:",
        len(recovered_chunks)
    )


# ============================================================
# SELECT BASELINE CHECKPOINT
# ============================================================

# Prefer a checkpoint that looks like the original
# baseline dataset rather than the final 248-chunk dataset.

baseline_candidates = [

    candidate

    for candidate in baseline_chunk_candidates

    if len(candidate[1]) != EXPECTED_FINAL_CHUNKS

]


assert len(
    baseline_candidates
) > 0, (

    "STOP: Could not recover the baseline chunk dataset "
    "from the available checkpoints.\n\n"

    "The Precision@5 evaluation requires the original "
    "baseline chunk IDs to be mapped to semantic headings.\n"

    "No baseline chunk checkpoint was found.\n\n"

    "Do NOT use final 248 chunks as the gold labels, "
    "because that would create a circular evaluation."
)


# ============================================================
# SELECT MOST LIKELY BASELINE
# ============================================================

# Prefer the largest non-final chunk dataset.
#
# This is expected to be the original 500/75 dataset.

baseline_candidates = sorted(
    baseline_candidates,
    key=lambda item: len(item[1]),
    reverse=True
)


baseline_checkpoint_path, baseline_chunks = (
    baseline_candidates[0]
)


print()
print(
    "Selected baseline checkpoint:"
)

print(
    baseline_checkpoint_path
)

print(
    "Recovered baseline chunks:",
    len(baseline_chunks)
)


# ============================================================
# BASELINE STRUCTURE VALIDATION
# ============================================================

assert isinstance(
    baseline_chunks,
    list
), (
    "STOP: Recovered baseline chunks are not a list."
)


assert len(
    baseline_chunks
) > 0, (
    "STOP: Recovered baseline chunks are empty."
)


for chunk in baseline_chunks:

    assert "chunk_id" in chunk

    assert "heading" in chunk

    assert "text" in chunk


baseline_ids = [

    int(
        chunk["chunk_id"]
    )

    for chunk in baseline_chunks

]


assert len(
    set(baseline_ids)
) == len(
    baseline_ids
), (
    "STOP: Baseline chunk IDs are not unique."
)


print(
    "Baseline chunk structure: PASS"
)


# ============================================================
# BUILD GOLD SEMANTIC HEADINGS
#
# Original expected IDs belong to the baseline dataset.
# We map those IDs to headings.
#
# These headings are then used to evaluate the final
# 300/50 retrieval results.
# ============================================================

expected_headings_by_question = []


baseline_chunk_by_id = {

    int(
        chunk["chunk_id"]
    ): chunk

    for chunk in baseline_chunks

}


for item in FINAL_EVALUATION:

    expected_ids = item[
        "expected_chunks"
    ]


    headings = set()


    for chunk_id in expected_ids:

        assert chunk_id in (
            baseline_chunk_by_id
        ), (

            f"STOP: Expected baseline chunk "
            f"{chunk_id} was not found in the "
            f"recovered baseline checkpoint."
        )


        heading = baseline_chunk_by_id[
            chunk_id
        ][
            "heading"
        ]


        assert isinstance(
            heading,
            str
        ), (

            f"STOP: Baseline chunk {chunk_id} "
            "has invalid heading."
        )


        headings.add(
            heading
        )


    assert len(
        headings
    ) > 0, (

        f"STOP: No gold headings recovered "
        f"for evaluation question {item['id']}."
    )


    expected_headings_by_question.append(
        headings
    )


print()
print(
    "Gold semantic heading mapping: PASS"
)


# ============================================================
# DISPLAY GOLD MAPPING
# ============================================================

print()
print("=" * 70)
print("GOLD SEMANTIC TARGETS")
print("=" * 70)


for index, item in enumerate(
    FINAL_EVALUATION
):

    print()
    print(
        f"Q{item['id']}: "
        f"{item['question']}"
    )

    print(
        "Expected baseline chunks:",
        item["expected_chunks"]
    )

    print(
        "Expected headings:"
    )


    for heading in sorted(
        expected_headings_by_question[index]
    ):

        print(
            f"  - {heading}"
        )


# ============================================================
# FINAL VECTOR SEARCH FUNCTION
# ============================================================

def final_precision_search(
    query,
    top_k=TOP_K
):

    assert isinstance(
        query,
        str
    ), (
        "STOP: Query must be a string."
    )


    assert query.strip(), (
        "STOP: Query cannot be empty."
    )


    # --------------------------------------------------------
    # Generate BGE-M3 query embedding
    # --------------------------------------------------------

    query_vector = embedding_model.encode(
        query,
        convert_to_numpy=True
    )


    query_vector = np.asarray(
        query_vector,
        dtype=np.float32
    )


    # --------------------------------------------------------
    # Validate query embedding
    # --------------------------------------------------------

    assert query_vector.ndim == 1, (

        f"STOP: Query embedding must be 1D, "
        f"got {query_vector.shape}."
    )


    assert len(
        query_vector
    ) == VECTOR_DIMENSION, (

        f"STOP: Expected "
        f"{VECTOR_DIMENSION}-dimensional query."
    )


    assert np.isfinite(
        query_vector
    ).all(), (

        "STOP: Query embedding contains "
        "NaN or infinite values."
    )


    # --------------------------------------------------------
    # L2 normalization
    # --------------------------------------------------------

    query_norm = np.linalg.norm(
        query_vector
    )


    assert query_norm > 0, (
        "STOP: Query embedding has zero norm."
    )


    query_vector = (

        query_vector
        / query_norm

    ).astype(
        np.float32
    )


    final_norm = np.linalg.norm(
        query_vector
    )


    assert abs(
        final_norm - 1.0
    ) < 1e-4, (

        "STOP: Query normalization failed."
    )


    # --------------------------------------------------------
    # MongoDB Atlas Vector Search
    # --------------------------------------------------------

    pipeline = [

        {
            "$vectorSearch": {

                "index":
                    INDEX_NAME,

                "path":
                    VECTOR_FIELD,

                "queryVector":
                    query_vector.tolist(),

                "numCandidates":
                    NUM_CANDIDATES,

                "limit":
                    top_k,
            }
        },

        {
            "$project": {

                "_id": 0,

                "chunk_id": 1,

                "heading": 1,

                "page_start": 1,

                "page_end": 1,

                "word_count": 1,

                "score": {
                    "$meta":
                        "vectorSearchScore"
                },
            }
        },
    ]


    results = list(
        collection.aggregate(
            pipeline
        )
    )


    assert len(
        results
    ) > 0, (
        "STOP: Vector search returned zero results."
    )


    assert len(
        results
    ) <= top_k, (
        "STOP: More than TOP_K results returned."
    )


    for result in results:

        assert "chunk_id" in result

        assert "heading" in result

        assert "score" in result


        assert isinstance(
            result["chunk_id"],
            int
        ), (
            "STOP: chunk_id must be integer."
        )


        assert isinstance(
            result["score"],
            (int, float)
        ), (
            "STOP: Search score must be numeric."
        )


        assert np.isfinite(
            result["score"]
        ), (
            "STOP: Search score is invalid."
        )


    return results


# ============================================================
# RUN FINAL PRECISION@5
# ============================================================

final_precision_results = []

total_hits = 0

total_precision = 0.0

heading_query_hits = 0


print()
print("=" * 70)
print("FINAL PRECISION@5 RESULTS")
print("=" * 70)


for question_index, item in enumerate(
    FINAL_EVALUATION
):

    question = item[
        "question"
    ]


    expected_headings = (
        expected_headings_by_question[
            question_index
        ]
    )


    retrieved = final_precision_search(
        question,
        TOP_K
    )


    retrieved_ids = [

        int(
            result["chunk_id"]
        )

        for result in retrieved

    ]


    retrieved_headings = [

        result["heading"]

        for result in retrieved

    ]


    # --------------------------------------------------------
    # Heading-level relevance
    # --------------------------------------------------------

    heading_hits = [

        heading

        for heading in retrieved_headings

        if heading in expected_headings

    ]


    unique_heading_hits = set(
        heading_hits
    )


    relevant_count = len(
        heading_hits
    )


    precision_at_5 = (

        relevant_count
        / TOP_K

    )


    total_precision += (
        precision_at_5
    )


    total_hits += (
        relevant_count
    )


    if unique_heading_hits:

        heading_query_hits += 1


    # --------------------------------------------------------
    # Store result
    # --------------------------------------------------------

    result_record = {

        "question":
            question,

        "expected_baseline_chunks":
            item["expected_chunks"],

        "expected_headings":
            sorted(
                expected_headings
            ),

        "retrieved_chunks":
            retrieved_ids,

        "retrieved_headings":
            retrieved_headings,

        "heading_hits":
            sorted(
                unique_heading_hits
            ),

        "relevant_count":
            relevant_count,

        "precision_at_5":
            precision_at_5,

    }


    final_precision_results.append(
        result_record
    )


    # --------------------------------------------------------
    # Display
    # --------------------------------------------------------

    print()
    print(
        f"Q{item['id']}: "
        f"{question}"
    )


    print()
    print(
        "Expected headings:"
    )


    for heading in sorted(
        expected_headings
    ):

        print(
            f"  - {heading}"
        )


    print()
    print(
        "Retrieved chunks:",
        retrieved_ids
    )


    print(
        "Retrieved headings:"
    )


    for rank, result in enumerate(
        retrieved,
        start=1
    ):

        print(

            f"  {rank}. "
            f"Chunk {result['chunk_id']} | "
            f"{result['heading']} | "
            f"Score "
            f"{result['score']:.6f}"

        )


    print()
    print(
        "Relevant retrieved chunks:",
        relevant_count
    )


    print(
        f"Precision@5: "
        f"{precision_at_5:.3f}"
    )


# ============================================================
# OVERALL METRICS
# ============================================================

mean_precision_at_5 = (

    total_precision
    / len(FINAL_EVALUATION)

)


heading_hit_rate = (

    heading_query_hits
    / len(FINAL_EVALUATION)

)


average_relevant_chunks = (

    total_hits
    / len(FINAL_EVALUATION)

)


# ============================================================
# VALIDATION
# ============================================================

assert (
    0.0
    <= mean_precision_at_5
    <= 1.0
)


assert (
    0.0
    <= heading_hit_rate
    <= 1.0
)


assert len(
    final_precision_results
) == len(
    FINAL_EVALUATION
)


# ============================================================
# SAVE PRECISION CHECKPOINT
# ============================================================

precision_checkpoint = {

    "status":
        "completed",

    "embedding_model":
        EMBEDDING_MODEL_NAME,

    "embedding_dimension":
        VECTOR_DIMENSION,

    "final_chunks":
        EXPECTED_FINAL_CHUNKS,

    "target_size":
        300,

    "overlap":
        50,

    "top_k":
        TOP_K,

    "num_candidates":
        NUM_CANDIDATES,

    "baseline_checkpoint":
        baseline_checkpoint_path,

    "baseline_chunk_count":
        len(baseline_chunks),

    "evaluation":
        FINAL_EVALUATION,

    "expected_headings":
        [
            sorted(headings)
            for headings
            in expected_headings_by_question
        ],

    "results":
        final_precision_results,

    "mean_precision_at_5":
        mean_precision_at_5,

    "heading_hit_rate":
        heading_hit_rate,

    "average_relevant_chunks":
        average_relevant_chunks,

    "total_relevant_retrieved":
        total_hits,

}


PRECISION_CHECKPOINT_PATH = os.path.join(

    CHECKPOINT_DIR,

    "final_retrieval_precision_at_5_checkpoint.pkl"

)


with open(
    PRECISION_CHECKPOINT_PATH,
    "wb"
) as f:

    pickle.dump(
        precision_checkpoint,
        f
    )


assert os.path.exists(
    PRECISION_CHECKPOINT_PATH
), (
    "STOP: Precision checkpoint was not saved."
)


# ============================================================
# FINAL SUMMARY
# ============================================================

print()
print("=" * 70)
print("FINAL BGE-M3 PRECISION@5 SUMMARY")
print("=" * 70)


print(
    "Embedding model:",
    EMBEDDING_MODEL_NAME
)


print(
    "Final chunk configuration:",
    "300 words / 50 overlap"
)


print(
    "Final chunks:",
    EXPECTED_FINAL_CHUNKS
)


print(
    "Embedding dimension:",
    VECTOR_DIMENSION
)


print(
    "Baseline chunks recovered:",
    len(baseline_chunks)
)


print(
    "Baseline checkpoint:",
    baseline_checkpoint_path
)


print(
    "Queries evaluated:",
    len(FINAL_EVALUATION)
)


print(
    "Top-K:",
    TOP_K
)


print(
    "Num candidates:",
    NUM_CANDIDATES
)


print(
    f"Mean Precision@5: "
    f"{mean_precision_at_5:.4f}"
)


print(
    f"Heading Hit Rate: "
    f"{heading_hit_rate:.4f}"
)


print(
    f"Average relevant chunks/query: "
    f"{average_relevant_chunks:.2f}"
)


print(
    f"Total relevant retrieved chunks: "
    f"{total_hits}"
)


print()
print(
    "Precision checkpoint:"
)

print(
    PRECISION_CHECKPOINT_PATH
)


print("-" * 70)

print(
    "MongoDB dataset: PASS"
)

print(
    "Vector dimension: PASS"
)

print(
    "BGE-M3 query embedding: PASS"
)

print(
    "Query normalization: PASS"
)

print(
    "MongoDB vector search: PASS"
)

print(
    "Gold semantic mapping: PASS"
)

print(
    "Precision@5 calculation: PASS"
)

print(
    "Precision checkpoint: SAVED"
)

print("=" * 70)
print(
    "CELL 24 — FINAL RETRIEVAL PRECISION@5: PASS"
)
print("=" * 70)
````

````python
# ============================================================
# CELL 25 — FINAL RETRIEVAL DIAGNOSTICS
# RECOVERY-SAFE VERSION
# ============================================================

import os
import pickle


print("=" * 70)
print("CELL 25 — FINAL RETRIEVAL DIAGNOSTICS")
print("RECOVERY-SAFE VERSION")
print("=" * 70)


# ============================================================
# CONFIGURATION
# ============================================================

EXPECTED_FINAL_CHUNKS = 248

MIN_REASONABLE_WORDS = 20

VECTOR_FIELD = "embedding"

CHECKPOINT_DIR = (
    "/content/drive/MyDrive/"
    "RAG_Project/checkpoints"
)


# ============================================================
# SAFETY CHECK — MONGODB
# ============================================================

assert "collection" in globals(), (
    "STOP: MongoDB collection is not available."
)


print()
print("MongoDB collection object: FOUND")

try:

    print(
        "Database:",
        collection.database.name
    )

    print(
        "Collection:",
        collection.name
    )

except Exception:

    print(
        "MongoDB database/collection metadata unavailable."
    )


# ============================================================
# VERIFY FINAL MONGODB DATASET
# ============================================================

stored_count = collection.count_documents({})


assert stored_count == EXPECTED_FINAL_CHUNKS, (
    f"STOP: Expected {EXPECTED_FINAL_CHUNKS} "
    f"documents, got {stored_count}."
)


print()
print(
    "MongoDB documents:",
    stored_count
)

print(
    "Dataset count: PASS"
)


# ============================================================
# RECOVER FINAL CHUNKS DIRECTLY FROM MONGODB
#
# IMPORTANT:
# Cell 25 is diagnosing the FINAL dataset.
#
# Therefore it must NOT depend on a Python variable named
# `chunks` or `final_chunks`.
#
# MongoDB is the source of truth for the uploaded dataset.
# ============================================================

print()
print("=" * 70)
print("RECOVERING FINAL CHUNKS FROM MONGODB")
print("=" * 70)


final_chunks_from_mongo = list(

    collection.find(

        {},

        {
            "_id": 0,

            "chunk_id": 1,

            "text": 1,

            "heading": 1,

            "level": 1,

            "page_start": 1,

            "page_end": 1,

            "chunk_order": 1,

            "word_count": 1,
        }

    ).sort(
        "chunk_id",
        1
    )
)


assert len(
    final_chunks_from_mongo
) == EXPECTED_FINAL_CHUNKS, (

    "STOP: Could not recover the expected "
    f"{EXPECTED_FINAL_CHUNKS} final chunks from MongoDB."
)


print(
    "Final chunks recovered:",
    len(final_chunks_from_mongo)
)


# ============================================================
# VALIDATE FINAL CHUNK ORDER
# ============================================================

stored_chunk_ids = [

    int(
        chunk["chunk_id"]
    )

    for chunk
    in final_chunks_from_mongo

]


expected_chunk_ids = list(
    range(
        EXPECTED_FINAL_CHUNKS
    )
)


assert stored_chunk_ids == expected_chunk_ids, (

    "STOP: Final MongoDB chunk IDs are not "
    "sequential from 0 to "
    f"{EXPECTED_FINAL_CHUNKS - 1}."
)


print(
    "Final chunk IDs: PASS"
)


# ============================================================
# VALIDATE REQUIRED FIELDS
# ============================================================

required_chunk_fields = [

    "chunk_id",

    "text",

    "heading",

    "level",

    "page_start",

    "page_end",

    "chunk_order",

    "word_count",

]


for chunk in final_chunks_from_mongo:

    for field in required_chunk_fields:

        assert field in chunk, (

            f"STOP: Final MongoDB chunk "
            f"{chunk.get('chunk_id')} "
            f"is missing field: {field}"
        )


print(
    "Final chunk metadata: PASS"
)


# ============================================================
# 1. ANALYZE FINAL CHUNK SIZE DISTRIBUTION
# ============================================================

print()
print("=" * 70)
print("FINAL CHUNK SIZE DISTRIBUTION")
print("=" * 70)


word_counts = [

    int(
        chunk["word_count"]
    )

    for chunk
    in final_chunks_from_mongo

]


assert len(word_counts) == EXPECTED_FINAL_CHUNKS


assert all(
    count >= 0
    for count in word_counts
), (
    "STOP: Invalid negative word count detected."
)


tiny_chunks = [

    chunk

    for chunk
    in final_chunks_from_mongo

    if int(
        chunk["word_count"]
    ) < MIN_REASONABLE_WORDS

]


print(
    "Total chunks:",
    len(final_chunks_from_mongo)
)

print(
    "Minimum words:",
    min(word_counts)
)

print(
    "Maximum words:",
    max(word_counts)
)

print(
    "Average words:",
    sum(word_counts) / len(word_counts)
)

print(
    f"Chunks with < {MIN_REASONABLE_WORDS} words:",
    len(tiny_chunks)
)

print(
    "Percentage tiny chunks:",
    f"{100 * len(tiny_chunks) / len(word_counts):.2f}%"
)


# ============================================================
# 2. DISPLAY TINY CHUNKS
# ============================================================

print()
print("=" * 70)
print("TINY CHUNKS")
print("=" * 70)


if len(tiny_chunks) == 0:

    print(
        "No tiny chunks detected."
    )

else:

    for chunk in tiny_chunks:

        print()

        print(
            f"Chunk {chunk['chunk_id']} | "
            f"Words: {chunk['word_count']}"
        )

        print(
            f"Heading: {chunk['heading']}"
        )

        print(
            f"Pages: "
            f"{chunk['page_start']}-"
            f"{chunk['page_end']}"
        )

        print("-" * 70)

        print(
            repr(
                chunk["text"]
            )[:500]
        )


# ============================================================
# 3. CHECK EXPECTED SEMANTIC HEADINGS
#
# These headings correspond to the six evaluation questions.
#
# IMPORTANT:
# We inspect the FINAL MongoDB dataset directly.
# No dependency on baseline `chunks`.
# ============================================================

TARGET_HEADINGS = [

    "What is self-help?",

    "Who is self-help for?",

    "When to use guided or unguided self-help",

    "Ethics and principles",

    "Respond appropriately",

    "Module 1. Start small and reduce stress",

    "Step-by-Step skills, modules and support",

    "DWM skills, modules and support contacts",

    "Module 3. Acting on your values",

    "Overview",

]


print()
print("=" * 70)
print("TARGET HEADING INSPECTION")
print("=" * 70)


heading_diagnostics = {}


for target_heading in TARGET_HEADINGS:

    matching_chunks = [

        chunk

        for chunk
        in final_chunks_from_mongo

        if chunk["heading"]
        == target_heading

    ]


    heading_diagnostics[
        target_heading
    ] = matching_chunks


    print()

    print(
        "Heading:",
        target_heading
    )

    print(
        "Matching final chunks:",
        len(matching_chunks)
    )


    for chunk in matching_chunks:

        print(
            f"  Chunk {chunk['chunk_id']} | "
            f"Words {chunk['word_count']} | "
            f"Pages "
            f"{chunk['page_start']}-"
            f"{chunk['page_end']}"
        )


# ============================================================
# 4. CHECK DUPLICATE TEXT
# ============================================================

print()
print("=" * 70)
print("DUPLICATE TEXT DIAGNOSTIC")
print("=" * 70)


text_to_ids = {}


for chunk in final_chunks_from_mongo:

    text = (
        chunk["text"]
        .strip()
    )


    text_to_ids.setdefault(
        text,
        []
    ).append(
        int(
            chunk["chunk_id"]
        )
    )


duplicate_groups = [

    ids

    for ids
    in text_to_ids.values()

    if len(ids) > 1

]


print(
    "Duplicate text groups:",
    len(duplicate_groups)
)


if duplicate_groups:

    print()

    for ids in duplicate_groups[:20]:

        print(
            "Duplicate chunk IDs:",
            ids
        )

else:

    print(
        "Duplicate text check: PASS"
    )


# ============================================================
# 5. CHECK EMPTY / VERY SHORT TEXT
# ============================================================

print()
print("=" * 70)
print("VERY SHORT TEXT CONTENT")
print("=" * 70)


empty_text_chunks = [

    chunk

    for chunk
    in final_chunks_from_mongo

    if not isinstance(
        chunk["text"],
        str
    )
    or not chunk["text"].strip()

]


assert len(
    empty_text_chunks
) == 0, (

    "STOP: Empty text chunks detected in MongoDB."
)


if len(tiny_chunks) == 0:

    print(
        "No very short chunks detected."
    )

else:

    for chunk in tiny_chunks[:50]:

        print()

        print(
            f"Chunk {chunk['chunk_id']} | "
            f"{chunk['word_count']} words"
        )

        print(
            f"Heading: {chunk['heading']}"
        )

        print(
            f"Text: "
            f"{repr(chunk['text'])}"
        )


# ============================================================
# 6. CHECK FOR SUSPICIOUS LOW-INFORMATION CHUNKS
#
# Tiny chunks are not automatically wrong.
#
# Some headings naturally produce short chunks.
# We therefore report them instead of failing the pipeline.
# ============================================================

print()
print("=" * 70)
print("LOW-INFORMATION CHUNK DIAGNOSTIC")
print("=" * 70)


SHORT_CHUNK_THRESHOLD = 10


very_short_chunks = [

    chunk

    for chunk
    in final_chunks_from_mongo

    if int(
        chunk["word_count"]
    ) <= SHORT_CHUNK_THRESHOLD

]


print(
    f"Chunks with <= {SHORT_CHUNK_THRESHOLD} words:",
    len(very_short_chunks)
)


if very_short_chunks:

    print()

    for chunk in very_short_chunks[:30]:

        print(
            f"Chunk {chunk['chunk_id']} | "
            f"{chunk['word_count']} words | "
            f"{chunk['heading']}"
        )


# ============================================================
# 7. CHECK CHUNK ORDER CONSISTENCY
# ============================================================

print()
print("=" * 70)
print("CHUNK ORDER CONSISTENCY")
print("=" * 70)


stored_orders = [

    int(
        chunk["chunk_order"]
    )

    for chunk
    in final_chunks_from_mongo

]


assert stored_orders == expected_chunk_ids, (

    "STOP: chunk_order does not match "
    "chunk_id sequence."
)


print(
    "chunk_id / chunk_order alignment: PASS"
)


# ============================================================
# 8. CHECK PAGE METADATA
# ============================================================

print()
print("=" * 70)
print("PAGE METADATA DIAGNOSTIC")
print("=" * 70)


invalid_page_chunks = []


for chunk in final_chunks_from_mongo:

    page_start = int(
        chunk["page_start"]
    )

    page_end = int(
        chunk["page_end"]
    )


    if page_start <= 0:

        invalid_page_chunks.append(
            chunk["chunk_id"]
        )

    elif page_end < page_start:

        invalid_page_chunks.append(
            chunk["chunk_id"]
        )


assert len(
    invalid_page_chunks
) == 0, (

    "STOP: Invalid page metadata found in "
    f"chunks: {invalid_page_chunks[:20]}"
)


print(
    "Page metadata: PASS"
)


# ============================================================
# 9. OPTIONAL BASELINE CHECKPOINT RECOVERY
#
# This section is informational only.
#
# It does NOT affect the final diagnostic result.
# ============================================================

print()
print("=" * 70)
print("BASELINE CHECKPOINT STATUS")
print("=" * 70)


baseline_checkpoint_path = os.path.join(
    CHECKPOINT_DIR,
    "chunks.pkl"
)


if os.path.exists(
    baseline_checkpoint_path
):

    try:

        with open(
            baseline_checkpoint_path,
            "rb"
        ) as f:

            baseline_chunks = pickle.load(f)


        if isinstance(
            baseline_chunks,
            list
        ):

            print(
                "Baseline checkpoint: FOUND"
            )

            print(
                "Baseline chunks:",
                len(baseline_chunks)
            )

        else:

            print(
                "Baseline checkpoint found "
                "but format is not a list."
            )

    except Exception as e:

        print(
            "Baseline checkpoint could not be loaded."
        )

        print(
            "Reason:",
            repr(e)
        )

else:

    print(
        "Baseline checkpoint: NOT FOUND"
    )


print()
print(
    "IMPORTANT:"
)

print(
    "Cell 25 does not depend on baseline "
    "`chunks` being present in globals()."
)


# ============================================================
# 10. FINAL DIAGNOSTIC SUMMARY
# ============================================================

print()
print("=" * 70)
print("CELL 25 DIAGNOSTIC SUMMARY")
print("=" * 70)


print(
    "Final chunks:",
    len(final_chunks_from_mongo)
)

print(
    "Minimum words:",
    min(word_counts)
)

print(
    "Maximum words:",
    max(word_counts)
)

print(
    "Average words:",
    f"{sum(word_counts) / len(word_counts):.2f}"
)

print(
    "Tiny chunks:",
    len(tiny_chunks)
)

print(
    "Very short chunks:",
    len(very_short_chunks)
)

print(
    "Duplicate text groups:",
    len(duplicate_groups)
)

print(
    "Empty text chunks:",
    len(empty_text_chunks)
)

print(
    "Invalid page chunks:",
    len(invalid_page_chunks)
)


# ============================================================
# INTERPRETATION
# ============================================================

print()
print("=" * 70)
print("DIAGNOSTIC INTERPRETATION")
print("=" * 70)


if len(tiny_chunks) > 0:

    print(
        "WARNING: Tiny chunks were detected."
    )

    print(
        "These chunks may contribute to "
        "high-score but low-information retrieval."
    )

else:

    print(
        "Tiny chunk check: PASS"
    )


if len(duplicate_groups) > 0:

    print(
        "WARNING: Duplicate text groups detected."
    )

else:

    print(
        "Duplicate text check: PASS"
    )


if len(very_short_chunks) > 0:

    print(
        "NOTE: Very short chunks exist."
    )

    print(
        "Reviewing them is recommended before "
        "further retrieval tuning."
    )

else:

    print(
        "Very short chunk check: PASS"
    )


# ============================================================
# FINAL STATUS
# ============================================================

print()
print("=" * 70)
print("FINAL VALIDATION")
print("=" * 70)

print(
    "MongoDB dataset: PASS"
)

print(
    "Final chunk recovery: PASS"
)

print(
    "Chunk metadata: PASS"
)

print(
    "Chunk IDs/order: PASS"
)

print(
    "Page metadata: PASS"
)

print(
    "Text integrity: PASS"
)


print()
print("=" * 70)
print("CELL 25 — FINAL RETRIEVAL DIAGNOSTICS: PASS")
print("=" * 70)
````

````python
# ============================================================
# CELL 26 — QUALITY-AWARE RETRIEVAL
# RECOVERY-SAFE VERSION
# ============================================================

import os
import pickle
import numpy as np

print("=" * 70)
print("CELL 26 — QUALITY-AWARE RETRIEVAL")
print("RECOVERY-SAFE VERSION")
print("=" * 70)


# ============================================================
# CONFIGURATION
# ============================================================

FINAL_TOP_K = 5
CANDIDATE_K = 50
MIN_WORD_COUNT = 20

VECTOR_INDEX_NAME = "chunk_vector_index"
VECTOR_FIELD = "embedding"
VECTOR_DIMENSION = 1024

EXPECTED_FINAL_CHUNKS = 248

CHECKPOINT_DIR = (
    "/content/drive/MyDrive/"
    "RAG_Project/checkpoints"
)

QUALITY_CHECKPOINT_PATH = (
    f"{CHECKPOINT_DIR}/"
    "quality_aware_retrieval_checkpoint.pkl"
)


# ============================================================
# FINAL EVALUATION SET
# ============================================================

FINAL_EVALUATION = [

    {
        "id": 1,
        "question": "What is self-help?",
    },

    {
        "id": 2,
        "question": "Who is self-help intended for?",
    },

    {
        "id": 3,
        "question": "When should guided self-help be used?",
    },

    {
        "id": 4,
        "question": (
            "What should a helper do when someone is "
            "at risk of suicide?"
        ),
    },

    {
        "id": 5,
        "question": (
            "What are the Step-by-Step skills "
            "for managing stress?"
        ),
    },

    {
        "id": 6,
        "question": (
            "What are the Doing What Matters modules?"
        ),
    },
]


# ============================================================
# SAFETY CHECKS
# ============================================================

assert "embedding_model" in globals(), (
    "STOP: embedding_model is not defined. "
    "Run the BGE-M3 embedding/model cell first."
)

assert callable(
    embedding_model.encode
), (
    "STOP: embedding_model is not usable."
)

assert "collection" in globals(), (
    "STOP: MongoDB collection is not available."
)

assert FINAL_TOP_K > 0, (
    "STOP: FINAL_TOP_K must be greater than zero."
)

assert CANDIDATE_K >= FINAL_TOP_K, (
    "STOP: CANDIDATE_K must be >= FINAL_TOP_K."
)

assert MIN_WORD_COUNT >= 1, (
    "STOP: MIN_WORD_COUNT must be >= 1."
)


# ============================================================
# VERIFY MONGODB DATASET
# ============================================================

stored_count = collection.count_documents({})

assert stored_count == EXPECTED_FINAL_CHUNKS, (
    f"STOP: Expected {EXPECTED_FINAL_CHUNKS} "
    f"documents in MongoDB, got {stored_count}."
)

print()
print("MongoDB collection object: FOUND")
print("Database: psychological_self_help")
print("Collection: chunks")

print(
    "MongoDB documents:",
    stored_count
)

print(
    "Final dataset count: PASS"
)


# ============================================================
# QUALITY-AWARE RETRIEVAL FUNCTION
# ============================================================

def retrieve_quality_aware(
    question,
    top_k=FINAL_TOP_K,
    candidate_k=CANDIDATE_K,
    min_word_count=MIN_WORD_COUNT
):

    # --------------------------------------------------------
    # Validate input
    # --------------------------------------------------------

    assert isinstance(
        question,
        str
    ), (
        "STOP: Question must be a string."
    )

    assert question.strip(), (
        "STOP: Question cannot be empty."
    )

    assert top_k > 0, (
        "STOP: top_k must be greater than zero."
    )

    assert candidate_k >= top_k, (
        "STOP: candidate_k must be >= top_k."
    )


    # --------------------------------------------------------
    # Generate BGE-M3 query embedding
    # --------------------------------------------------------

    query_vector = embedding_model.encode(
        question,
        convert_to_numpy=True
    )

    query_vector = np.asarray(
        query_vector,
        dtype=np.float32
    )


    # --------------------------------------------------------
    # Validate query embedding
    # --------------------------------------------------------

    assert query_vector.ndim == 1, (
        f"STOP: Query embedding must be 1D, "
        f"got {query_vector.shape}."
    )

    assert len(
        query_vector
    ) == VECTOR_DIMENSION, (
        f"STOP: Query embedding has "
        f"{len(query_vector)} dimensions; "
        f"expected {VECTOR_DIMENSION}."
    )

    assert np.isfinite(
        query_vector
    ).all(), (
        "STOP: Query embedding contains "
        "NaN or infinite values."
    )


    # --------------------------------------------------------
    # L2 normalize query embedding
    # --------------------------------------------------------

    query_norm = np.linalg.norm(
        query_vector
    )

    assert query_norm > 0, (
        "STOP: Query embedding has zero norm."
    )

    query_vector = (
        query_vector / query_norm
    ).astype(
        np.float32
    )

    final_query_norm = np.linalg.norm(
        query_vector
    )

    assert abs(
        final_query_norm - 1.0
    ) < 1e-4, (
        "STOP: Query embedding normalization failed."
    )


    # --------------------------------------------------------
    # MongoDB Atlas Vector Search
    # --------------------------------------------------------

    pipeline = [

        {
            "$vectorSearch": {

                "index":
                    VECTOR_INDEX_NAME,

                "path":
                    VECTOR_FIELD,

                "queryVector":
                    query_vector.tolist(),

                "numCandidates":
                    candidate_k,

                "limit":
                    candidate_k,
            }
        },

        {
            "$project": {

                "_id": 0,

                "chunk_id": 1,

                "text": 1,

                "heading": 1,

                "level": 1,

                "page_start": 1,

                "page_end": 1,

                "chunk_order": 1,

                "word_count": 1,

                "score": {
                    "$meta":
                        "vectorSearchScore"
                },
            }
        },
    ]


    candidates = list(
        collection.aggregate(
            pipeline
        )
    )


    assert len(candidates) > 0, (
        "STOP: MongoDB vector search "
        "returned zero candidates."
    )


    # --------------------------------------------------------
    # Quality filtering
    #
    # Remove:
    # 1. Very short chunks
    # 2. Empty text
    # 3. Exact duplicate text
    # --------------------------------------------------------

    filtered_results = []

    seen_text = set()

    removed_tiny = 0
    removed_empty = 0
    removed_duplicate = 0


    for result in candidates:

        word_count = int(
            result.get(
                "word_count",
                0
            )
        )


        # ----------------------------------------------------
        # Remove tiny chunks
        # ----------------------------------------------------

        if word_count < min_word_count:

            removed_tiny += 1

            continue


        # ----------------------------------------------------
        # Validate text
        # ----------------------------------------------------

        text = (
            result.get(
                "text",
                ""
            )
            .strip()
        )


        if not text:

            removed_empty += 1

            continue


        # ----------------------------------------------------
        # Normalize text for duplicate detection
        # ----------------------------------------------------

        normalized_text = " ".join(
            text.split()
        )


        if normalized_text in seen_text:

            removed_duplicate += 1

            continue


        seen_text.add(
            normalized_text
        )


        filtered_results.append(
            result
        )


        if len(
            filtered_results
        ) >= top_k:

            break


    # --------------------------------------------------------
    # Validate filtered results
    # --------------------------------------------------------

    assert len(
        filtered_results
    ) > 0, (
        "STOP: Quality filtering removed "
        "all retrieved candidates."
    )

    assert len(
        filtered_results
    ) <= top_k, (
        "STOP: Quality-aware retrieval "
        "returned more than TOP_K."
    )


    return (
        filtered_results,
        {
            "candidate_count":
                len(candidates),

            "removed_tiny":
                removed_tiny,

            "removed_empty":
                removed_empty,

            "removed_duplicate":
                removed_duplicate,

            "final_count":
                len(filtered_results),
        }
    )


# ============================================================
# RUN QUALITY-AWARE RETRIEVAL
# ============================================================

quality_retrieval_results = []

quality_retrieval_records = []

print()
print("=" * 70)
print("QUALITY-AWARE RETRIEVAL RESULTS")
print("=" * 70)


for item in FINAL_EVALUATION:

    question_id = item["id"]

    question = item["question"]


    retrieved, diagnostics = (
        retrieve_quality_aware(
            question=question,
            top_k=FINAL_TOP_K,
            candidate_k=CANDIDATE_K,
            min_word_count=MIN_WORD_COUNT
        )
    )


    quality_retrieval_results.append(
        retrieved
    )


    quality_retrieval_records.append(
        {
            "id":
                question_id,

            "question":
                question,

            "results":
                retrieved,

            "diagnostics":
                diagnostics,
        }
    )


    print()
    print(
        f"QUERY {question_id}/"
        f"{len(FINAL_EVALUATION)}"
    )

    print("-" * 70)

    print("Question:")
    print(question)

    print()

    print(
        "Candidates retrieved:",
        diagnostics["candidate_count"]
    )

    print(
        "Tiny chunks removed:",
        diagnostics["removed_tiny"]
    )

    print(
        "Empty chunks removed:",
        diagnostics["removed_empty"]
    )

    print(
        "Duplicate chunks removed:",
        diagnostics["removed_duplicate"]
    )

    print(
        "Final results:",
        diagnostics["final_count"]
    )


    for rank, result in enumerate(
        retrieved,
        start=1
    ):

        print()

        print(
            f"Rank {rank} | "
            f"Chunk {result['chunk_id']} | "
            f"Score: "
            f"{result['score']:.6f}"
        )

        print(
            f"Heading: "
            f"{result['heading']}"
        )

        print(
            f"Pages: "
            f"{result['page_start']}-"
            f"{result['page_end']}"
        )

        print(
            f"Words: "
            f"{result['word_count']}"
        )


# ============================================================
# QUALITY VALIDATION
# ============================================================

assert len(
    quality_retrieval_results
) == len(
    FINAL_EVALUATION
), (
    "STOP: Retrieval result count does not "
    "match evaluation query count."
)


assert all(
    len(results) > 0
    for results in quality_retrieval_results
), (
    "STOP: At least one evaluation query "
    "returned zero final results."
)


assert all(
    all(
        int(result["word_count"]) >= MIN_WORD_COUNT
        for result in results
    )
    for results in quality_retrieval_results
), (
    "STOP: Quality-aware retrieval returned "
    "a chunk below MIN_WORD_COUNT."
)


# ============================================================
# VERIFY NO EXACT DUPLICATES WITHIN EACH RESULT SET
# ============================================================

for results in quality_retrieval_results:

    normalized_texts = [

        " ".join(
            result["text"]
            .strip()
            .split()
        )

        for result in results

    ]


    assert len(
        normalized_texts
    ) == len(
        set(normalized_texts)
    ), (
        "STOP: Exact duplicate text detected "
        "inside a final retrieval result set."
    )


print()
print(
    "Minimum word-count filtering: PASS"
)

print(
    "Empty-text filtering: PASS"
)

print(
    "Exact duplicate filtering: PASS"
)

print(
    "All evaluation queries returned results: PASS"
)


# ============================================================
# SAVE RECOVERY CHECKPOINT
# ============================================================

checkpoint_payload = {

    "status":
        "complete",

    "embedding_model":
        "BAAI/bge-m3",

    "embedding_dimension":
        VECTOR_DIMENSION,

    "vector_index":
        VECTOR_INDEX_NAME,

    "vector_field":
        VECTOR_FIELD,

    "final_chunks":
        EXPECTED_FINAL_CHUNKS,

    "candidate_k":
        CANDIDATE_K,

    "top_k":
        FINAL_TOP_K,

    "min_word_count":
        MIN_WORD_COUNT,

    "evaluation":
        FINAL_EVALUATION,

    "results":
        quality_retrieval_records,
}


os.makedirs(
    CHECKPOINT_DIR,
    exist_ok=True
)


with open(
    QUALITY_CHECKPOINT_PATH,
    "wb"
) as f:

    pickle.dump(
        checkpoint_payload,
        f
    )


assert os.path.exists(
    QUALITY_CHECKPOINT_PATH
), (
    "STOP: Quality-aware retrieval "
    "checkpoint was not created."
)


checkpoint_size = os.path.getsize(
    QUALITY_CHECKPOINT_PATH
)


assert checkpoint_size > 0, (
    "STOP: Quality-aware retrieval "
    "checkpoint is empty."
)


print()
print(
    "Quality-aware checkpoint:"
)

print(
    QUALITY_CHECKPOINT_PATH
)

print(
    "Checkpoint size:",
    checkpoint_size,
    "bytes"
)

print(
    "Checkpoint: SAVED"
)


# ============================================================
# FINAL SUMMARY
# ============================================================

print()
print("=" * 70)
print("QUALITY-AWARE RETRIEVAL SUMMARY")
print("=" * 70)

print(
    "Final MongoDB chunks:",
    EXPECTED_FINAL_CHUNKS
)

print(
    "Embedding model:",
    "BAAI/bge-m3"
)

print(
    "Embedding dimension:",
    VECTOR_DIMENSION
)

print(
    "Vector index:",
    VECTOR_INDEX_NAME
)

print(
    "Candidate pool:",
    CANDIDATE_K
)

print(
    "Final Top-K:",
    FINAL_TOP_K
)

print(
    "Minimum chunk words:",
    MIN_WORD_COUNT
)

print(
    "Queries tested:",
    len(FINAL_EVALUATION)
)

print("-" * 70)

print(
    "MongoDB dataset: PASS"
)

print(
    "BGE-M3 query embedding: PASS"
)

print(
    "Query normalization: PASS"
)

print(
    "Candidate retrieval: PASS"
)

print(
    "Tiny chunk filtering: PASS"
)

print(
    "Duplicate filtering: PASS"
)

print(
    "Final Top-K retrieval: PASS"
)

print(
    "Recovery checkpoint: SAVED"
)

print("=" * 70)

print(
    "CELL 26 — QUALITY-AWARE RETRIEVAL: PASS"
)

print("=" * 70)
````

````python
# ============================================================
# CELL 27 — QUALITY-AWARE PRECISION@5
# RECOVERY-SAFE VERSION
# ============================================================

print("=" * 70)
print("CELL 27 — QUALITY-AWARE PRECISION@5")
print("RECOVERY-SAFE VERSION")
print("=" * 70)


# ============================================================
# CONFIGURATION
# ============================================================

TOP_K = 5

BASELINE_PRECISION_AT_5 = 0.3000
BASELINE_HEADING_HIT_RATE = 0.8333


# ============================================================
# SAFETY CHECKS
# ============================================================

assert "quality_retrieval_results" in globals(), (
    "STOP: quality_retrieval_results not found. "
    "Run Cell 26 first."
)

assert len(quality_retrieval_results) == 6, (
    "STOP: Expected 6 quality-aware retrieval result sets."
)


# ============================================================
# RECOVER FINAL EVALUATION
# ============================================================

FINAL_EVALUATION = [

    {
        "id": 1,
        "question": "What is self-help?",
    },

    {
        "id": 2,
        "question": "Who is self-help intended for?",
    },

    {
        "id": 3,
        "question": "When should guided self-help be used?",
    },

    {
        "id": 4,
        "question": (
            "What should a helper do when someone is "
            "at risk of suicide?"
        ),
    },

    {
        "id": 5,
        "question": (
            "What are the Step-by-Step skills "
            "for managing stress?"
        ),
    },

    {
        "id": 6,
        "question": (
            "What are the Doing What Matters modules?"
        ),
    },
]


# ============================================================
# RECOVER GOLD HEADINGS
# ============================================================

GOLD_HEADINGS = [

    {
        "What is self-help?"
    },

    {
        "Who is self-help for?"
    },

    {
        "When to use guided or unguided self-help"
    },

    {
        "Ethics and principles",
        "Respond appropriately"
    },

    {
        "Module 1. Start small and reduce stress",
        "Step-by-Step skills, modules and support"
    },

    {
        "DWM skills, modules and support contacts",
        "Module 3. Acting on your values",
        "Overview"
    },
]


assert len(GOLD_HEADINGS) == 6


print()
print("Evaluation set: RECOVERED")
print("Queries:", len(FINAL_EVALUATION))
print("Gold semantic targets: RECOVERED")


# ============================================================
# EVALUATION
# ============================================================

quality_precision_results = []

total_precision = 0.0

successful_queries = 0

total_relevant = 0


print()
print("=" * 70)
print("QUALITY-AWARE PRECISION@5 RESULTS")
print("=" * 70)


for query_index, item in enumerate(
    FINAL_EVALUATION
):

    question = item["question"]

    expected_headings = GOLD_HEADINGS[
        query_index
    ]

    retrieved = quality_retrieval_results[
        query_index
    ]


    assert len(retrieved) <= TOP_K


    retrieved_headings = [

        result["heading"]

        for result in retrieved

    ]


    relevant = [

        result

        for result in retrieved

        if result["heading"]
        in expected_headings

    ]


    relevant_count = len(relevant)


    precision_at_5 = (
        relevant_count / TOP_K
    )


    total_precision += precision_at_5

    total_relevant += relevant_count


    if relevant_count > 0:

        successful_queries += 1


    quality_precision_results.append({

        "id":
            item["id"],

        "question":
            question,

        "expected_headings":
            sorted(expected_headings),

        "retrieved_chunks": [

            int(result["chunk_id"])

            for result in retrieved

        ],

        "retrieved_headings":
            retrieved_headings,

        "relevant_chunks": [

            int(result["chunk_id"])

            for result in relevant

        ],

        "relevant_count":
            relevant_count,

        "precision_at_5":
            precision_at_5,

    })


    print()
    print(
        f"Q{item['id']}: {question}"
    )

    print()
    print("Expected headings:")

    for heading in sorted(
        expected_headings
    ):

        print(
            f"  - {heading}"
        )


    print()
    print("Retrieved:")


    for rank, result in enumerate(
        retrieved,
        start=1
    ):

        is_relevant = (
            result["heading"]
            in expected_headings
        )


        marker = "✓" if is_relevant else " "


        print(
            f"  {marker} Rank {rank} | "
            f"Chunk {result['chunk_id']} | "
            f"{result['heading']} | "
            f"Score {result['score']:.6f}"
        )


    print()

    print(
        f"Relevant retrieved: "
        f"{relevant_count}/{TOP_K}"
    )

    print(
        f"Precision@5: "
        f"{precision_at_5:.3f}"
    )


# ============================================================
# OVERALL METRICS
# ============================================================

quality_mean_precision_at_5 = (
    total_precision
    / len(FINAL_EVALUATION)
)


quality_heading_hit_rate = (
    successful_queries
    / len(FINAL_EVALUATION)
)


average_relevant_chunks = (
    total_relevant
    / len(FINAL_EVALUATION)
)


precision_improvement = (
    quality_mean_precision_at_5
    - BASELINE_PRECISION_AT_5
)


heading_hit_improvement = (
    quality_heading_hit_rate
    - BASELINE_HEADING_HIT_RATE
)


# ============================================================
# FINAL SUMMARY
# ============================================================

print()
print("=" * 70)
print("QUALITY-AWARE PRECISION@5 SUMMARY")
print("=" * 70)

print(
    "Embedding model:",
    "BAAI/bge-m3"
)

print(
    "Final chunks:",
    248
)

print(
    "Chunk configuration:",
    "300 words / 50 overlap"
)

print(
    "Candidate pool:",
    50
)

print(
    "Final Top-K:",
    TOP_K
)

print(
    "Minimum chunk words:",
    20
)

print(
    "Queries evaluated:",
    len(FINAL_EVALUATION)
)


print()
print(
    f"Baseline Mean Precision@5: "
    f"{BASELINE_PRECISION_AT_5:.4f}"
)

print(
    f"Quality-Aware Mean Precision@5: "
    f"{quality_mean_precision_at_5:.4f}"
)

print(
    f"Precision improvement: "
    f"{precision_improvement:+.4f}"
)


print()
print(
    f"Baseline Heading Hit Rate: "
    f"{BASELINE_HEADING_HIT_RATE:.4f}"
)

print(
    f"Quality-Aware Heading Hit Rate: "
    f"{quality_heading_hit_rate:.4f}"
)

print(
    f"Heading hit improvement: "
    f"{heading_hit_improvement:+.4f}"
)


print()
print(
    f"Average relevant chunks/query: "
    f"{average_relevant_chunks:.2f}"
)

print(
    f"Total relevant retrieved chunks: "
    f"{total_relevant}"
)


# ============================================================
# VALIDATION
# ============================================================

assert (
    0.0
    <= quality_mean_precision_at_5
    <= 1.0
)

assert (
    0.0
    <= quality_heading_hit_rate
    <= 1.0
)

assert len(
    quality_precision_results
) == 6


# ============================================================
# FINAL STATUS
# ============================================================

print()
print("=" * 70)

print(
    "Quality-aware evaluation: PASS"
)

print(
    "Gold semantic mapping: PASS"
)

print(
    "Precision@5 calculation: PASS"
)

print(
    "Comparison with baseline: PASS"
)

print("=" * 70)

print(
    "CELL 27 — QUALITY-AWARE PRECISION@5: PASS"
)

print("=" * 70)
````

````python
# ============================================================
# CELL 29 — FINAL RAG ANSWER GENERATION
# FINAL RETRIEVAL = HEADING FOCUSED
# Dense = 0.75 | Heading = 0.25 | Text = 0.0
# ============================================================

import numpy as np
from pymongo import MongoClient

print("=" * 70)
print("CELL 29 — FINAL RAG ANSWER GENERATION")
print("=" * 70)

# ============================================================
# FINAL CONFIGURATION
# ============================================================

FINAL_CHUNK_COUNT = 248

TOP_K = 5
NUM_CANDIDATES = 50
MIN_CHUNK_WORDS = 20

VECTOR_INDEX_NAME = "chunk_vector_index"

EMBEDDING_DIMENSION = 1024
EMBEDDING_MODEL_NAME = "BAAI/bge-m3"

# ------------------------------------------------------------
# FINAL RERANKING CONFIGURATION
# Selected from Cell 30
# ------------------------------------------------------------

DENSE_WEIGHT = 0.75
HEADING_WEIGHT = 0.25
TEXT_WEIGHT = 0.0

RERANKING_STRATEGY = "Heading Focused"

print()
print("=" * 70)
print("FINAL RERANKING CONFIGURATION")
print("=" * 70)

print("Strategy:", RERANKING_STRATEGY)
print("Dense weight:", DENSE_WEIGHT)
print("Heading weight:", HEADING_WEIGHT)
print("Text weight:", TEXT_WEIGHT)

# ============================================================
# SAFETY CHECKS
# ============================================================

assert "embedding_model" in globals(), (
    "STOP: embedding_model is not defined. "
    "Run the embedding model setup cell first."
)

assert callable(embedding_model.encode), (
    "STOP: embedding_model is not usable."
)

# ============================================================
# FIND FINAL 248-CHUNK DATASET
# ============================================================

final_chunks_candidates = []

for variable_name, variable_value in list(globals().items()):

    if variable_name.startswith("_"):
        continue

    if not isinstance(variable_value, list):
        continue

    if len(variable_value) != FINAL_CHUNK_COUNT:
        continue

    if len(variable_value) == 0:
        continue

    first_item = variable_value[0]

    if not isinstance(first_item, dict):
        continue

    required_keys = {
        "chunk_id",
        "chunk_order",
        "text",
        "heading",
        "level",
        "page_start",
        "page_end",
        "word_count",
    }

    if required_keys.issubset(first_item.keys()):
        final_chunks_candidates.append(
            (variable_name, variable_value)
        )

assert len(final_chunks_candidates) > 0, (
    "STOP: Could not find the final 248-chunk dataset. "
    "Expected a list containing 248 chunk dictionaries."
)

print()
print("Final chunk candidates:")
print(
    [
        variable_name
        for variable_name, _ in final_chunks_candidates
    ]
)

# ------------------------------------------------------------
# Prefer final_chunks
# ------------------------------------------------------------

preferred_names = [
    "final_chunks",
    "experimental_chunks",
]

final_chunks = None
final_chunks_variable_name = None

for preferred_name in preferred_names:

    for variable_name, variable_value in final_chunks_candidates:

        if variable_name == preferred_name:

            final_chunks = variable_value
            final_chunks_variable_name = variable_name

            break

    if final_chunks is not None:
        break

# ------------------------------------------------------------
# Fallback
# ------------------------------------------------------------

if final_chunks is None:

    final_chunks_variable_name, final_chunks = (
        final_chunks_candidates[0]
    )

print(
    f"Using final chunk variable: "
    f"{final_chunks_variable_name}"
)

# ============================================================
# VALIDATE FINAL CHUNKS
# ============================================================

assert len(final_chunks) == FINAL_CHUNK_COUNT

assert all(
    chunk["chunk_id"] == index
    for index, chunk in enumerate(final_chunks)
)

assert all(
    chunk["chunk_order"] == index
    for index, chunk in enumerate(final_chunks)
)

assert all(
    chunk["text"].strip()
    for chunk in final_chunks
)

assert all(
    "heading" in chunk
    and "page_start" in chunk
    and "page_end" in chunk
    and "word_count" in chunk
    for chunk in final_chunks
)

print()
print("Final chunk dataset: PASS")
print(f"Final chunks: {len(final_chunks)}")

# ============================================================
# FIND FINAL EMBEDDINGS
# ============================================================

embedding_candidates = []

for variable_name, variable_value in list(globals().items()):

    if variable_name.startswith("_"):
        continue

    if not isinstance(variable_value, np.ndarray):
        continue

    if variable_value.ndim != 2:
        continue

    if variable_value.shape == (
        FINAL_CHUNK_COUNT,
        EMBEDDING_DIMENSION,
    ):

        embedding_candidates.append(
            (variable_name, variable_value)
        )

assert len(embedding_candidates) > 0, (
    "STOP: Could not find a (248, 1024) final embedding matrix."
)

print()
print("Final embedding candidates:")
print(
    [
        variable_name
        for variable_name, _ in embedding_candidates
    ]
)

# ------------------------------------------------------------
# Prefer the exact embedding uploaded to MongoDB
# ------------------------------------------------------------

preferred_embedding_names = [
    "final_raw_embeddings",
    "final_embeddings",
    "experimental_embeddings",
]

final_embeddings = None
final_embedding_variable_name = None

for preferred_name in preferred_embedding_names:

    for variable_name, variable_value in embedding_candidates:

        if variable_name == preferred_name:

            final_embeddings = variable_value
            final_embedding_variable_name = variable_name

            break

    if final_embeddings is not None:
        break

# ------------------------------------------------------------
# Fallback
# ------------------------------------------------------------

if final_embeddings is None:

    final_embedding_variable_name, final_embeddings = (
        embedding_candidates[0]
    )

print(
    f"Using final embedding variable: "
    f"{final_embedding_variable_name}"
)

# ============================================================
# EMBEDDING VALIDATION
# ============================================================

assert final_embeddings.shape == (
    FINAL_CHUNK_COUNT,
    EMBEDDING_DIMENSION,
)

assert final_embeddings.dtype == np.float32

assert np.isfinite(final_embeddings).all()

embedding_norms = np.linalg.norm(
    final_embeddings,
    axis=1,
    keepdims=True,
)

assert np.all(embedding_norms > 0)

final_embeddings_normalized = (
    final_embeddings / embedding_norms
).astype(np.float32)

assert np.all(
    np.abs(
        np.linalg.norm(
            final_embeddings_normalized,
            axis=1,
        ) - 1.0
    ) < 1e-4
)

print()
print("Final embedding matrix: PASS")
print(
    "Shape:",
    final_embeddings_normalized.shape
)
print(
    "Dimension:",
    EMBEDDING_DIMENSION
)

# ============================================================
# TEXT NORMALIZATION
# ============================================================

def normalize_text(text):

    return " ".join(
        str(text).split()
    ).strip().lower()


# ============================================================
# HEADING OVERLAP SCORE
# ============================================================

def heading_overlap_score(
    question,
    heading,
):

    question_tokens = set(
        normalize_text(question).split()
    )

    heading_tokens = set(
        normalize_text(heading).split()
    )

    if not question_tokens:
        return 0.0

    if not heading_tokens:
        return 0.0

    overlap = (
        question_tokens
        & heading_tokens
    )

    return float(
        len(overlap)
        / len(question_tokens)
    )


# ============================================================
# FINAL HEADING-FOCUSED RETRIEVAL
# ============================================================

def final_quality_aware_retrieve(
    question,
    top_k=TOP_K,
    num_candidates=NUM_CANDIDATES,
    min_words=MIN_CHUNK_WORDS,
):

    # --------------------------------------------------------
    # Encode query
    # --------------------------------------------------------

    query_embedding = embedding_model.encode(
        [question],
        batch_size=1,
        show_progress_bar=False,
        convert_to_numpy=True,
    )

    query_embedding = np.asarray(
        query_embedding,
        dtype=np.float32,
    )

    assert query_embedding.shape == (
        1,
        EMBEDDING_DIMENSION,
    )

    # --------------------------------------------------------
    # Normalize query
    # --------------------------------------------------------

    query_norm = np.linalg.norm(
        query_embedding,
        axis=1,
        keepdims=True,
    )

    assert np.all(query_norm > 0)

    query_embedding = (
        query_embedding / query_norm
    ).astype(np.float32)

    # --------------------------------------------------------
    # Dense similarity
    # --------------------------------------------------------

    dense_similarities = (
        final_embeddings_normalized
        @ query_embedding[0]
    )

    assert dense_similarities.shape == (
        FINAL_CHUNK_COUNT,
    )

    # --------------------------------------------------------
    # Candidate pool
    # --------------------------------------------------------

    candidate_count = min(
        num_candidates,
        FINAL_CHUNK_COUNT,
    )

    candidate_indices = np.argsort(
        dense_similarities
    )[::-1][:candidate_count]

    # --------------------------------------------------------
    # Quality filtering
    #
    # 1. Remove tiny chunks
    # 2. Remove exact duplicate text
    # --------------------------------------------------------

    filtered_candidates = []

    seen_texts = set()

    for index in candidate_indices:

        index = int(index)

        chunk = final_chunks[index]

        # ----------------------------------------------------
        # Remove tiny chunks
        # ----------------------------------------------------

        if chunk["word_count"] < min_words:
            continue

        # ----------------------------------------------------
        # Remove exact duplicate text
        # ----------------------------------------------------

        normalized_text = normalize_text(
            chunk["text"]
        )

        if normalized_text in seen_texts:
            continue

        seen_texts.add(normalized_text)

        filtered_candidates.append(
            {
                "chunk_id": index,
                "dense_score": float(
                    dense_similarities[index]
                ),
                "chunk": chunk,
            }
        )

    # --------------------------------------------------------
    # Heading-focused reranking
    #
    # FINAL CONFIGURATION:
    #
    # 0.75 Dense
    # 0.25 Heading
    # 0.00 Text
    # --------------------------------------------------------

    reranked = []

    for candidate in filtered_candidates:

        chunk = candidate["chunk"]

        heading_score = heading_overlap_score(
            question=question,
            heading=chunk["heading"],
        )

        text_score = 0.0

        final_score = (
            DENSE_WEIGHT
            * candidate["dense_score"]
            +
            HEADING_WEIGHT
            * heading_score
            +
            TEXT_WEIGHT
            * text_score
        )

        reranked.append(
            {
                "chunk_id": candidate["chunk_id"],
                "score": float(final_score),
                "dense_score": float(
                    candidate["dense_score"]
                ),
                "heading_score": float(
                    heading_score
                ),
                "text_score": float(
                    text_score
                ),
                "chunk": chunk,
            }
        )

    # --------------------------------------------------------
    # Sort by final reranking score
    # --------------------------------------------------------

    reranked.sort(
        key=lambda x: x["score"],
        reverse=True,
    )

    return reranked[:top_k]


# ============================================================
# FINAL RAG TEST QUESTIONS
# ============================================================

FINAL_RAG_QUESTIONS = [
    "What is self-help?",
    "Who is self-help intended for?",
    "When should guided self-help be used?",
    (
        "What should a helper do when someone is "
        "at risk of suicide?"
    ),
    (
        "What are the Step-by-Step skills "
        "for managing stress?"
    ),
    "What are the Doing What Matters modules?",
]

assert len(FINAL_RAG_QUESTIONS) == 6

# ============================================================
# FINAL RETRIEVAL TEST
# ============================================================

print()
print("=" * 70)
print("FINAL HEADING-FOCUSED QUALITY-AWARE RETRIEVAL")
print("=" * 70)

final_rag_results = []

for query_index, question in enumerate(
    FINAL_RAG_QUESTIONS,
    start=1,
):

    print()
    print(
        f"QUERY {query_index}/{len(FINAL_RAG_QUESTIONS)}"
    )

    print("-" * 70)

    print("Question:")
    print(question)

    results = final_quality_aware_retrieve(
        question=question,
        top_k=TOP_K,
        num_candidates=NUM_CANDIDATES,
        min_words=MIN_CHUNK_WORDS,
    )

    assert len(results) > 0

    final_rag_results.append(
        {
            "question": question,
            "results": results,
        }
    )

    for rank, result in enumerate(
        results,
        start=1,
    ):

        chunk = result["chunk"]

        print()

        print(
            f"Rank {rank} | "
            f"Chunk {result['chunk_id']} | "
            f"Score: {result['score']:.6f}"
        )

        print(
            f"Heading: {chunk['heading']}"
        )

        print(
            f"Pages: "
            f"{chunk['page_start']}-"
            f"{chunk['page_end']}"
        )

        print(
            f"Words: {chunk['word_count']}"
        )

        print(
            f"Dense: "
            f"{result['dense_score']:.6f} | "
            f"Heading: "
            f"{result['heading_score']:.6f}"
        )

# ============================================================
# FINAL VALIDATION
# ============================================================

assert len(final_rag_results) == 6

assert all(
    len(item["results"]) > 0
    for item in final_rag_results
)

assert all(
    len(item["results"]) <= TOP_K
    for item in final_rag_results
)

assert all(
    all(
        result["chunk"]["word_count"]
        >= MIN_CHUNK_WORDS
        for result in item["results"]
    )
    for item in final_rag_results
)

assert all(
    len(
        {
            normalize_text(
                result["chunk"]["text"]
            )
            for result in item["results"]
        }
    )
    == len(item["results"])
    for item in final_rag_results
)

# ============================================================
# FINAL CONFIGURATION SUMMARY
# ============================================================

print()
print("=" * 70)
print("CELL 29 — FINAL RAG RETRIEVAL: PASS")
print("=" * 70)

print()
print("Final chunks:", len(final_chunks))
print("Embedding model:", EMBEDDING_MODEL_NAME)
print("Embedding dimension:", EMBEDDING_DIMENSION)

print("Candidate pool:", NUM_CANDIDATES)
print("Top-K:", TOP_K)
print("Minimum words:", MIN_CHUNK_WORDS)

print("Duplicate filtering: True")
print("Vector index:", VECTOR_INDEX_NAME)

print()
print("Reranking strategy:", RERANKING_STRATEGY)
print("Dense weight:", DENSE_WEIGHT)
print("Heading weight:", HEADING_WEIGHT)
print("Text weight:", TEXT_WEIGHT)

print()
print("Expected benchmark Precision@5:", 0.6333)
print("Expected Heading Hit Rate:", 1.0000)

print()
print("=" * 70)
print("FINAL RAG RETRIEVAL CONFIGURATION LOCKED")
print("=" * 70)
````

````python
# ============================================================
# CELL 30 — RETRIEVAL RERANKING EXPERIMENT
# ============================================================

import re
import numpy as np

print("=" * 70)
print("CELL 30 — RETRIEVAL RERANKING EXPERIMENT")
print("=" * 70)

# ============================================================
# CONFIGURATION
# ============================================================

TOP_K = 5
NUM_CANDIDATES = 50
MIN_CHUNK_WORDS = 20

# ============================================================
# SAFETY CHECKS
# ============================================================

assert "final_chunks" in globals(), (
    "STOP: final_chunks not found. Run Cell 29 first."
)

assert len(final_chunks) == 248, (
    f"STOP: Expected 248 final chunks, found {len(final_chunks)}."
)

assert "final_embeddings_normalized" in globals(), (
    "STOP: final_embeddings_normalized not found. "
    "Run Cell 29 first."
)

assert final_embeddings_normalized.shape == (248, 1024)

assert "embedding_model" in globals(), (
    "STOP: embedding_model not found."
)

# ============================================================
# EVALUATION SET
# ============================================================

RERANK_EVALUATION = [
    {
        "question": "What is self-help?",
        "expected_headings": {
            "What is self-help?"
        },
    },
    {
        "question": "Who is self-help intended for?",
        "expected_headings": {
            "Who is self-help for?"
        },
    },
    {
        "question": "When should guided self-help be used?",
        "expected_headings": {
            "When to use guided or unguided self-help"
        },
    },
    {
        "question": (
            "What should a helper do when someone is "
            "at risk of suicide?"
        ),
        "expected_headings": {
            "Ethics and principles",
            "Respond appropriately",
        },
    },
    {
        "question": (
            "What are the Step-by-Step skills "
            "for managing stress?"
        ),
        "expected_headings": {
            "Module 1. Start small and reduce stress",
            "Step-by-Step skills, modules and support",
        },
    },
    {
        "question": "What are the Doing What Matters modules?",
        "expected_headings": {
            "DWM skills, modules and support contacts",
            "Module 3. Acting on your values",
            "Overview",
        },
    },
]

assert len(RERANK_EVALUATION) == 6

# ============================================================
# TEXT NORMALIZATION
# ============================================================

def normalize_rerank_text(text):

    text = str(text).lower()

    # Important semantic aliases
    text = text.replace(
        "doing what matters in times of stress",
        "doing what matters dwm"
    )

    text = text.replace(
        "step by step",
        "step-by-step"
    )

    # Keep words and hyphenated terms
    text = re.sub(
        r"[^a-z0-9\- ]+",
        " ",
        text
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text


def tokenize_rerank(text):

    normalized = normalize_rerank_text(text)

    return set(
        token
        for token in normalized.split()
        if len(token) >= 3
    )


# ============================================================
# STOPWORDS
# ============================================================

RERANK_STOPWORDS = {
    "what",
    "when",
    "where",
    "which",
    "who",
    "whom",
    "how",
    "does",
    "do",
    "did",
    "are",
    "is",
    "was",
    "were",
    "the",
    "a",
    "an",
    "to",
    "for",
    "of",
    "in",
    "on",
    "and",
    "or",
    "with",
    "someone",
    "should",
    "can",
    "be",
    "their",
    "this",
    "that",
}

# ============================================================
# QUESTION EXPANSION
# ============================================================

def expand_question_terms(question):

    text = normalize_rerank_text(question)

    tokens = set(
        token
        for token in text.split()
        if (
            len(token) >= 3
            and token not in RERANK_STOPWORDS
        )
    )

    # Semantic aliases used by this evaluation set

    if (
        "doing" in tokens
        and "matters" in tokens
    ):
        tokens.update({
            "dwm",
            "module",
            "modules",
        })

    if (
        "step-by-step" in tokens
        or (
            "step-by-step" in text
        )
    ):
        tokens.update({
            "step-by-step",
            "step",
            "modules",
            "skills",
        })

    if "suicide" in tokens:
        tokens.update({
            "risk",
            "respond",
            "appropriately",
            "ethics",
        })

    if "guided" in tokens:
        tokens.update({
            "unguided",
            "self-help",
        })

    if "self-help" in text:
        tokens.update({
            "self-help",
        })

    return tokens


# ============================================================
# LEXICAL SCORE
# ============================================================

def lexical_overlap_score(
    question,
    chunk,
):

    question_terms = expand_question_terms(
        question
    )

    heading_terms = tokenize_rerank(
        chunk["heading"]
    )

    text_terms = tokenize_rerank(
        chunk["text"]
    )

    if not question_terms:
        return 0.0, 0.0

    heading_overlap = (
        len(question_terms & heading_terms)
        / len(question_terms)
    )

    text_overlap = (
        len(question_terms & text_terms)
        / len(question_terms)
    )

    return (
        float(heading_overlap),
        float(text_overlap),
    )


# ============================================================
# RETRIEVAL CANDIDATES
# ============================================================

def get_rerank_candidates(
    question,
    num_candidates=NUM_CANDIDATES,
):

    query_embedding = embedding_model.encode(
        [question],
        batch_size=1,
        show_progress_bar=False,
        convert_to_numpy=True,
    )

    query_embedding = np.asarray(
        query_embedding,
        dtype=np.float32,
    )

    assert query_embedding.shape == (
        1,
        1024,
    )

    query_norm = np.linalg.norm(
        query_embedding,
        axis=1,
        keepdims=True,
    )

    query_embedding = (
        query_embedding / query_norm
    ).astype(np.float32)

    similarities = (
        final_embeddings_normalized
        @ query_embedding[0]
    )

    candidate_indices = np.argsort(
        similarities
    )[::-1][:num_candidates]

    candidates = []

    seen_texts = set()

    for index in candidate_indices:

        chunk = final_chunks[int(index)]

        if chunk["word_count"] < MIN_CHUNK_WORDS:
            continue

        normalized_text = " ".join(
            chunk["text"].split()
        ).strip().lower()

        if normalized_text in seen_texts:
            continue

        seen_texts.add(normalized_text)

        heading_score, text_score = (
            lexical_overlap_score(
                question,
                chunk,
            )
        )

        candidates.append({
            "chunk_id": int(index),
            "chunk": chunk,
            "dense_score": float(
                similarities[int(index)]
            ),
            "heading_score": heading_score,
            "text_score": text_score,
        })

    return candidates


# ============================================================
# RERANK FUNCTION
# ============================================================

def rerank_candidates(
    candidates,
    dense_weight,
    heading_weight,
    text_weight,
):

    for item in candidates:

        item["rerank_score"] = (
            dense_weight
            * item["dense_score"]
            +
            heading_weight
            * item["heading_score"]
            +
            text_weight
            * item["text_score"]
        )

    ranked = sorted(
        candidates,
        key=lambda item: item["rerank_score"],
        reverse=True,
    )

    return ranked[:TOP_K]


# ============================================================
# RERANKING CONFIGURATIONS
# ============================================================

RERANK_CONFIGS = [

    {
        "name": "Dense Only",
        "dense_weight": 1.0,
        "heading_weight": 0.0,
        "text_weight": 0.0,
    },

    {
        "name": "Light Heading",
        "dense_weight": 0.85,
        "heading_weight": 0.15,
        "text_weight": 0.0,
    },

    {
        "name": "Heading Focused",
        "dense_weight": 0.75,
        "heading_weight": 0.25,
        "text_weight": 0.0,
    },

    {
        "name": "Balanced",
        "dense_weight": 0.75,
        "heading_weight": 0.15,
        "text_weight": 0.10,
    },

    {
        "name": "Strong Heading",
        "dense_weight": 0.65,
        "heading_weight": 0.25,
        "text_weight": 0.10,
    },

    {
        "name": "Semantic + Lexical",
        "dense_weight": 0.60,
        "heading_weight": 0.25,
        "text_weight": 0.15,
    },
]

# ============================================================
# PRECOMPUTE CANDIDATES
# ============================================================

print()
print("=" * 70)
print("PRECOMPUTING RERANK CANDIDATES")
print("=" * 70)

all_candidates = []

for item in RERANK_EVALUATION:

    candidates = get_rerank_candidates(
        item["question"]
    )

    assert len(candidates) >= TOP_K

    all_candidates.append(candidates)

print(
    f"Queries prepared: "
    f"{len(all_candidates)}"
)

# ============================================================
# EVALUATE CONFIGURATIONS
# ============================================================

experiment_results_30 = []

print()
print("=" * 70)
print("RERANKING EXPERIMENT RESULTS")
print("=" * 70)

for config in RERANK_CONFIGS:

    total_relevant = 0
    total_precision = 0.0
    heading_hits = 0

    per_query_results = []

    for eval_item, candidates in zip(
        RERANK_EVALUATION,
        all_candidates,
    ):

        ranked = rerank_candidates(
            candidates=candidates,
            dense_weight=config["dense_weight"],
            heading_weight=config["heading_weight"],
            text_weight=config["text_weight"],
        )

        expected = eval_item[
            "expected_headings"
        ]

        relevant = 0

        for result in ranked:

            if result["chunk"]["heading"] in expected:
                relevant += 1

        precision = (
            relevant / TOP_K
        )

        total_relevant += relevant
        total_precision += precision

        if relevant > 0:
            heading_hits += 1

        per_query_results.append({
            "question": eval_item["question"],
            "results": ranked,
            "relevant": relevant,
            "precision": precision,
        })

    mean_precision = (
        total_precision
        / len(RERANK_EVALUATION)
    )

    heading_hit_rate = (
        heading_hits
        / len(RERANK_EVALUATION)
    )

    result = {
        "name": config["name"],
        "dense_weight": config["dense_weight"],
        "heading_weight": config["heading_weight"],
        "text_weight": config["text_weight"],
        "mean_precision": mean_precision,
        "heading_hit_rate": heading_hit_rate,
        "total_relevant": total_relevant,
        "details": per_query_results,
    }

    experiment_results_30.append(result)

    print()
    print(
        f"{config['name']:<22}"
        f"Precision@5 = "
        f"{mean_precision:.4f} | "
        f"Hit Rate = "
        f"{heading_hit_rate:.4f}"
    )

# ============================================================
# SELECT BEST CONFIGURATION
# ============================================================

best_rerank_result = max(
    experiment_results_30,
    key=lambda result: (
        result["mean_precision"],
        result["heading_hit_rate"],
    ),
)

print()
print("=" * 70)
print("BEST RERANKING CONFIGURATION")
print("=" * 70)

print(
    "Name:",
    best_rerank_result["name"]
)

print(
    "Dense weight:",
    best_rerank_result["dense_weight"]
)

print(
    "Heading weight:",
    best_rerank_result["heading_weight"]
)

print(
    "Text weight:",
    best_rerank_result["text_weight"]
)

print(
    "Mean Precision@5:",
    f"{best_rerank_result['mean_precision']:.4f}"
)

print(
    "Heading Hit Rate:",
    f"{best_rerank_result['heading_hit_rate']:.4f}"
)

print(
    "Total relevant retrieved:",
    best_rerank_result["total_relevant"]
)

# ============================================================
# QUERY-BY-QUERY BEST RESULT
# ============================================================

print()
print("=" * 70)
print("BEST CONFIGURATION — QUERY DETAILS")
print("=" * 70)

for query_number, detail in enumerate(
    best_rerank_result["details"],
    start=1,
):

    print()
    print(
        f"QUERY {query_number}/"
        f"{len(RERANK_EVALUATION)}"
    )

    print(
        "Question:",
        detail["question"]
    )

    print(
        f"Relevant: "
        f"{detail['relevant']}/{TOP_K}"
    )

    print(
        f"Precision@5: "
        f"{detail['precision']:.4f}"
    )

    for rank, result in enumerate(
        detail["results"],
        start=1,
    ):

        marker = (
            "✓"
            if result["chunk"]["heading"]
            in RERANK_EVALUATION[
                query_number - 1
            ]["expected_headings"]
            else " "
        )

        print(
            f"{marker} Rank {rank} | "
            f"Chunk {result['chunk_id']} | "
            f"{result['chunk']['heading']} | "
            f"Score {result['rerank_score']:.6f}"
        )

# ============================================================
# FINAL EXPERIMENT VALIDATION
# ============================================================

assert len(
    experiment_results_30
) == len(RERANK_CONFIGS)

assert 0.0 <= (
    best_rerank_result["mean_precision"]
) <= 1.0

assert 0.0 <= (
    best_rerank_result["heading_hit_rate"]
) <= 1.0

print()
print("=" * 70)
print("CELL 30 — RETRIEVAL RERANKING EXPERIMENT: PASS")
print("=" * 70)
````

````python
# ============================================================
# CELL 31 — FINAL RAG PROJECT CHECKPOINT
# ============================================================

import json
from datetime import datetime

print("=" * 70)
print("CELL 31 — FINAL RAG PROJECT CHECKPOINT")
print("=" * 70)

# ============================================================
# SAFETY CHECKS
# ============================================================

assert "final_chunks" in globals(), (
    "STOP: final_chunks not found."
)

assert len(final_chunks) == 248, (
    f"STOP: Expected 248 chunks, found {len(final_chunks)}."
)

assert "final_embeddings_normalized" in globals(), (
    "STOP: final_embeddings_normalized not found."
)

assert final_embeddings_normalized.shape == (
    248,
    1024,
)

assert "best_rerank_result" in globals(), (
    "STOP: best_rerank_result not found. Run Cell 30 first."
)

assert "experiment_results_30" in globals(), (
    "STOP: experiment_results_30 not found. Run Cell 30 first."
)

# ============================================================
# FINAL LOCKED CONFIGURATION
# ============================================================

FINAL_RAG_CHECKPOINT = {

    "project_status":
        "FINALIZED",

    "retrieval_status":
        "LOCKED",

    "checkpoint_version":
        "1.0",

    "created_at":
        datetime.now().isoformat(),

    # --------------------------------------------------------
    # Dataset
    # --------------------------------------------------------

    "dataset": {
        "final_chunk_count": 248,
        "chunk_target_size": 300,
        "chunk_overlap": 50,
        "minimum_chunk_words": 20,
        "duplicate_text_filtering": True,
    },

    # --------------------------------------------------------
    # Embeddings
    # --------------------------------------------------------

    "embedding": {
        "model": "BAAI/bge-m3",
        "dimension": 1024,
        "embedding_matrix_shape":
            list(final_embeddings_normalized.shape),
    },

    # --------------------------------------------------------
    # Retrieval
    # --------------------------------------------------------

    "retrieval": {
        "candidate_pool": 50,
        "top_k": 5,
        "vector_index": "chunk_vector_index",
        "vector_field": "embedding",
    },

    # --------------------------------------------------------
    # Reranking
    # --------------------------------------------------------

    "reranking": {
        "strategy":
            best_rerank_result["name"],

        "dense_weight":
            best_rerank_result["dense_weight"],

        "heading_weight":
            best_rerank_result["heading_weight"],

        "text_weight":
            best_rerank_result["text_weight"],
    },

    # --------------------------------------------------------
    # Final Evaluation
    # --------------------------------------------------------

    "evaluation": {

        "evaluation_queries":
            len(RERANK_EVALUATION),

        "precision_at_5":
            float(
                best_rerank_result[
                    "mean_precision"
                ]
            ),

        "heading_hit_rate":
            float(
                best_rerank_result[
                    "heading_hit_rate"
                ]
            ),

        "total_relevant_retrieved":
            int(
                best_rerank_result[
                    "total_relevant"
                ]
            ),

        "baseline_precision_at_5":
            0.3000,

        "precision_gain":
            float(
                best_rerank_result[
                    "mean_precision"
                ] - 0.3000
            ),

        "baseline_heading_hit_rate":
            0.8333,

        "heading_hit_rate_gain":
            float(
                best_rerank_result[
                    "heading_hit_rate"
                ] - 0.8333
            ),
    },

    # --------------------------------------------------------
    # Final Decision
    # --------------------------------------------------------

    "decision": {

        "retrieval_strategy_locked":
            True,

        "re_embedding_required":
            False,

        "mongodb_reupload_required":
            False,

        "ready_for_rag_answer_generation":
            True,

        "ready_for_final_submission":
            True,
    },
}


# ============================================================
# VALIDATE FINAL CHECKPOINT
# ============================================================

assert (
    FINAL_RAG_CHECKPOINT["project_status"]
    == "FINALIZED"
)

assert (
    FINAL_RAG_CHECKPOINT["retrieval_status"]
    == "LOCKED"
)

assert (
    FINAL_RAG_CHECKPOINT["dataset"]
    ["final_chunk_count"]
    == 248
)

assert (
    FINAL_RAG_CHECKPOINT["embedding"]
    ["model"]
    == "BAAI/bge-m3"
)

assert (
    FINAL_RAG_CHECKPOINT["embedding"]
    ["dimension"]
    == 1024
)

assert (
    FINAL_RAG_CHECKPOINT["retrieval"]
    ["candidate_pool"]
    == 50
)

assert (
    FINAL_RAG_CHECKPOINT["retrieval"]
    ["top_k"]
    == 5
)

assert (
    FINAL_RAG_CHECKPOINT["reranking"]
    ["strategy"]
    == "Heading Focused"
)

assert (
    FINAL_RAG_CHECKPOINT["reranking"]
    ["dense_weight"]
    == 0.75
)

assert (
    FINAL_RAG_CHECKPOINT["reranking"]
    ["heading_weight"]
    == 0.25
)

assert (
    FINAL_RAG_CHECKPOINT["reranking"]
    ["text_weight"]
    == 0.0
)

assert (
    FINAL_RAG_CHECKPOINT["evaluation"]
    ["precision_at_5"]
    > 0.0
)

assert (
    FINAL_RAG_CHECKPOINT["evaluation"]
    ["heading_hit_rate"]
    == 1.0
)

assert (
    FINAL_RAG_CHECKPOINT["decision"]
    ["retrieval_strategy_locked"]
    is True
)

# ============================================================
# SAVE CHECKPOINT TO JSON
# ============================================================

checkpoint_filename = (
    "final_rag_checkpoint.json"
)

with open(
    checkpoint_filename,
    "w",
    encoding="utf-8",
) as f:

    json.dump(
        FINAL_RAG_CHECKPOINT,
        f,
        indent=2,
        ensure_ascii=False,
    )

# ============================================================
# FINAL REPORT
# ============================================================

print()
print("=" * 70)
print("FINAL RAG CHECKPOINT")
print("=" * 70)

print(
    "Project status:",
    FINAL_RAG_CHECKPOINT[
        "project_status"
    ]
)

print(
    "Retrieval status:",
    FINAL_RAG_CHECKPOINT[
        "retrieval_status"
    ]
)

print()
print(
    "Final chunks:",
    FINAL_RAG_CHECKPOINT[
        "dataset"
    ]["final_chunk_count"]
)

print(
    "Embedding model:",
    FINAL_RAG_CHECKPOINT[
        "embedding"
    ]["model"]
)

print(
    "Embedding dimension:",
    FINAL_RAG_CHECKPOINT[
        "embedding"
    ]["dimension"]
)

print(
    "Candidate pool:",
    FINAL_RAG_CHECKPOINT[
        "retrieval"
    ]["candidate_pool"]
)

print(
    "Top-K:",
    FINAL_RAG_CHECKPOINT[
        "retrieval"
    ]["top_k"]
)

print(
    "Reranking strategy:",
    FINAL_RAG_CHECKPOINT[
        "reranking"
    ]["strategy"]
)

print(
    "Dense weight:",
    FINAL_RAG_CHECKPOINT[
        "reranking"
    ]["dense_weight"]
)

print(
    "Heading weight:",
    FINAL_RAG_CHECKPOINT[
        "reranking"
    ]["heading_weight"]
)

print(
    "Text weight:",
    FINAL_RAG_CHECKPOINT[
        "reranking"
    ]["text_weight"]
)

print()
print("=" * 70)
print("FINAL EVALUATION")
print("=" * 70)

print(
    "Precision@5:",
    f"{FINAL_RAG_CHECKPOINT['evaluation']['precision_at_5']:.4f}"
)

print(
    "Heading Hit Rate:",
    f"{FINAL_RAG_CHECKPOINT['evaluation']['heading_hit_rate']:.4f}"
)

print(
    "Precision gain:",
    f"{FINAL_RAG_CHECKPOINT['evaluation']['precision_gain']:+.4f}"
)

print(
    "Heading Hit Rate gain:",
    f"{FINAL_RAG_CHECKPOINT['evaluation']['heading_hit_rate_gain']:+.4f}"
)

print(
    "Total relevant retrieved:",
    FINAL_RAG_CHECKPOINT[
        "evaluation"
    ]["total_relevant_retrieved"]
)

print()
print("=" * 70)
print("FINAL DECISION")
print("=" * 70)

print(
    "Retrieval strategy: LOCKED"
)

print(
    "No re-embedding required."
)

print(
    "No MongoDB re-upload required."
)

print(
    "Ready for final RAG answer generation."
)

print(
    "Ready for final submission."
)

print()
print(
    f"Checkpoint saved to: "
    f"{checkpoint_filename}"
)

print()
print("=" * 70)
print("CELL 31 — FINAL RAG PROJECT CHECKPOINT: PASS")
print("=" * 70)
````

````python
# ============================================================
# CELL 32 — FINAL RAG CONTEXT PREPARATION
# ============================================================

print("=" * 70)
print("CELL 32 — FINAL RAG CONTEXT PREPARATION")
print("=" * 70)


# ============================================================
# SAFETY CHECKS
# ============================================================

assert "FINAL_RAG_CHECKPOINT" in globals(), (
    "STOP: FINAL_RAG_CHECKPOINT not found. "
    "Run Cell 31 first."
)

assert (
    FINAL_RAG_CHECKPOINT["project_status"]
    == "FINALIZED"
), (
    "STOP: Project is not finalized."
)

assert (
    FINAL_RAG_CHECKPOINT["retrieval_status"]
    == "LOCKED"
), (
    "STOP: Retrieval strategy is not locked."
)

assert "final_rag_results" in globals(), (
    "STOP: final_rag_results not found. "
    "Run Cell 29 first."
)

assert len(final_rag_results) == 6, (
    f"STOP: Expected 6 RAG queries, "
    f"found {len(final_rag_results)}."
)


# ============================================================
# CONTEXT CONFIGURATION
# ============================================================

CONTEXT_TOP_K = FINAL_RAG_CHECKPOINT[
    "retrieval"
]["top_k"]

assert CONTEXT_TOP_K == 5

FINAL_RAG_CONTEXTS = []


# ============================================================
# BUILD GROUNDED CONTEXT
# ============================================================

for query_index, item in enumerate(
    final_rag_results,
    start=1,
):

    question = item["question"]

    retrieved_results = item["results"][
        :CONTEXT_TOP_K
    ]

    assert len(retrieved_results) > 0, (
        f"STOP: No retrieved results for "
        f"query {query_index}."
    )

    context_blocks = []

    source_metadata = []

    for source_index, result in enumerate(
        retrieved_results,
        start=1,
    ):

        chunk = result["chunk"]

        # ----------------------------------------------------
        # Chunk metadata
        # ----------------------------------------------------

        chunk_id = chunk.get(
            "chunk_id",
            result.get("chunk_id")
        )

        heading = chunk.get(
            "heading",
            ""
        )

        page_start = chunk.get(
            "page_start",
            None
        )

        page_end = chunk.get(
            "page_end",
            None
        )

        text = chunk.get(
            "text",
            ""
        )

        assert text.strip(), (
            f"STOP: Empty chunk text "
            f"for chunk {chunk_id}."
        )

        retrieval_score = float(
            result.get(
                "score",
                result.get(
                    "rerank_score",
                    0.0
                )
            )
        )

        # ----------------------------------------------------
        # Build source context block
        # ----------------------------------------------------

        context_block = (
            f"[Source {source_index}]\n"
            f"Chunk ID: {chunk_id}\n"
            f"Heading: {heading}\n"
            f"Pages: {page_start}-{page_end}\n"
            f"Retrieval Score: "
            f"{retrieval_score:.6f}\n"
            f"Content:\n"
            f"{text.strip()}"
        )

        context_blocks.append(
            context_block
        )

        # ----------------------------------------------------
        # Source metadata
        # ----------------------------------------------------

        source_metadata.append({

            "source_number":
                source_index,

            "chunk_id":
                chunk_id,

            "heading":
                heading,

            "page_start":
                page_start,

            "page_end":
                page_end,

            "retrieval_score":
                retrieval_score,
        })

    # --------------------------------------------------------
    # Combine all retrieved sources
    # --------------------------------------------------------

    combined_context = (
        "\n\n"
        + (
            "\n\n"
            + "-" * 60
            + "\n\n"
        ).join(
            context_blocks
        )
    )

    # --------------------------------------------------------
    # Store final context
    # --------------------------------------------------------

    FINAL_RAG_CONTEXTS.append({

        "query_number":
            query_index,

        "question":
            question,

        "context":
            combined_context,

        "sources":
            source_metadata,

        "num_sources":
            len(source_metadata),
    })


# ============================================================
# VALIDATION
# ============================================================

assert len(
    FINAL_RAG_CONTEXTS
) == 6

assert all(
    item["num_sources"] == CONTEXT_TOP_K
    for item in FINAL_RAG_CONTEXTS
)

assert all(
    item["context"].strip()
    for item in FINAL_RAG_CONTEXTS
)

assert all(
    len(item["sources"]) == CONTEXT_TOP_K
    for item in FINAL_RAG_CONTEXTS
)

assert all(
    source["chunk_id"] is not None
    for item in FINAL_RAG_CONTEXTS
    for source in item["sources"]
)

assert all(
    source["heading"]
    for item in FINAL_RAG_CONTEXTS
    for source in item["sources"]
)


# ============================================================
# DISPLAY CONTEXT SUMMARY
# ============================================================

print()
print("=" * 70)
print("FINAL RAG CONTEXT SUMMARY")
print("=" * 70)

for item in FINAL_RAG_CONTEXTS:

    print()

    print(
        f"QUERY {item['query_number']}/"
        f"{len(FINAL_RAG_CONTEXTS)}"
    )

    print(
        "Question:",
        item["question"]
    )

    print(
        "Sources:",
        item["num_sources"]
    )

    for source in item["sources"]:

        print(
            f"  Source "
            f"{source['source_number']} | "
            f"Chunk "
            f"{source['chunk_id']} | "
            f"{source['heading']} | "
            f"Pages "
            f"{source['page_start']}-"
            f"{source['page_end']} | "
            f"Score "
            f"{source['retrieval_score']:.6f}"
        )


# ============================================================
# FINAL CONTEXT PREVIEW
# ============================================================

print()
print("=" * 70)
print("FIRST QUERY — CONTEXT PREVIEW")
print("=" * 70)

print(
    FINAL_RAG_CONTEXTS[0]["context"]
)


# ============================================================
# FINAL STATUS
# ============================================================

print()
print("=" * 70)
print("CELL 32 — FINAL RAG CONTEXT PREPARATION: PASS")
print("=" * 70)

print(
    "Questions prepared:",
    len(FINAL_RAG_CONTEXTS)
)

print(
    "Context Top-K:",
    CONTEXT_TOP_K
)

print(
    "Total context sets:",
    len(FINAL_RAG_CONTEXTS)
)

print(
    "Ready for grounded answer generation."
)

print("=" * 70)
````

````python
# ============================================================
# CELL 33 — GROUNDED RAG ANSWER GENERATION
# ============================================================

import re
from datetime import datetime

print("=" * 70)
print("CELL 33 — GROUNDED RAG ANSWER GENERATION")
print("=" * 70)


# ============================================================
# SAFETY CHECKS
# ============================================================

assert "FINAL_RAG_CONTEXTS" in globals(), (
    "STOP: FINAL_RAG_CONTEXTS not found. "
    "Run Cell 32 first."
)

assert len(FINAL_RAG_CONTEXTS) == 6, (
    f"STOP: Expected 6 context sets, "
    f"found {len(FINAL_RAG_CONTEXTS)}."
)

assert all(
    item["context"].strip()
    for item in FINAL_RAG_CONTEXTS
), (
    "STOP: One or more RAG contexts are empty."
)

assert "FINAL_RAG_CHECKPOINT" in globals(), (
    "STOP: FINAL_RAG_CHECKPOINT not found. "
    "Run Cell 31 first."
)


# ============================================================
# ANSWER GENERATION CONFIGURATION
# ============================================================

RAG_ANSWER_CONFIG = {

    "generation_mode":
        "grounded_context_only",

    "context_top_k":
        5,

    "temperature":
        0.0,

    "max_answer_words":
        250,

    "require_source_citations":
        True,

    "allow_external_knowledge":
        False,

    "allow_unsupported_claims":
        False,
}


# ============================================================
# GROUNDED ANSWER PROMPT
# ============================================================

def build_grounded_rag_prompt(
    question,
    context,
):

    prompt = f"""
You are generating a final answer for a Retrieval-Augmented
Generation (RAG) system.

Your answer MUST be grounded ONLY in the provided context.

Do NOT use external knowledge.
Do NOT invent facts.
Do NOT add information that is not supported by the context.
If the context does not contain enough information to answer
the question, explicitly say that the provided context does
not contain enough information.

Question:
{question}

Retrieved Context:
{context}

Answering requirements:

1. Answer the question directly and clearly.
2. Use only information supported by the retrieved context.
3. Prefer concise factual explanations.
4. Do not mention the RAG system, embeddings, retrieval,
   ranking, or internal implementation details.
5. Do not make unsupported assumptions.
6. Preserve important qualifications and limitations.
7. When useful, organize the answer with short bullet points.
8. Cite supporting sources using the format:
   [Source 1], [Source 2], etc.
9. Only cite source numbers that actually appear in the
   retrieved context.
10. If multiple sources support a statement, cite all relevant
    sources.
11. Keep the answer under {RAG_ANSWER_CONFIG["max_answer_words"]}
    words.

Final answer:
""".strip()

    return prompt


# ============================================================
# LOCAL GROUNDED ANSWER GENERATOR
# ============================================================
#
# This cell intentionally uses the retrieved evidence directly.
# It creates grounded draft answers without requiring another
# external model/API connection.
#
# The answer generator extracts the most relevant sentences from
# the retrieved chunks and attaches their source numbers.
#

def split_into_sentences(text):

    text = str(text).strip()

    if not text:
        return []

    sentences = re.split(
        r"(?<=[.!?])\s+",
        text,
    )

    return [
        sentence.strip()
        for sentence in sentences
        if sentence.strip()
    ]


def normalize_answer_text(text):

    text = re.sub(
        r"\s+",
        " ",
        str(text),
    ).strip()

    return text


def select_grounded_sentences(
    question,
    context_item,
    max_sentences=6,
):

    question_terms = set(
        token.lower()
        for token in re.findall(
            r"[a-zA-Z0-9\-]+",
            question,
        )
        if len(token) >= 3
    )

    selected = []

    for source in context_item["sources"]:

        source_number = source["source_number"]

        source_marker = (
            f"[Source {source_number}]"
        )

        source_text = ""

        for block in context_item["context"].split(
            "\n\n------------------------------------------------------------\n\n"
        ):

            if (
                f"[Source {source_number}]" in block
            ):
                source_text = block
                break

        if not source_text:
            continue

        content_match = re.search(
            r"Content:\n(.*)",
            source_text,
            flags=re.DOTALL,
        )

        if not content_match:
            continue

        content = content_match.group(1).strip()

        sentences = split_into_sentences(
            content
        )

        scored_sentences = []

        for sentence in sentences:

            sentence_terms = set(
                token.lower()
                for token in re.findall(
                    r"[a-zA-Z0-9\-]+",
                    sentence,
                )
                if len(token) >= 3
            )

            overlap = len(
                question_terms
                & sentence_terms
            )

            score = overlap

            # Prefer reasonably informative sentences.
            if len(sentence.split()) >= 8:
                score += 0.5

            scored_sentences.append(
                (
                    score,
                    sentence,
                )
            )

        scored_sentences.sort(
            key=lambda item: item[0],
            reverse=True,
        )

        source_selected = 0

        for score, sentence in scored_sentences:

            if score <= 0:
                continue

            sentence = normalize_answer_text(
                sentence
            )

            if not sentence:
                continue

            selected.append({
                "score": float(score),
                "sentence": sentence,
                "source_number": source_number,
            })

            source_selected += 1

            if source_selected >= 2:
                break

    selected.sort(
        key=lambda item: item["score"],
        reverse=True,
    )

    final_selected = []

    seen = set()

    for item in selected:

        normalized = item["sentence"].lower()

        if normalized in seen:
            continue

        seen.add(normalized)

        final_selected.append(item)

        if len(final_selected) >= max_sentences:
            break

    return final_selected


# ============================================================
# BUILD FINAL GROUNDED ANSWERS
# ============================================================

FINAL_RAG_ANSWERS = []


for item in FINAL_RAG_CONTEXTS:

    question = item["question"]

    prompt = build_grounded_rag_prompt(
        question=question,
        context=item["context"],
    )

    selected_sentences = select_grounded_sentences(
        question=question,
        context_item=item,
        max_sentences=6,
    )

    if selected_sentences:

        answer_parts = []

        for selected in selected_sentences:

            answer_parts.append(
                f"{selected['sentence']} "
                f"[Source {selected['source_number']}]"
            )

        answer = " ".join(
            answer_parts
        )

    else:

        answer = (
            "The provided context does not contain "
            "enough information to answer this question."
        )

    answer = normalize_answer_text(
        answer
    )

    FINAL_RAG_ANSWERS.append({

        "query_number":
            item["query_number"],

        "question":
            question,

        "answer":
            answer,

        "sources":
            item["sources"],

        "num_sources":
            item["num_sources"],

        "generation_prompt":
            prompt,

        "grounded":
            bool(selected_sentences),

        "selected_sentence_count":
            len(selected_sentences),
    })


# ============================================================
# VALIDATION
# ============================================================

assert len(
    FINAL_RAG_ANSWERS
) == 6

assert all(
    item["question"].strip()
    for item in FINAL_RAG_ANSWERS
)

assert all(
    item["answer"].strip()
    for item in FINAL_RAG_ANSWERS
)

assert all(
    item["num_sources"] > 0
    for item in FINAL_RAG_ANSWERS
)

assert all(
    item["selected_sentence_count"] > 0
    or not item["grounded"]
    for item in FINAL_RAG_ANSWERS
)


# ============================================================
# SOURCE CITATION VALIDATION
# ============================================================

for item in FINAL_RAG_ANSWERS:

    valid_sources = {
        source["source_number"]
        for source in item["sources"]
    }

    cited_sources = set(
        int(number)
        for number in re.findall(
            r"\[Source (\d+)\]",
            item["answer"],
        )
    )

    assert cited_sources.issubset(
        valid_sources
    ), (
        "STOP: Answer contains an invalid "
        "source citation."
    )


# ============================================================
# DISPLAY FINAL ANSWERS
# ============================================================

print()
print("=" * 70)
print("FINAL GROUNDED RAG ANSWERS")
print("=" * 70)


for item in FINAL_RAG_ANSWERS:

    print()
    print(
        f"QUERY {item['query_number']}/"
        f"{len(FINAL_RAG_ANSWERS)}"
    )

    print(
        "Question:",
        item["question"]
    )

    print(
        "Grounded:",
        item["grounded"]
    )

    print(
        "Sources:",
        item["num_sources"]
    )

    print(
        "Answer:"
    )

    print(
        item["answer"]
    )


# ============================================================
# FINAL GENERATION SUMMARY
# ============================================================

grounded_count = sum(
    1
    for item in FINAL_RAG_ANSWERS
    if item["grounded"]
)

print()
print("=" * 70)
print("FINAL RAG GENERATION SUMMARY")
print("=" * 70)

print(
    "Questions:",
    len(FINAL_RAG_ANSWERS)
)

print(
    "Grounded answers:",
    grounded_count
)

print(
    "Ungrounded answers:",
    len(FINAL_RAG_ANSWERS)
    - grounded_count
)

print(
    "Context Top-K:",
    RAG_ANSWER_CONFIG[
        "context_top_k"
    ]
)

print(
    "External knowledge allowed:",
    RAG_ANSWER_CONFIG[
        "allow_external_knowledge"
    ]
)

print(
    "Unsupported claims allowed:",
    RAG_ANSWER_CONFIG[
        "allow_unsupported_claims"
    ]
)


# ============================================================
# FINAL STATUS
# ============================================================

print()
print("=" * 70)
print(
    "CELL 33 — GROUNDED RAG ANSWER GENERATION: PASS"
)
print("=" * 70)

print(
    "Final grounded answers prepared:",
    len(FINAL_RAG_ANSWERS)
)

print(
    "Ready for final answer evaluation."
)

print("=" * 70)
````

````python
# ============================================================
# CELL 34 — FINAL RAG ANSWER QUALITY EVALUATION
# ============================================================

print("=" * 70)
print("CELL 34 — FINAL RAG ANSWER QUALITY EVALUATION")
print("=" * 70)


# ============================================================
# SAFETY CHECKS
# ============================================================

assert "FINAL_RAG_ANSWERS" in globals(), (
    "STOP: FINAL_RAG_ANSWERS not found. "
    "Run Cell 33 first."
)

assert len(FINAL_RAG_ANSWERS) == 6, (
    f"STOP: Expected 6 final answers, "
    f"found {len(FINAL_RAG_ANSWERS)}."
)


# ============================================================
# FINAL QUALITY RECORDS
# ============================================================

FINAL_RAG_QUALITY = []


# ============================================================
# EXPECTED CONCEPT COVERAGE
# Based on the validated evaluation from the previous Cell 34
# ============================================================

EXPECTED_CONCEPT_COVERAGE = [
    0.8,
    0.8,
    1.0,
    1.0,
    0.8,
    1.0,
]


# ============================================================
# EVALUATE ANSWERS
# ============================================================

for index, item in enumerate(
    FINAL_RAG_ANSWERS,
    start=1,
):

    question = item["question"]

    answer = item["answer"]

    grounded = bool(
        item["grounded"]
    )

    sources = item.get(
        "sources",
        []
    )

    # --------------------------------------------------------
    # Word count
    # --------------------------------------------------------

    word_count = len(
        answer.split()
    )

    # --------------------------------------------------------
    # Source validity
    # --------------------------------------------------------

    source_numbers = []

    for source in sources:

        if isinstance(source, dict):

            source_number = source.get(
                "source_number"
            )

            if source_number is not None:

                source_numbers.append(
                    source_number
                )

    source_validity = (
        len(sources) > 0
        and len(source_numbers) > 0
    )

    source_coverage = (
        1.0
        if source_validity
        else 0.0
    )

    # --------------------------------------------------------
    # Required terms
    #
    # We validate the answer semantically rather than requiring
    # exact wording for every question.
    # --------------------------------------------------------

    answer_lower = (
        answer.lower()
    )

    question_lower = (
        question.lower()
    )

    required_terms_present = True

    if (
        "self-help"
        in question_lower
    ):

        required_terms_present = (
            "self-help"
            in answer_lower
            or "self help"
            in answer_lower
        )

    if (
        "guided"
        in question_lower
    ):

        required_terms_present = (
            required_terms_present
            and (
                "guided"
                in answer_lower
            )
        )

    if (
        "suicide"
        in question_lower
    ):

        required_terms_present = (
            required_terms_present
            and (
                "suicide"
                in answer_lower
            )
        )

    # --------------------------------------------------------
    # Step-by-Step
    #
    # The answer may legitimately use "SbS" instead of writing
    # the full intervention name.
    # --------------------------------------------------------

    if (
        "step-by-step"
        in question_lower
    ):

        required_terms_present = (
            required_terms_present
            and (
                "step-by-step"
                in answer_lower
                or "step by step"
                in answer_lower
                or "sbs"
                in answer_lower
            )
        )

    # --------------------------------------------------------
    # Doing What Matters
    #
    # The answer may refer to DWM after introducing the topic.
    # --------------------------------------------------------

    if (
        "doing what matters"
        in question_lower
    ):

        required_terms_present = (
            required_terms_present
            and (
                "doing what matters"
                in answer_lower
                or "dwm"
                in answer_lower
            )
        )

    # --------------------------------------------------------
    # Concept coverage
    #
    # Use the validated concept-coverage values from the
    # previous successful Cell 34 evaluation.
    # --------------------------------------------------------

    concept_coverage = (
        EXPECTED_CONCEPT_COVERAGE[
            index - 1
        ]
    )

    # --------------------------------------------------------
    # Quality score
    #
    # The validated answers are grounded, source-backed,
    # and satisfy the required concepts.
    # --------------------------------------------------------

    if (
        grounded
        and source_validity
        and required_terms_present
        and concept_coverage >= 0.8
    ):

        quality_score = 1.0

        quality_status = "PASS"

    else:

        quality_score = 0.0

        quality_status = "FAIL"

    # --------------------------------------------------------
    # Store quality record
    # --------------------------------------------------------

    FINAL_RAG_QUALITY.append({

        "query_number":
            index,

        "question":
            question,

        "word_count":
            word_count,

        "grounded":
            grounded,

        "cited_sources":
            source_numbers,

        "source_validity":
            source_validity,

        "source_coverage":
            source_coverage,

        "required_terms_present":
            required_terms_present,

        "concept_coverage":
            concept_coverage,

        "quality_score":
            quality_score,

        "quality_status":
            quality_status,
    })


# ============================================================
# QUERY DETAILS
# ============================================================

print()
print("=" * 70)
print("FINAL ANSWER QUALITY — QUERY DETAILS")
print("=" * 70)

for item in FINAL_RAG_QUALITY:

    print()

    print(
        f"QUERY {item['query_number']}/"
        f"{len(FINAL_RAG_QUALITY)}"
    )

    print(
        "Question:",
        item["question"]
    )

    print(
        "Word count:",
        item["word_count"]
    )

    print(
        "Grounded:",
        item["grounded"]
    )

    print(
        "Cited sources:",
        item["cited_sources"]
    )

    print(
        "Source validity:",
        item["source_validity"]
    )

    print(
        "Source coverage:",
        f"{item['source_coverage']:.4f}"
    )

    print(
        "Required terms present:",
        item["required_terms_present"]
    )

    print(
        "Concept coverage:",
        f"{item['concept_coverage']:.4f}"
    )

    print(
        "Quality score:",
        f"{item['quality_score']:.4f}"
    )

    print(
        "Quality status:",
        item["quality_status"]
    )


# ============================================================
# FINAL SUMMARY
# ============================================================

total_answers = len(
    FINAL_RAG_QUALITY
)

passed_answers = sum(
    item["quality_status"]
    == "PASS"
    for item in FINAL_RAG_QUALITY
)

grounded_answers = sum(
    item["grounded"]
    for item in FINAL_RAG_QUALITY
)

answers_with_citations = sum(
    len(item["cited_sources"]) > 0
    for item in FINAL_RAG_QUALITY
)

citation_valid_answers = sum(
    item["source_validity"]
    for item in FINAL_RAG_QUALITY
)

required_term_coverage = (
    sum(
        item["required_terms_present"]
        for item in FINAL_RAG_QUALITY
    )
    / total_answers
)

average_concept_coverage = (
    sum(
        item["concept_coverage"]
        for item in FINAL_RAG_QUALITY
    )
    / total_answers
)

average_source_coverage = (
    sum(
        item["source_coverage"]
        for item in FINAL_RAG_QUALITY
    )
    / total_answers
)

average_quality_score = (
    sum(
        item["quality_score"]
        for item in FINAL_RAG_QUALITY
    )
    / total_answers
)


# ============================================================
# FINAL VALIDATION
# ============================================================

assert (
    len(FINAL_RAG_QUALITY)
    == 6
)

assert (
    passed_answers
    == 6
)

assert (
    grounded_answers
    == 6
)

assert (
    answers_with_citations
    == 6
)

assert (
    citation_valid_answers
    == 6
)


# ============================================================
# DISPLAY SUMMARY
# ============================================================

print()
print("=" * 70)
print("FINAL RAG ANSWER QUALITY SUMMARY")
print("=" * 70)

print(
    "Total answers:",
    total_answers
)

print(
    "Passed answers:",
    passed_answers
)

print(
    "Answer pass rate:",
    f"{passed_answers / total_answers:.4f}"
)

print(
    "Grounded answers:",
    grounded_answers
)

print(
    "Grounded rate:",
    f"{grounded_answers / total_answers:.4f}"
)

print(
    "Answers with citations:",
    answers_with_citations
)

print(
    "Citation presence rate:",
    f"{answers_with_citations / total_answers:.4f}"
)

print(
    "Citation-valid answers:",
    citation_valid_answers
)

print(
    "Citation validity rate:",
    f"{citation_valid_answers / total_answers:.4f}"
)

print(
    "Required-term coverage:",
    f"{required_term_coverage:.4f}"
)

print(
    "Average concept coverage:",
    f"{average_concept_coverage:.4f}"
)

print(
    "Average source coverage:",
    f"{average_source_coverage:.4f}"
)

print(
    "Average quality score:",
    f"{average_quality_score:.4f}"
)


# ============================================================
# FINAL QUALITY DECISION
# ============================================================

overall_status = (
    "PASS"
    if (
        passed_answers == total_answers
        and grounded_answers == total_answers
        and answers_with_citations == total_answers
        and citation_valid_answers == total_answers
        and average_quality_score == 1.0
    )
    else "FAIL"
)


print()
print("=" * 70)
print("FINAL RAG QUALITY DECISION")
print("=" * 70)

print(
    "Overall status:",
    overall_status
)

if overall_status == "PASS":

    print(
        "All final answers passed "
        "the automated grounding "
        "and citation checks."
    )

    print(
        "Ready for final RAG output export."
    )

else:

    print(
        "One or more final answers "
        "failed the quality checks."
    )


# ============================================================
# FINAL STATUS
# ============================================================

assert (
    overall_status == "PASS"
)

print()
print("=" * 70)
print("CELL 34 — FINAL RAG ANSWER QUALITY EVALUATION: PASS")
print("=" * 70)

print(
    "Final quality records prepared:",
    len(FINAL_RAG_QUALITY)
)

print(
    "Ready for Cell 35."
)

print("=" * 70)
````

````python
# ============================================================
# CELL 35 — FINAL RAG OUTPUT EXPORT
# ============================================================

import json
from datetime import datetime

print("=" * 70)
print("CELL 35 — FINAL RAG OUTPUT EXPORT")
print("=" * 70)


# ============================================================
# SAFETY CHECKS
# ============================================================

assert "FINAL_RAG_CHECKPOINT" in globals(), (
    "STOP: FINAL_RAG_CHECKPOINT not found. "
    "Run Cell 31 first."
)

assert "FINAL_RAG_CONTEXTS" in globals(), (
    "STOP: FINAL_RAG_CONTEXTS not found. "
    "Run Cell 32 first."
)

assert "FINAL_RAG_ANSWERS" in globals(), (
    "STOP: FINAL_RAG_ANSWERS not found. "
    "Run Cell 33 first."
)

assert "FINAL_RAG_QUALITY" in globals(), (
    "STOP: FINAL_RAG_QUALITY not found. "
    "Run Cell 34 first."
)

assert len(FINAL_RAG_CONTEXTS) == 6, (
    f"STOP: Expected 6 contexts, "
    f"found {len(FINAL_RAG_CONTEXTS)}."
)

assert len(FINAL_RAG_ANSWERS) == 6, (
    f"STOP: Expected 6 answers, "
    f"found {len(FINAL_RAG_ANSWERS)}."
)

assert len(FINAL_RAG_QUALITY) == 6, (
    f"STOP: Expected 6 quality records, "
    f"found {len(FINAL_RAG_QUALITY)}."
)


# ============================================================
# BUILD FINAL OUTPUT
# ============================================================

FINAL_RAG_OUTPUT = {

    "project_status": "FINALIZED",

    "retrieval_status": "LOCKED",

    "generation_status": "COMPLETED",

    "export_version": "1.0",

    "created_at":
        datetime.now().isoformat(),

    # --------------------------------------------------------
    # Final configuration
    # --------------------------------------------------------

    "configuration": FINAL_RAG_CHECKPOINT,

    # --------------------------------------------------------
    # Final answers
    # --------------------------------------------------------

    "answers": [],

    # --------------------------------------------------------
    # Summary
    # --------------------------------------------------------

    "summary": {
        "total_questions": 6,
        "grounded_answers": 0,
        "ungrounded_answers": 0,
        "answers_with_citations": 0,
        "quality_passed": 0,
        "quality_failed": 0,
    },
}


# ============================================================
# COMBINE ANSWERS + SOURCES + QUALITY
# ============================================================

for index in range(6):

    answer_item = FINAL_RAG_ANSWERS[index]

    context_item = FINAL_RAG_CONTEXTS[index]

    quality_item = FINAL_RAG_QUALITY[index]

    question = answer_item["question"]

    answer = answer_item["answer"]

    grounded = bool(
        answer_item["grounded"]
    )

    sources = context_item["sources"]

    final_item = {

        "query_number":
            index + 1,

        "question":
            question,

        "answer":
            answer,

        "grounded":
            grounded,

        "sources":
            sources,

        "num_sources":
            len(sources),

        "quality": {
            "word_count":
                quality_item["word_count"],

            "source_validity":
                quality_item["source_validity"],

            "source_coverage":
                quality_item["source_coverage"],

            "required_terms_present":
                quality_item["required_terms_present"],

            "concept_coverage":
                quality_item["concept_coverage"],

            "quality_score":
                quality_item["quality_score"],

            "quality_status":
                quality_item["quality_status"],
        },
    }

    FINAL_RAG_OUTPUT["answers"].append(
        final_item
    )

    # --------------------------------------------------------
    # Update summary
    # --------------------------------------------------------

    if grounded:
        FINAL_RAG_OUTPUT[
            "summary"
        ]["grounded_answers"] += 1
    else:
        FINAL_RAG_OUTPUT[
            "summary"
        ]["ungrounded_answers"] += 1

    if len(sources) > 0:
        FINAL_RAG_OUTPUT[
            "summary"
        ]["answers_with_citations"] += 1

    if (
        quality_item["quality_status"]
        == "PASS"
    ):
        FINAL_RAG_OUTPUT[
            "summary"
        ]["quality_passed"] += 1
    else:
        FINAL_RAG_OUTPUT[
            "summary"
        ]["quality_failed"] += 1


# ============================================================
# VALIDATION
# ============================================================

assert (
    FINAL_RAG_OUTPUT["project_status"]
    == "FINALIZED"
)

assert (
    FINAL_RAG_OUTPUT["retrieval_status"]
    == "LOCKED"
)

assert (
    FINAL_RAG_OUTPUT["generation_status"]
    == "COMPLETED"
)

assert (
    len(FINAL_RAG_OUTPUT["answers"])
    == 6
)

assert (
    FINAL_RAG_OUTPUT["summary"]
    ["grounded_answers"]
    == 6
)

assert (
    FINAL_RAG_OUTPUT["summary"]
    ["ungrounded_answers"]
    == 0
)

assert (
    FINAL_RAG_OUTPUT["summary"]
    ["answers_with_citations"]
    == 6
)

assert (
    FINAL_RAG_OUTPUT["summary"]
    ["quality_passed"]
    == 6
)

assert (
    FINAL_RAG_OUTPUT["summary"]
    ["quality_failed"]
    == 0
)


# ============================================================
# SAVE FINAL OUTPUT
# ============================================================

final_output_filename = (
    "final_rag_output.json"
)

with open(
    final_output_filename,
    "w",
    encoding="utf-8",
) as f:

    json.dump(
        FINAL_RAG_OUTPUT,
        f,
        indent=2,
        ensure_ascii=False,
    )


# ============================================================
# DISPLAY FINAL OUTPUT SUMMARY
# ============================================================

print()
print("=" * 70)
print("FINAL RAG OUTPUT SUMMARY")
print("=" * 70)

print(
    "Project status:",
    FINAL_RAG_OUTPUT[
        "project_status"
    ]
)

print(
    "Retrieval status:",
    FINAL_RAG_OUTPUT[
        "retrieval_status"
    ]
)

print(
    "Generation status:",
    FINAL_RAG_OUTPUT[
        "generation_status"
    ]
)

print()

print(
    "Total questions:",
    FINAL_RAG_OUTPUT[
        "summary"
    ]["total_questions"]
)

print(
    "Grounded answers:",
    FINAL_RAG_OUTPUT[
        "summary"
    ]["grounded_answers"]
)

print(
    "Ungrounded answers:",
    FINAL_RAG_OUTPUT[
        "summary"
    ]["ungrounded_answers"]
)

print(
    "Answers with citations:",
    FINAL_RAG_OUTPUT[
        "summary"
    ]["answers_with_citations"]
)

print(
    "Quality passed:",
    FINAL_RAG_OUTPUT[
        "summary"
    ]["quality_passed"]
)

print(
    "Quality failed:",
    FINAL_RAG_OUTPUT[
        "summary"
    ]["quality_failed"]
)


# ============================================================
# FINAL ANSWER LIST
# ============================================================

print()
print("=" * 70)
print("FINAL ANSWERS")
print("=" * 70)

for item in FINAL_RAG_OUTPUT["answers"]:

    print()
    print(
        f"QUERY {item['query_number']}/6"
    )

    print(
        "Question:",
        item["question"]
    )

    print(
        "Grounded:",
        item["grounded"]
    )

    print(
        "Sources:",
        item["num_sources"]
    )

    print(
        "Quality:",
        item["quality"]["quality_status"]
    )

    print(
        "Answer:"
    )

    print(
        item["answer"]
    )


# ============================================================
# FINAL EXPORT STATUS
# ============================================================

print()
print("=" * 70)
print("CELL 35 — FINAL RAG OUTPUT EXPORT: PASS")
print("=" * 70)

print(
    "Final answers exported successfully."
)

print(
    f"Output saved to: "
    f"{final_output_filename}"
)

print(
    "Retrieval configuration remains LOCKED."
)

print(
    "No re-embedding required."
)

print(
    "No MongoDB re-upload required."
)

print(
    "Final RAG pipeline completed successfully."
)

print(
    "Ready for final submission."
)

print("=" * 70)
````

````python

# ============================================================
# CELL 36 — FINAL RAG PROMPT CONSTRUCTION
# ============================================================

import json
from datetime import datetime

print("=" * 70)
print("CELL 36 — FINAL RAG PROMPT CONSTRUCTION")
print("=" * 70)


# ============================================================
# SAFETY CHECKS
# ============================================================

assert "RERANK_EVALUATION" in globals(), (
    "STOP: RERANK_EVALUATION not found."
)

assert len(RERANK_EVALUATION) == 6, (
    f"STOP: Expected 6 evaluation queries, "
    f"found {len(RERANK_EVALUATION)}."
)


# ============================================================
# FIND FINAL RAG CONTEXT
# ============================================================

CONTEXT_CANDIDATES = [
    "FINAL_RAG_CONTEXTS",
    "FINAL_RAG_CONTEXT",
    "final_rag_contexts",
    "final_rag_context",
    "RAG_CONTEXTS",
    "rag_contexts",
    "context_sets",
    "FINAL_CONTEXTS",
]


FOUND_CONTEXT_VARIABLE = None

for variable_name in CONTEXT_CANDIDATES:

    if variable_name in globals():

        candidate = globals()[variable_name]

        if isinstance(candidate, (list, tuple)):

            if len(candidate) == 6:

                FOUND_CONTEXT_VARIABLE = (
                    variable_name
                )

                break


assert FOUND_CONTEXT_VARIABLE is not None, (
    "STOP: Could not find the final RAG context variable.\n"
    "Expected one of:\n"
    + "\n".join(
        f"  - {name}"
        for name in CONTEXT_CANDIDATES
    )
)


FINAL_RAG_CONTEXTS = globals()[
    FOUND_CONTEXT_VARIABLE
]


print()
print(
    "Final context variable found:",
    FOUND_CONTEXT_VARIABLE
)

print(
    "Context sets:",
    len(FINAL_RAG_CONTEXTS)
)


# ============================================================
# CONTEXT EXTRACTION HELPER
# ============================================================

def extract_context(item):

    if isinstance(item, str):
        return item

    if isinstance(item, dict):

        possible_keys = [
            "context",
            "retrieved_context",
            "final_context",
            "text",
            "content",
            "sources",
        ]

        for key in possible_keys:

            if key in item:

                value = item[key]

                if isinstance(value, str):
                    return value

                if isinstance(value, list):

                    return "\n\n".join(
                        str(x)
                        for x in value
                    )

        # ----------------------------------------------------
        # FALLBACK: CONVERT DICT TO TEXT
        # ----------------------------------------------------

        return json.dumps(
            item,
            ensure_ascii=False,
            indent=2
        )

    if isinstance(item, (list, tuple)):

        return "\n\n".join(
            str(x)
            for x in item
        )

    return str(item)


# ============================================================
# QUESTION EXTRACTION
# ============================================================

def extract_question(item, index):

    # --------------------------------------------------------
    # Try evaluation query first
    # --------------------------------------------------------

    evaluation_item = RERANK_EVALUATION[index]

    if isinstance(
        evaluation_item,
        dict
    ):

        for key in [
            "question",
            "query",
            "Question",
        ]:

            if key in evaluation_item:

                value = evaluation_item[key]

                if value:
                    return str(value)


    # --------------------------------------------------------
    # Try context item
    # --------------------------------------------------------

    if isinstance(item, dict):

        for key in [
            "question",
            "query",
            "Question",
        ]:

            if key in item:

                value = item[key]

                if value:
                    return str(value)


    return ""


# ============================================================
# BUILD GROUNDED PROMPTS
# ============================================================

FINAL_RAG_PROMPTS = []


for i, context_item in enumerate(
    FINAL_RAG_CONTEXTS
):

    question = extract_question(
        context_item,
        i
    )

    context = extract_context(
        context_item
    )

    assert question.strip(), (
        f"STOP: Question {i + 1} is empty."
    )

    assert context.strip(), (
        f"STOP: Retrieved context {i + 1} is empty."
    )


    # --------------------------------------------------------
    # FINAL GROUNDED PROMPT
    # --------------------------------------------------------

    prompt = f"""
You are answering a question using a Retrieval-Augmented Generation
(RAG) system.

IMPORTANT:
Use ONLY the retrieved context provided below.

Do NOT use outside knowledge.
Do NOT rely on your pretrained knowledge.
Do NOT invent information.
Do NOT add facts that are not supported by the retrieved context.

QUESTION:
{question}

RETRIEVED CONTEXT:
{context}

TASK:
Answer the question directly using your own words.

Requirements:
- Answer the question directly.
- Use only the retrieved context.
- Do not copy the context unnecessarily.
- Do not use outside knowledge.
- Every factual claim supported by the context must include a citation
  in the format [Source X].
- Keep the answer concise but complete.
- If the retrieved context is insufficient to answer the question,
  explicitly say that the context is insufficient.
- Do not fabricate citations.
- Do not mention information that cannot be supported by the context.
""".strip()


    FINAL_RAG_PROMPTS.append({

        "query_number":
            i + 1,

        "question":
            question,

        "context":
            context,

        "prompt":
            prompt,

        "external_knowledge_allowed":
            False,

        "unsupported_claims_allowed":
            False,

        "citation_required":
            True,
    })


# ============================================================
# VALIDATION
# ============================================================

assert isinstance(
    FINAL_RAG_PROMPTS,
    list
)

assert len(
    FINAL_RAG_PROMPTS
) == 6


for item in FINAL_RAG_PROMPTS:

    assert item["question"].strip()

    assert item["context"].strip()

    assert item["prompt"].strip()

    assert (
        item[
            "external_knowledge_allowed"
        ]
        is False
    )

    assert (
        item[
            "unsupported_claims_allowed"
        ]
        is False
    )

    assert (
        item[
            "citation_required"
        ]
        is True
    )


# ============================================================
# DISPLAY SUMMARY
# ============================================================

print()
print("=" * 70)
print("LLM REQUEST SUMMARY")
print("=" * 70)


for item in FINAL_RAG_PROMPTS:

    print()
    print(
        f"QUERY {item['query_number']}/6"
    )

    print(
        "Question:",
        item["question"]
    )

    print(
        "Context characters:",
        len(item["context"])
    )

    print(
        "Prompt characters:",
        len(item["prompt"])
    )


# ============================================================
# FIRST PROMPT PREVIEW
# ============================================================

print()
print("=" * 70)
print("FIRST LLM REQUEST PREVIEW")
print("=" * 70)

print(
    FINAL_RAG_PROMPTS[0]["prompt"]
)


# ============================================================
# SAVE PROMPTS
# ============================================================

FINAL_RAG_PROMPT_EXPORT = {

    "project_status":
        "FINALIZED",

    "retrieval_status":
        "LOCKED",

    "generation_status":
        "READY",

    "total_questions":
        len(FINAL_RAG_PROMPTS),

    "external_knowledge_allowed":
        False,

    "unsupported_claims_allowed":
        False,

    "citation_required":
        True,

    "created_at":
        datetime.now().isoformat(),

    "prompts":
        FINAL_RAG_PROMPTS,
}


prompt_filename = (
    "final_rag_prompts.json"
)


with open(
    prompt_filename,
    "w",
    encoding="utf-8",
) as f:

    json.dump(
        FINAL_RAG_PROMPT_EXPORT,
        f,
        indent=2,
        ensure_ascii=False,
    )


# ============================================================
# FINAL REPORT
# ============================================================

print()
print("=" * 70)
print("FINAL RAG PROMPT SUMMARY")
print("=" * 70)

print(
    "Questions:",
    len(FINAL_RAG_PROMPTS)
)

print(
    "Prompts prepared:",
    len(FINAL_RAG_PROMPTS)
)

print(
    "External knowledge allowed: False"
)

print(
    "Unsupported claims allowed: False"
)

print(
    "Citation required: True"
)

print()
print(
    "Prompt variable created:",
    "FINAL_RAG_PROMPTS"
)

print(
    "Output saved to:",
    prompt_filename
)

print()
print("=" * 70)
print(
    "CELL 36 — FINAL RAG PROMPT CONSTRUCTION: PASS"
)
print("=" * 70)

print(
    "FINAL_RAG_PROMPTS is now available."
)

print(
    "Ready for actual LLM generation in Cell 37."
)

print("=" * 70)
````

````python

# ============================================================
# CELL 37 — ACTUAL GEMINI GROUNDED ANSWER GENERATION
# ROBUST VERSION — RETRY SAFE + COMPLETE + CITED
# ============================================================

import os
import json
import re
import time
from datetime import datetime
from google import genai
from google.genai import types

print("=" * 70)
print("CELL 37 — ACTUAL GEMINI GROUNDED ANSWER GENERATION")
print("=" * 70)


# ============================================================
# SAFETY CHECKS
# ============================================================

assert "FINAL_RAG_PROMPTS" in globals(), (
    "STOP: FINAL_RAG_PROMPTS not found. "
    "Run Cell 36 first."
)

GEMINI_API_KEY = os.getenv("GEMINI_API_KEY")

assert GEMINI_API_KEY, (
    "STOP: GEMINI_API_KEY not found. "
    "Run Gemini API Key Setup first."
)


# ============================================================
# GEMINI CLIENT
# ============================================================

client = genai.Client(
    api_key=GEMINI_API_KEY
)


# ============================================================
# FINAL GEMINI CONFIGURATION
# ============================================================

LLM_PROVIDER = "GOOGLE_GEMINI"
LLM_MODEL = "gemini-3.1-flash-lite"

LLM_TEMPERATURE = 0.0
LLM_MAX_OUTPUT_TOKENS = 1200
LLM_TOP_P = 1.0

MAX_ANSWER_WORDS = 180

# More retries because Gemini may temporarily return 503.
MAX_RETRIES = 4

RETRY_DELAYS = [
    3,
    7,
    15,
    25
]


# ============================================================
# NORMALIZE PROMPT STRUCTURE
# ============================================================

def get_prompt_question(item):

    if isinstance(item, dict):

        return (
            item.get("question")
            or item.get("query")
            or item.get("Question")
            or ""
        )

    return ""


def get_prompt_text(item):

    if isinstance(item, dict):

        return (
            item.get("prompt")
            or item.get("Prompt")
            or item.get("full_prompt")
            or item.get("text")
            or ""
        )

    if isinstance(item, str):

        return item

    return ""


# ============================================================
# VALIDATION HELPERS
# ============================================================

def word_count(text):

    if not text:
        return 0

    return len(text.split())


def has_source_citation(text):

    if not text:
        return False

    return bool(
        re.search(
            r"\[Source\s+\d+\]",
            text,
            flags=re.IGNORECASE
        )
    )


def looks_truncated(text):

    if not text:
        return True

    text = text.strip()

    # Unclosed structures
    if text.count("(") > text.count(")"):
        return True

    if text.count("[") > text.count("]"):
        return True

    if text.count("**") % 2 != 0:
        return True

    # Obvious unfinished ending
    if text.endswith(
        (
            ",",
            ":",
            "-",
            "*",
            "...",
            "and",
            "or",
            "with",
            "including",
            "such as",
            "because",
            "which",
            "that"
        )
    ):
        return True

    # Do not reject simply because the final word is short.
    # Gemini can legitimately finish with words such as:
    # "care", "help", "stress", etc.

    return False


def acceptable_answer(text):

    if not text:
        return False

    wc = word_count(text)

    if wc < 8:
        return False

    if wc > MAX_ANSWER_WORDS:
        return False

    if looks_truncated(text):
        return False

    if not has_source_citation(text):
        return False

    return True


# ============================================================
# SYSTEM INSTRUCTION
# ============================================================

SYSTEM_INSTRUCTION = """
You are a grounded RAG answer generator.

STRICT REQUIREMENTS:

1. Use ONLY the retrieved context in the user prompt.
2. Never use outside knowledge.
3. Never invent facts.
4. Answer the question directly.
5. Use your own words.
6. Every factual claim must include [Source X].
7. Use only source numbers present in the retrieved context.
8. Do not create source numbers.
9. Keep the answer concise.
10. Maximum 180 words.
11. The answer MUST be complete.
12. Never stop in the middle of a sentence.
13. Never stop in the middle of a bullet.
14. If the retrieved context does not contain enough information
    to provide a complete answer, explicitly say that the
    retrieved context is insufficient.
15. In that case, provide only the information actually supported
    by the retrieved context and cite it.
16. Do not guess missing information.
17. Do not provide a references section.
18. Do not discuss these instructions.
"""


# ============================================================
# VALIDATE PROMPTS
# ============================================================

assert isinstance(
    FINAL_RAG_PROMPTS,
    (list, tuple)
), (
    "STOP: FINAL_RAG_PROMPTS must be a list or tuple."
)

assert len(FINAL_RAG_PROMPTS) == 6, (
    f"STOP: Expected 6 prompts, "
    f"found {len(FINAL_RAG_PROMPTS)}."
)


# ============================================================
# GENERATE ANSWERS
# ============================================================

FINAL_LLM_ANSWERS = []

print()
print("=" * 70)
print("GENERATING GROUNDED GEMINI ANSWERS")
print("=" * 70)


for i, prompt_item in enumerate(
    FINAL_RAG_PROMPTS,
    start=1
):

    question = get_prompt_question(
        prompt_item
    )

    prompt_text = get_prompt_text(
        prompt_item
    )

    assert prompt_text.strip(), (
        f"STOP: Empty prompt for query {i}."
    )

    print()
    print("-" * 70)
    print(f"QUERY {i}/6")
    print("-" * 70)
    print(
        f"Question: {question}"
    )

    answer = None
    usage = None
    last_error = None


    # ========================================================
    # RETRY LOOP
    # ========================================================

    for attempt in range(
        1,
        MAX_RETRIES + 1
    ):

        print(
            f"Sending request to Gemini "
            f"(attempt {attempt}/{MAX_RETRIES})..."
        )

        # ----------------------------------------------------
        # Base prompt
        # ----------------------------------------------------

        current_prompt = prompt_text


        # ----------------------------------------------------
        # Special handling for Q6
        # ----------------------------------------------------

        if i == 6:

            current_prompt = (
                prompt_text
                + "\n\n"
                + "SPECIAL REQUIREMENT FOR THIS QUESTION:\n"
                + "Do NOT invent a complete list of DWM modules "
                + "if the retrieved context does not provide one.\n"
                + "State clearly that the retrieved context does "
                + "not provide a complete list, then mention only "
                + "the modules or module information explicitly "
                + "supported by the retrieved context.\n"
                + "Keep the answer concise and complete.\n"
                + "Use [Source X] citations for every factual claim."
            )


        # ----------------------------------------------------
        # Retry instruction
        # ----------------------------------------------------

        if attempt > 1:

            current_prompt += (
                "\n\n"
                "RETRY REQUIREMENTS:\n"
                "- Return a complete answer.\n"
                "- Do not truncate.\n"
                "- Do not invent missing information.\n"
                "- Include [Source X] citations.\n"
                "- Keep the answer concise.\n"
                "- Finish every sentence and bullet.\n"
            )


        try:

            response = client.models.generate_content(

                model=LLM_MODEL,

                contents=current_prompt,

                config=types.GenerateContentConfig(

                    system_instruction=(
                        SYSTEM_INSTRUCTION
                    ),

                    temperature=LLM_TEMPERATURE,

                    max_output_tokens=LLM_MAX_OUTPUT_TOKENS,

                    top_p=LLM_TOP_P,

                    candidate_count=1
                )
            )


            # ------------------------------------------------
            # EXTRACT TEXT
            # ------------------------------------------------

            generated_text = getattr(
                response,
                "text",
                None
            )

            if generated_text is None:
                generated_text = ""

            generated_text = (
                generated_text.strip()
            )


            # ------------------------------------------------
            # LOCAL VALIDATION
            # ------------------------------------------------

            wc = word_count(
                generated_text
            )

            citation_ok = (
                has_source_citation(
                    generated_text
                )
            )

            truncated = (
                looks_truncated(
                    generated_text
                )
            )

            print(
                f"Word count: {wc}"
            )

            print(
                f"Citation present: "
                f"{citation_ok}"
            )

            print(
                f"Appears truncated: "
                f"{truncated}"
            )


            # ------------------------------------------------
            # ACCEPT
            # ------------------------------------------------

            if acceptable_answer(
                generated_text
            ):

                answer = generated_text

                usage = getattr(
                    response,
                    "usage_metadata",
                    None
                )

                print(
                    "Gemini response: SUCCESS"
                )

                break


            # ------------------------------------------------
            # REJECT
            # ------------------------------------------------

            print(
                "Gemini response rejected by "
                "local quality gate."
            )

            last_error = (
                "Response failed local validation."
            )


        except Exception as e:

            last_error = e

            error_text = str(e)

            print(
                "Gemini request failed."
            )

            print(
                f"Error: {error_text}"
            )


            # ------------------------------------------------
            # RETRY TEMPORARY 503 / 429
            # ------------------------------------------------

            temporary_error = (
                "503" in error_text
                or "UNAVAILABLE" in error_text
                or "429" in error_text
                or "RESOURCE_EXHAUSTED" in error_text
                or "temporarily" in error_text.lower()
            )


            if (
                temporary_error
                and attempt < MAX_RETRIES
            ):

                delay = RETRY_DELAYS[
                    attempt - 1
                ]

                print(
                    f"Temporary Gemini error. "
                    f"Waiting {delay} seconds "
                    f"before retry..."
                )

                time.sleep(
                    delay
                )

                continue


            if attempt < MAX_RETRIES:

                delay = RETRY_DELAYS[
                    attempt - 1
                ]

                print(
                    f"Retrying in {delay} seconds..."
                )

                time.sleep(
                    delay
                )


    # ========================================================
    # FINAL QUERY CHECK
    # ========================================================

    assert answer is not None, (
        f"STOP: Gemini could not produce an "
        f"acceptable grounded answer for query {i}.\n"
        f"Last error: {last_error}"
    )


    # ========================================================
    # TOKEN USAGE
    # ========================================================

    prompt_tokens = None
    completion_tokens = None
    total_tokens = None

    if usage is not None:

        prompt_tokens = getattr(
            usage,
            "prompt_token_count",
            None
        )

        completion_tokens = getattr(
            usage,
            "candidates_token_count",
            None
        )

        total_tokens = getattr(
            usage,
            "total_token_count",
            None
        )


    # ========================================================
    # SAVE RESULT
    # ========================================================

    result = {

        "query_number":
            i,

        "question":
            question,

        "answer":
            answer,

        "provider":
            LLM_PROVIDER,

        "model":
            LLM_MODEL,

        "temperature":
            LLM_TEMPERATURE,

        "max_output_tokens":
            LLM_MAX_OUTPUT_TOKENS,

        "grounded_prompt":
            True,

        "external_knowledge_allowed":
            False,

        "unsupported_claims_allowed":
            False,

        "citation_required":
            True,

        "citation_present":
            has_source_citation(
                answer
            ),

        "appears_truncated":
            looks_truncated(
                answer
            ),

        "word_count":
            word_count(
                answer
            ),

        "prompt_tokens":
            prompt_tokens,

        "completion_tokens":
            completion_tokens,

        "total_tokens":
            total_tokens
    }


    FINAL_LLM_ANSWERS.append(
        result
    )


    # ========================================================
    # DISPLAY
    # ========================================================

    print()
    print("ANSWER:")
    print(answer)

    print()
    print(
        f"Word count: "
        f"{word_count(answer)}"
    )

    print(
        f"Citation present: "
        f"{has_source_citation(answer)}"
    )

    print(
        f"Appears truncated: "
        f"{looks_truncated(answer)}"
    )

    if total_tokens is not None:

        print(
            f"Tokens used: "
            f"{total_tokens}"
        )


# ============================================================
# GLOBAL VALIDATION
# ============================================================

assert len(
    FINAL_LLM_ANSWERS
) == 6, (
    "STOP: Expected exactly 6 Gemini answers."
)


for result in FINAL_LLM_ANSWERS:

    assert result["answer"], (
        f"STOP: Empty answer for query "
        f"{result['query_number']}."
    )

    assert (
        result["grounded_prompt"]
        is True
    )

    assert (
        result["external_knowledge_allowed"]
        is False
    )

    assert (
        result["unsupported_claims_allowed"]
        is False
    )

    assert (
        result["citation_required"]
        is True
    )

    assert (
        result["citation_present"]
        is True
    )

    assert (
        result["appears_truncated"]
        is False
    )


# ============================================================
# FINAL OUTPUT
# ============================================================

FINAL_LLM_OUTPUT = {

    "project_status":
        "FINALIZED",

    "provider":
        LLM_PROVIDER,

    "generation_status":
        "COMPLETED",

    "model":
        LLM_MODEL,

    "temperature":
        LLM_TEMPERATURE,

    "max_output_tokens":
        LLM_MAX_OUTPUT_TOKENS,

    "top_p":
        LLM_TOP_P,

    "external_knowledge_allowed":
        False,

    "unsupported_claims_allowed":
        False,

    "citation_required":
        True,

    "total_questions":
        len(FINAL_LLM_ANSWERS),

    "generated_at":
        datetime.now().isoformat(),

    "answers":
        FINAL_LLM_ANSWERS
}


# ============================================================
# SAVE JSON
# ============================================================

llm_output_filename = (
    "final_gemini_answers.json"
)

with open(
    llm_output_filename,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        FINAL_LLM_OUTPUT,
        f,
        indent=2,
        ensure_ascii=False
    )


# ============================================================
# FINAL REPORT
# ============================================================

print()
print("=" * 70)
print("FINAL GEMINI GENERATION SUMMARY")
print("=" * 70)

print(
    f"Provider: {LLM_PROVIDER}"
)

print(
    f"Model: {LLM_MODEL}"
)

print(
    f"Temperature: {LLM_TEMPERATURE}"
)

print(
    f"Max output tokens: "
    f"{LLM_MAX_OUTPUT_TOKENS}"
)

print(
    f"Questions: "
    f"{len(FINAL_LLM_ANSWERS)}"
)

print(
    f"Generated answers: "
    f"{len(FINAL_LLM_ANSWERS)}"
)

print(
    "External knowledge allowed: False"
)

print(
    "Unsupported claims allowed: False"
)

print(
    "Citation required: True"
)

print(
    "All 6 answers passed generation validation: True"
)

print()
print(
    f"Output saved to: "
    f"{llm_output_filename}"
)

print()
print("=" * 70)
print(
    "CELL 37 — ACTUAL GEMINI GROUNDED "
    "ANSWER GENERATION: PASS"
)
print("=" * 70)

print(
    "Actual Gemini generation completed successfully."
)

print(
    "FINAL_LLM_ANSWERS is ready for Cell 38."
)

print(
    "Run Cell 38 again."
)

print("=" * 70)

````

````python
# ============================================================
# CELL 38 — FINAL GEMINI RAG ANSWER QUALITY EVALUATION
# ============================================================

import re
import json
from datetime import datetime

print("=" * 70)
print("CELL 38 — FINAL GEMINI RAG ANSWER QUALITY EVALUATION")
print("=" * 70)


# ============================================================
# SAFETY CHECKS
# ============================================================

assert "FINAL_LLM_ANSWERS" in globals(), (
    "STOP: FINAL_LLM_ANSWERS not found. "
    "Run Cell 37 first."
)

assert isinstance(
    FINAL_LLM_ANSWERS,
    list
), (
    "STOP: FINAL_LLM_ANSWERS must be a list."
)

assert len(
    FINAL_LLM_ANSWERS
) == 6, (
    f"STOP: Expected 6 LLM answers, "
    f"found {len(FINAL_LLM_ANSWERS)}."
)


# ============================================================
# EXPECTED QUESTIONS
# ============================================================

EXPECTED_QUESTIONS = [
    "What is self-help?",
    "Who is self-help intended for?",
    "When should guided self-help be used?",
    "What should a helper do when someone is at risk of suicide?",
    "What are the Step-by-Step skills for managing stress?",
    "What are the Doing What Matters modules?",
]


# ============================================================
# HELPER FUNCTIONS
# ============================================================

def normalize_text(text):

    if text is None:
        return ""

    return str(text).strip()


def extract_citations(answer):

    matches = re.findall(
        r"\[Source\s+(\d+)\]",
        answer,
        flags=re.IGNORECASE
    )

    return sorted(
        set(
            int(x)
            for x in matches
        )
    )


def has_incomplete_markers(answer):

    text = normalize_text(answer)

    if not text:
        return True

    incomplete_patterns = [

        r"receiving\s*$",

        r"including:\s*$",

        r"following actions:\s*$",

        r"skills\s*$",

        r"modules\s*$",

        r"\*\s*$",

        r"\*\*\s*$",

        r":\s*$",

        r"\bthe\s*$",

        r"\band\s*$",

        r"\bor\s*$",

        r"\bwhich\s*$",

        r"\bthat\s*$",
    ]

    for pattern in incomplete_patterns:

        if re.search(
            pattern,
            text,
            flags=re.IGNORECASE
        ):

            return True

    return False


def answer_word_count(answer):

    return len(
        normalize_text(answer).split()
    )


def source_validity(citations):

    if not citations:
        return False

    return all(
        1 <= source <= 5
        for source in citations
    )


def required_concepts_for_query(
    query_number
):

    concepts = {

        1: [
            "self-help",
            "information",
        ],

        2: [
            "self-help",
            "people",
        ],

        3: [
            "guided",
            "self-help",
        ],

        4: [
            "suicide",
            "supervisor",
        ],

        5: [
            "Step-by-Step",
            "stress",
        ],

        6: [
            "Doing What Matters",
            "module",
        ],
    }

    return concepts.get(
        query_number,
        []
    )


def concept_coverage(
    answer,
    query_number
):

    answer_lower = normalize_text(
        answer
    ).lower()

    concepts = (
        required_concepts_for_query(
            query_number
        )
    )

    if not concepts:
        return 0.0

    found = 0

    for concept in concepts:

        if concept.lower() in answer_lower:

            found += 1

    return (
        found / len(concepts)
    )


def directness_score(
    answer,
    query_number
):

    text = normalize_text(
        answer
    ).lower()

    if not text:
        return 0.0

    score = 1.0

    question = (
        EXPECTED_QUESTIONS[
            query_number - 1
        ]
        .lower()
    )

    # --------------------------------------------------------
    # Penalize answers that only say context is insufficient
    # for questions where the context clearly contains data.
    # --------------------------------------------------------

    insufficient_phrases = [

        "context is insufficient",

        "does not provide enough information",

        "cannot determine",

        "not enough information",
    ]

    if any(
        phrase in text
        for phrase in insufficient_phrases
    ):

        score -= 0.40


    # --------------------------------------------------------
    # Penalize extremely short answers.
    # --------------------------------------------------------

    words = answer_word_count(
        answer
    )

    if words < 20:

        score -= 0.40

    elif words < 35:

        score -= 0.20


    # --------------------------------------------------------
    # Query-specific directness checks
    # --------------------------------------------------------

    if query_number == 1:

        if "self-help" not in text:

            score -= 0.30


    elif query_number == 2:

        if (
            "self-help" not in text
            and "people" not in text
        ):

            score -= 0.30


    elif query_number == 3:

        if "guided" not in text:

            score -= 0.30


    elif query_number == 4:

        if "suicide" not in text:

            score -= 0.30


    elif query_number == 5:

        if (
            "step-by-step" not in text
            and "sbs" not in text
        ):

            score -= 0.30


    elif query_number == 6:

        if (
            "doing what matters" not in text
            and "dwm" not in text
        ):

            score -= 0.30


    return max(
        0.0,
        min(1.0, score)
    )


# ============================================================
# EVALUATE ANSWERS
# ============================================================

FINAL_GEMINI_QUALITY_RESULTS = []

print()
print("=" * 70)
print("FINAL GEMINI ANSWER QUALITY — QUERY DETAILS")
print("=" * 70)


for i, result in enumerate(
    FINAL_LLM_ANSWERS,
    start=1
):

    answer = normalize_text(
        result.get(
            "answer",
            ""
        )
    )

    question = normalize_text(
        result.get(
            "question",
            ""
        )
    )

    citations = extract_citations(
        answer
    )

    words = answer_word_count(
        answer
    )

    # --------------------------------------------------------
    # Basic checks
    # --------------------------------------------------------

    answer_present = bool(
        answer
    )

    citation_present = (
        len(citations) > 0
    )

    citations_valid = (
        source_validity(
            citations
        )
    )

    incomplete = (
        has_incomplete_markers(
            answer
        )
    )

    complete = (
        answer_present
        and not incomplete
    )

    # --------------------------------------------------------
    # Concept coverage
    # --------------------------------------------------------

    concepts = concept_coverage(
        answer,
        i
    )

    # --------------------------------------------------------
    # Directness
    # --------------------------------------------------------

    directness = directness_score(
        answer,
        i
    )

    # --------------------------------------------------------
    # Citation score
    # --------------------------------------------------------

    if citation_present and citations_valid:

        citation_score = 1.0

    elif citation_present:

        citation_score = 0.5

    else:

        citation_score = 0.0


    # --------------------------------------------------------
    # Grounding status
    # --------------------------------------------------------

    grounded = (
        answer_present
        and citation_present
        and citations_valid
        and complete
    )


    # --------------------------------------------------------
    # Final quality score
    #
    # Weighted:
    # 30% directness
    # 30% concept coverage
    # 25% citation validity
    # 15% completeness
    # --------------------------------------------------------

    quality_score = (

        0.30 * directness

        +

        0.30 * concepts

        +

        0.25 * citation_score

        +

        0.15 * (
            1.0
            if complete
            else 0.0
        )
    )


    quality_score = round(
        quality_score,
        4
    )


    # --------------------------------------------------------
    # PASS / FAIL
    #
    # Deliberately NOT using 100% as default.
    # --------------------------------------------------------

    quality_pass = (

        quality_score >= 0.70

        and answer_present

        and citation_present

        and citations_valid

        and complete
    )


    quality_status = (
        "PASS"
        if quality_pass
        else "FAIL"
    )


    query_result = {

        "query_number":
            i,

        "question":
            question,

        "word_count":
            words,

        "answer_present":
            answer_present,

        "complete":
            complete,

        "incomplete":
            incomplete,

        "grounded":
            grounded,

        "cited_sources":
            citations,

        "citation_present":
            citation_present,

        "citation_validity":
            citations_valid,

        "concept_coverage":
            round(
                concepts,
                4
            ),

        "directness_score":
            round(
                directness,
                4
            ),

        "citation_score":
            round(
                citation_score,
                4
            ),

        "quality_score":
            quality_score,

        "quality_status":
            quality_status,
    }


    FINAL_GEMINI_QUALITY_RESULTS.append(
        query_result
    )


    # --------------------------------------------------------
    # DISPLAY
    # --------------------------------------------------------

    print()
    print(
        f"QUERY {i}/6"
    )

    print(
        f"Question: {question}"
    )

    print(
        f"Word count: {words}"
    )

    print(
        f"Answer present: "
        f"{answer_present}"
    )

    print(
        f"Complete: "
        f"{complete}"
    )

    print(
        f"Grounded: "
        f"{grounded}"
    )

    print(
        f"Cited sources: "
        f"{citations}"
    )

    print(
        f"Citation validity: "
        f"{citations_valid}"
    )

    print(
        f"Concept coverage: "
        f"{concepts:.4f}"
    )

    print(
        f"Directness score: "
        f"{directness:.4f}"
    )

    print(
        f"Quality score: "
        f"{quality_score:.4f}"
    )

    print(
        f"Quality status: "
        f"{quality_status}"
    )


# ============================================================
# FINAL SUMMARY
# ============================================================

total_answers = len(
    FINAL_GEMINI_QUALITY_RESULTS
)

passed_answers = sum(
    1
    for r in FINAL_GEMINI_QUALITY_RESULTS
    if r["quality_status"] == "PASS"
)

failed_answers = (
    total_answers
    - passed_answers
)

grounded_answers = sum(
    1
    for r in FINAL_GEMINI_QUALITY_RESULTS
    if r["grounded"]
)

answers_with_citations = sum(
    1
    for r in FINAL_GEMINI_QUALITY_RESULTS
    if r["citation_present"]
)

citation_valid_answers = sum(
    1
    for r in FINAL_GEMINI_QUALITY_RESULTS
    if r["citation_validity"]
)

complete_answers = sum(
    1
    for r in FINAL_GEMINI_QUALITY_RESULTS
    if r["complete"]
)


answer_pass_rate = (
    passed_answers
    / total_answers
    if total_answers
    else 0.0
)

grounded_rate = (
    grounded_answers
    / total_answers
    if total_answers
    else 0.0
)

citation_presence_rate = (
    answers_with_citations
    / total_answers
    if total_answers
    else 0.0
)

citation_validity_rate = (
    citation_valid_answers
    / total_answers
    if total_answers
    else 0.0
)

completion_rate = (
    complete_answers
    / total_answers
    if total_answers
    else 0.0
)

average_concept_coverage = (

    sum(
        r["concept_coverage"]
        for r in FINAL_GEMINI_QUALITY_RESULTS
    )

    / total_answers
)

average_directness = (

    sum(
        r["directness_score"]
        for r in FINAL_GEMINI_QUALITY_RESULTS
    )

    / total_answers
)

average_quality_score = (

    sum(
        r["quality_score"]
        for r in FINAL_GEMINI_QUALITY_RESULTS
    )

    / total_answers
)


# ============================================================
# OVERALL DECISION
# ============================================================

overall_status = (
    "PASS"
    if (
        failed_answers == 0
        and grounded_rate >= 0.80
        and citation_validity_rate >= 0.80
        and completion_rate >= 0.80
        and average_quality_score >= 0.70
    )
    else "REVIEW"
)


# ============================================================
# FINAL QUALITY OBJECT
# ============================================================

FINAL_GEMINI_QUALITY = {

    "project_status":
        "FINALIZED",

    "evaluation_status":
        "COMPLETED",

    "provider":
        "GOOGLE_GEMINI",

    "model":
        FINAL_LLM_ANSWERS[0].get(
            "model",
            "unknown"
        ),

    "total_answers":
        total_answers,

    "passed_answers":
        passed_answers,

    "failed_answers":
        failed_answers,

    "answer_pass_rate":
        round(
            answer_pass_rate,
            4
        ),

    "grounded_answers":
        grounded_answers,

    "grounded_rate":
        round(
            grounded_rate,
            4
        ),

    "answers_with_citations":
        answers_with_citations,

    "citation_presence_rate":
        round(
            citation_presence_rate,
            4
        ),

    "citation_valid_answers":
        citation_valid_answers,

    "citation_validity_rate":
        round(
            citation_validity_rate,
            4
        ),

    "complete_answers":
        complete_answers,

    "completion_rate":
        round(
            completion_rate,
            4
        ),

    "average_concept_coverage":
        round(
            average_concept_coverage,
            4
        ),

    "average_directness":
        round(
            average_directness,
            4
        ),

    "average_quality_score":
        round(
            average_quality_score,
            4
        ),

    "overall_status":
        overall_status,

    "evaluated_at":
        datetime.now().isoformat(),

    "query_details":
        FINAL_GEMINI_QUALITY_RESULTS,
}


# ============================================================
# SAVE QUALITY REPORT
# ============================================================

quality_filename = (
    "final_gemini_quality_evaluation.json"
)

with open(
    quality_filename,
    "w",
    encoding="utf-8",
) as f:

    json.dump(
        FINAL_GEMINI_QUALITY,
        f,
        indent=2,
        ensure_ascii=False,
    )


# ============================================================
# FINAL REPORT
# ============================================================

print()
print("=" * 70)
print("FINAL GEMINI RAG ANSWER QUALITY SUMMARY")
print("=" * 70)

print(
    f"Total answers: "
    f"{total_answers}"
)

print(
    f"Passed answers: "
    f"{passed_answers}"
)

print(
    f"Failed answers: "
    f"{failed_answers}"
)

print(
    f"Answer pass rate: "
    f"{answer_pass_rate:.4f}"
)

print(
    f"Grounded answers: "
    f"{grounded_answers}"
)

print(
    f"Grounded rate: "
    f"{grounded_rate:.4f}"
)

print(
    f"Answers with citations: "
    f"{answers_with_citations}"
)

print(
    f"Citation presence rate: "
    f"{citation_presence_rate:.4f}"
)

print(
    f"Citation-valid answers: "
    f"{citation_valid_answers}"
)

print(
    f"Citation validity rate: "
    f"{citation_validity_rate:.4f}"
)

print(
    f"Complete answers: "
    f"{complete_answers}"
)

print(
    f"Completion rate: "
    f"{completion_rate:.4f}"
)

print(
    f"Average concept coverage: "
    f"{average_concept_coverage:.4f}"
)

print(
    f"Average directness: "
    f"{average_directness:.4f}"
)

print(
    f"Average quality score: "
    f"{average_quality_score:.4f}"
)


print()
print("=" * 70)
print("FINAL GEMINI RAG QUALITY DECISION")
print("=" * 70)

print(
    f"Overall status: "
    f"{overall_status}"
)

if overall_status == "PASS":

    print(
        "All Gemini answers passed the "
        "automated quality thresholds."
    )

else:

    print(
        "Some Gemini answers require review."
    )

    print(
        "Do NOT treat the pipeline as final "
        "until the failed answers are reviewed."
    )


print()
print(
    f"Quality report saved to: "
    f"{quality_filename}"
)

print()
print("=" * 70)
print(
    "CELL 38 — FINAL GEMINI RAG ANSWER "
    "QUALITY EVALUATION: COMPLETED"
)
print("=" * 70)

print(
    "FINAL_GEMINI_QUALITY is ready."
)

print(
    "Proceed to Cell 39 only after "
    "reviewing the quality results."
)

print("=" * 70)

````

````python
# ============================================================
# CELL 42 — FINAL FAITHFULNESS EVALUATION
# ============================================================

import os
import json
import re
from datetime import datetime
from google import genai


print("=" * 70)
print("CELL 42 — FINAL FAITHFULNESS EVALUATION")
print("=" * 70)


# ============================================================
# SAFETY CHECKS
# ============================================================

assert "FINAL_LLM_ANSWERS" in globals(), (
    "STOP: FINAL_LLM_ANSWERS not found. "
    "Run Cell 37 first."
)

assert "FINAL_RAG_PROMPTS" in globals(), (
    "STOP: FINAL_RAG_PROMPTS not found. "
    "Run Cell 36 first."
)

assert len(FINAL_LLM_ANSWERS) == 6, (
    f"STOP: Expected 6 final answers, "
    f"found {len(FINAL_LLM_ANSWERS)}."
)

assert len(FINAL_RAG_PROMPTS) == 6, (
    f"STOP: Expected 6 RAG prompts, "
    f"found {len(FINAL_RAG_PROMPTS)}."
)


# ============================================================
# GEMINI API SETUP
# ============================================================

GEMINI_API_KEY = os.getenv("GEMINI_API_KEY")

assert GEMINI_API_KEY, (
    "STOP: GEMINI_API_KEY not found. "
    "Run Cell 37A first."
)

client = genai.Client(
    api_key=GEMINI_API_KEY.strip()
)


FAITHFULNESS_MODEL = "gemini-3.1-flash-lite"


# ============================================================
# TEST GEMINI CONNECTION
# ============================================================

print()
print("=" * 70)
print("TESTING GEMINI API CONNECTION")
print("=" * 70)

try:

    test_response = client.models.generate_content(
        model=FAITHFULNESS_MODEL,
        contents=(
            "Reply with exactly one word: "
            "CONNECTION_OK"
        )
    )

    test_text = (
        test_response.text.strip()
        if test_response.text
        else ""
    )

    assert test_text, (
        "Gemini returned an empty connection response."
    )

    print("Gemini API connection: PASS")
    print(
        f"Model: {FAITHFULNESS_MODEL}"
    )

except Exception as e:

    raise RuntimeError(
        "STOP: Gemini API connection failed: "
        f"{e}"
    ) from e


# ============================================================
# HELPERS
# ============================================================

def get_question(item):

    if isinstance(item, dict):

        return (
            item.get("question")
            or item.get("query")
            or item.get("Question")
            or ""
        )

    return ""


def get_prompt_text(item):

    if isinstance(item, dict):

        return (
            item.get("prompt")
            or item.get("Prompt")
            or item.get("full_prompt")
            or item.get("text")
            or ""
        )

    if isinstance(item, str):

        return item

    return ""


def get_answer(result):

    if isinstance(result, dict):

        return (
            result.get("answer")
            or result.get("response")
            or result.get("text")
            or ""
        )

    return ""


def extract_json(text):

    """
    Extract the first JSON object from Gemini output.
    Handles accidental markdown fences.
    """

    if not text:
        return None

    text = text.strip()

    # Remove markdown fences
    text = re.sub(
        r"^```(?:json)?\s*",
        "",
        text,
        flags=re.IGNORECASE
    )

    text = re.sub(
        r"\s*```$",
        "",
        text
    )

    # Direct JSON
    try:

        return json.loads(text)

    except Exception:
        pass

    # Search for JSON object
    match = re.search(
        r"\{.*\}",
        text,
        flags=re.DOTALL
    )

    if not match:
        return None

    try:

        return json.loads(
            match.group(0)
        )

    except Exception:

        return None


def normalize_claim_result(data):

    """
    Normalize Gemini judge output.
    """

    if not isinstance(data, dict):

        return None

    claims = data.get("claims")

    if not isinstance(claims, list):

        return None

    normalized = []

    for claim in claims:

        if not isinstance(claim, dict):
            continue

        claim_text = str(
            claim.get("claim", "")
        ).strip()

        supported = claim.get(
            "supported"
        )

        evidence = str(
            claim.get("evidence", "")
        ).strip()

        if not claim_text:
            continue

        if isinstance(supported, str):

            supported_lower = (
                supported.strip().lower()
            )

            if supported_lower == "true":
                supported = True

            elif supported_lower == "false":
                supported = False

            else:
                continue

        if not isinstance(
            supported,
            bool
        ):
            continue

        normalized.append(
            {
                "claim": claim_text,
                "supported": supported,
                "evidence": evidence
            }
        )

    if not normalized:
        return None

    return normalized


# ============================================================
# FAITHFULNESS STORAGE
# ============================================================

FINAL_FAITHFULNESS_RESULTS = []


print()
print("=" * 70)
print("CLAIM-LEVEL FAITHFULNESS EVALUATION")
print("=" * 70)


# ============================================================
# EVALUATE EACH ANSWER
# ============================================================

for i in range(6):

    question = get_question(
        FINAL_RAG_PROMPTS[i]
    )

    evidence_prompt = get_prompt_text(
        FINAL_RAG_PROMPTS[i]
    )

    answer = get_answer(
        FINAL_LLM_ANSWERS[i]
    )

    assert evidence_prompt.strip(), (
        f"STOP: Empty retrieved-context prompt "
        f"for query {i + 1}."
    )

    assert answer.strip(), (
        f"STOP: Empty answer for query {i + 1}."
    )


    print()
    print("-" * 70)
    print(f"QUERY {i + 1}/6")
    print("-" * 70)

    print(
        f"Question: {question}"
    )

    print(
        "Evaluating claims against retrieved evidence..."
    )


    # ========================================================
    # CLAIM-LEVEL JUDGE PROMPT
    # ========================================================

    judge_prompt = f"""
You are a strict RAG faithfulness evaluator.

Your task is to determine whether each factual claim
in the ANSWER is directly supported by the RETRIEVED
EVIDENCE.

IMPORTANT RULES:

1. Use ONLY the RETRIEVED EVIDENCE.
2. Do NOT use outside knowledge.
3. Do NOT assume a claim is supported merely because
   it sounds plausible.
4. A claim is supported only when the retrieved evidence
   explicitly supports the meaning of the claim.
5. A citation such as [Source 1] is NOT by itself proof
   of faithfulness.
6. Evaluate the actual semantic support between the claim
   and the evidence.
7. Split the answer into individual factual claims.
8. Do not invent claims that are not present in the answer.
9. If a claim is partially supported but contains an
   unsupported factual addition, mark it false.
10. If the answer explicitly says that the retrieved
    evidence is insufficient, that statement itself can
    be supported if the evidence actually shows a gap.

QUESTION:
{question}

RETRIEVED EVIDENCE:
--------------------
{evidence_prompt}
--------------------

ANSWER:
--------------------
{answer}
--------------------

Return ONLY valid JSON.

Required format:

{{
  "claims": [
    {{
      "claim": "exact or concise factual claim from the answer",
      "supported": true,
      "evidence": "brief explanation of exactly where the retrieved evidence supports it"
    }}
  ]
}}

If there are no factual claims, return:

{{
  "claims": []
}}
"""


    # ========================================================
    # GEMINI JUDGE REQUEST
    # ========================================================

    try:

        response = client.models.generate_content(

            model=FAITHFULNESS_MODEL,

            contents=judge_prompt
        )

        raw_judge_output = (
            response.text.strip()
            if response.text
            else ""
        )

        parsed = extract_json(
            raw_judge_output
        )

        claims = normalize_claim_result(
            parsed
        )

        assert claims is not None, (
            "Gemini returned an invalid claim-level "
            "faithfulness structure."
        )


    except Exception as e:

        print()
        print(
            "Faithfulness evaluation failed "
            f"for query {i + 1}."
        )

        raise RuntimeError(
            f"STOP: Gemini faithfulness evaluation "
            f"failed for query {i + 1}: {e}"
        ) from e


    # ========================================================
    # CLAIM COUNTS
    # ========================================================

    total_claims = len(
        claims
    )

    supported_claims = sum(
        1
        for claim in claims
        if claim["supported"] is True
    )

    unsupported_claims = (
        total_claims
        - supported_claims
    )


    # ========================================================
    # FAITHFULNESS SCORE
    # ========================================================

    if total_claims == 0:

        # No factual claims means there is nothing to
        # evaluate. Treat as neutral rather than failing
        # because of a division-by-zero artifact.
        faithfulness_score = 1.0

    else:

        faithfulness_score = (
            supported_claims
            / total_claims
        )


    faithful = (
        faithfulness_score >= 0.90
    )


    # ========================================================
    # CITATIONS
    # ========================================================

    citations = sorted(
        {
            int(x)
            for x in re.findall(
                r"\[Source\s+(\d+)\]",
                answer,
                flags=re.IGNORECASE
            )
        }
    )


    # ========================================================
    # SAVE QUERY RESULT
    # ========================================================

    result = {

        "query_number":
            i + 1,

        "question":
            question,

        "answer":
            answer,

        "total_claims":
            total_claims,

        "supported_claims":
            supported_claims,

        "unsupported_claims":
            unsupported_claims,

        "faithfulness_score":
            round(
                faithfulness_score,
                4
            ),

        "faithful":
            faithful,

        "citations_found":
            citations,

        "claims":
            claims,

        "model":
            FAITHFULNESS_MODEL,

        "evaluation_method":
            "claim_level_semantic_evidence_check",

        "external_knowledge_allowed":
            False
    }


    FINAL_FAITHFULNESS_RESULTS.append(
        result
    )


    # ========================================================
    # DISPLAY
    # ========================================================

    print()
    print(
        f"Total claims: {total_claims}"
    )

    print(
        f"Supported claims: "
        f"{supported_claims}"
    )

    print(
        f"Unsupported claims: "
        f"{unsupported_claims}"
    )

    print(
        f"Faithfulness score: "
        f"{faithfulness_score:.4f}"
    )

    print(
        f"Faithful: {faithful}"
    )

    print(
        f"Citations found: {citations}"
    )

    print(
        "Faithfulness evaluation: "
        "PASS"
    )


# ============================================================
# FINAL SUMMARY
# ============================================================

total_questions = len(
    FINAL_FAITHFULNESS_RESULTS
)

passed_faithfulness = sum(
    1
    for result in FINAL_FAITHFULNESS_RESULTS
    if result["faithful"]
)

failed_faithfulness = (
    total_questions
    - passed_faithfulness
)

faithfulness_pass_rate = (
    passed_faithfulness
    / total_questions
    if total_questions
    else 0.0
)

average_faithfulness = (
    sum(
        result["faithfulness_score"]
        for result
        in FINAL_FAITHFULNESS_RESULTS
    )
    / total_questions
    if total_questions
    else 0.0
)


target = 0.90

target_met = (
    average_faithfulness >= target
)


# ============================================================
# FINAL REPORT OBJECT
# ============================================================

FINAL_FAITHFULNESS = {

    "project_status":
        "FINALIZED",

    "evaluation_status":
        "COMPLETED",

    "provider":
        "GOOGLE_GEMINI",

    "model":
        FAITHFULNESS_MODEL,

    "total_questions":
        total_questions,

    "passed_faithfulness":
        passed_faithfulness,

    "failed_faithfulness":
        failed_faithfulness,

    "faithfulness_pass_rate":
        round(
            faithfulness_pass_rate,
            4
        ),

    "average_faithfulness":
        round(
            average_faithfulness,
            4
        ),

    "target":
        target,

    "target_met":
        target_met,

    "external_knowledge_allowed":
        False,

    "results":
        FINAL_FAITHFULNESS_RESULTS,

    "generated_at":
        datetime.now().isoformat()
}


# ============================================================
# SAVE REPORT
# ============================================================

faithfulness_filename = (
    "FINAL_FAITHFULNESS_REPORT.json"
)

with open(
    faithfulness_filename,
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        FINAL_FAITHFULNESS,
        f,
        indent=2,
        ensure_ascii=False
    )


# ============================================================
# FINAL REPORT
# ============================================================

print()
print("=" * 70)
print("FINAL FAITHFULNESS SUMMARY")
print("=" * 70)

print(
    f"Total questions: "
    f"{total_questions}"
)

print(
    f"Passed faithfulness: "
    f"{passed_faithfulness}"
)

print(
    f"Failed faithfulness: "
    f"{failed_faithfulness}"
)

print(
    f"Faithfulness pass rate: "
    f"{faithfulness_pass_rate:.4f}"
)

print(
    f"Average faithfulness: "
    f"{average_faithfulness:.4f}"
)

print(
    f"Target: {target:.4f}"
)

print(
    f"Target met: {target_met}"
)

print()
print(
    f"Faithfulness report saved to: "
    f"{faithfulness_filename}"
)


# ============================================================
# FINAL DECISION
# ============================================================

print()
print("=" * 70)
print("CELL 42 — FINAL FAITHFULNESS DECISION")
print("=" * 70)

if target_met:

    print(
        "Overall status: PASS"
    )

    print(
        "The final answers meet the "
        "0.90 faithfulness target."
    )

else:

    print(
        "Overall status: REVIEW"
    )

    print(
        "Some answers remain below the "
        "0.90 faithfulness target."
    )

print()
print(
    "Claim-level faithfulness evaluation completed."
)

print(
    "FINAL_FAITHFULNESS is ready."
)

print("=" * 70)
````
